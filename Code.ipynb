```python
# ============================================================
# MATH-619 Capstone Research Project
# Author  : Abdullah Hussain Alharbi (g202415480)
# Institution : King Fahd University of Petroleum and Minerals
# Degree  : Master in Data Science & Analytics
# Course  : MATH-619
# Advisor : Dr. Mousa Ahmad Al-Bashrawi
# Phase   : CRISP-DM — Data Understanding
# ============================================================

import os
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from matplotlib.ticker import MaxNLocator
import seaborn as sns

DARK    = '#0d1117'
GRID    = '#21262d'
BORDER  = '#30363d'
TEXT    = '#c9d1d9'
MUTED   = '#8b949e'
ACCENT  = '#58a6ff'
GREEN   = '#3fb950'
RED     = '#f85149'
YELLOW  = '#e3b341'
PURPLE  = '#bc8cff'
ORANGE  = '#f0883e'
TEAL    = '#39d353'
COLORS  = [ACCENT, GREEN, RED, YELLOW, PURPLE]

plt.rcParams.update({
    'figure.facecolor':  DARK,
    'axes.facecolor':    DARK,
    'axes.edgecolor':    BORDER,
    'axes.labelcolor':   TEXT,
    'xtick.color':       MUTED,
    'ytick.color':       MUTED,
    'text.color':        TEXT,
    'grid.color':        GRID,
    'grid.linestyle':    '--',
    'grid.alpha':        0.6,
    'font.family':       'monospace',
    'legend.facecolor':  '#161b22',
    'legend.edgecolor':  BORDER,
    'figure.dpi':        120,
})

print("Imports OK.")

# ── File configuration ───────────────────────────────────────
DATA_PATH = '/content/'
TICKERS   = ['AAPL', 'AMZN', 'GOOG', 'MSFT', 'TSLA']


def find_file(data_path, ticker, kind):
    candidates = [
        f'{ticker}({kind}).xlsx',
        f'{ticker}_{kind}_.xlsx',
        f'{ticker}_{kind}.xlsx',
        f'{ticker.lower()}({kind.lower()}).xlsx',
        f'{ticker.lower()}_{kind.lower()}_.xlsx',
        f'{ticker}({kind}).xlsx'.replace('xlsx', 'XLSX'),
    ]
    for name in candidates:
        full = os.path.join(data_path, name)
        if os.path.exists(full):
            return full
    try:
        all_files = os.listdir(data_path)
        for f in all_files:
            if (ticker.upper() in f.upper() and
                    kind.upper() in f.upper() and
                    f.endswith('.xlsx')):
                return os.path.join(data_path, f)
    except Exception:
        pass
    raise FileNotFoundError(
        f"Could not find {ticker} {kind} file in {data_path}.\n"
        f"Tried: {candidates}\n"
        f"Files present: {os.listdir(data_path)}"
    )


RENAME_MAP = {
    'time': 'Date', 'Time': 'Date',
    'open': 'Open',  'high': 'High',
    'low':  'Low',   'close': 'Close',
    'Volume': 'Volume', 'volume': 'Volume',
    'EMA_20': 'EMA_20', 'EMA_50': 'EMA_50',
    'RSI_14': 'RSI',    'RSI':    'RSI',
    'RSI-based_MA': 'RSI_MA', 'RSI-based MA': 'RSI_MA',
    'Histogram':  'Histogram',
    'BB_Basis':   'BB_Basis',
    'BB_Upper':   'BB_Upper',
    'BB_Lower':   'BB_Lower',
    'MACD':       'MACD',
    'Signal':     'Signal',
    'VIX_Close':  'VIX',  'VIX': 'VIX',
    'Sentiment_Score': 'Sentiment',
}

FEAT_TECH = ['Open','High','Low','Close','Volume',
             'EMA_20','EMA_50','RSI','RSI_MA',
             'Histogram','BB_Basis','BB_Upper','BB_Lower',
             'MACD','Signal','VIX']
FEAT_SENT = FEAT_TECH + ['Sentiment']


def load_file(path):
    df = pd.read_excel(path)
    df = df.rename(columns=RENAME_MAP)
    if 'Date' not in df.columns:
        date_candidates = [c for c in df.columns
                           if 'date' in c.lower() or 'time' in c.lower()]
        if date_candidates:
            df = df.rename(columns={date_candidates[0]: 'Date'})
    df['Date'] = pd.to_datetime(df['Date'])
    df = df.sort_values('Date').reset_index(drop=True)
    return df


def load_all(data_path):
    data = {}
    for ticker in TICKERS:
        path_t = find_file(data_path, ticker, 'T')
        path_s = find_file(data_path, ticker, 'S')
        data[ticker] = {
            'T': load_file(path_t),
            'S': load_file(path_s),
            'path_T': path_t,
            'path_S': path_s,
        }
        print(f"  {ticker}  T: {data[ticker]['T'].shape}  "
              f"S: {data[ticker]['S'].shape}  "
              f"[{os.path.basename(path_t)}, "
              f"{os.path.basename(path_s)}]")
    return data


print("Loading files...")
DATA = load_all(DATA_PATH)
print("\nAll files loaded successfully.")

# ── DU-1: Dataset inventory ──────────────────────────────────
print("=" * 65)
print("DU-1  DATASET INVENTORY")
print("=" * 65)

rows = []
for ticker in TICKERS:
    for kind in ['T', 'S']:
        df  = DATA[ticker][kind]
        has_sent = 'Sentiment' in df.columns
        missing  = df.isnull().sum().sum()
        rows.append({
            'Ticker':         ticker,
            'File':           kind,
            'Rows':           len(df),
            'Cols':           len(df.columns),
            'Date_start':     df['Date'].min().date(),
            'Date_end':       df['Date'].max().date(),
            'Has_sentiment':  has_sent,
            'Missing_values': missing,
        })

inv = pd.DataFrame(rows)
print(inv.to_string(index=False))

print("\nColumn names per file:")
for ticker in TICKERS:
    print(f"\n  {ticker}_T: {list(DATA[ticker]['T'].columns)}")
    print(f"  {ticker}_S: {list(DATA[ticker]['S'].columns)}")

# ── DU-2: Calendar verification ──────────────────────────────
print("=" * 65)
print("DU-2  CALENDAR VERIFICATION")
print("=" * 65)

t_dates = {t: set(DATA[t]['T']['Date'].dt.date) for t in TICKERS}
base = t_dates['AAPL']
print("\nT-file date alignment (all should match AAPL):")
for ticker in TICKERS:
    diff = base.symmetric_difference(t_dates[ticker])
    print(f"  {ticker}: {len(t_dates[ticker])} trading days  "
          f"difference from AAPL: {len(diff)} days "
          f"{'OK' if len(diff) == 0 else 'MISMATCH - investigate!'}")

s_dates = {t: set(DATA[t]['S']['Date'].dt.date) for t in TICKERS}
s_base = s_dates['AAPL']
print("\nS-file date alignment:")
for ticker in TICKERS:
    diff = s_base.symmetric_difference(s_dates[ticker])
    print(f"  {ticker}: {len(s_dates[ticker])} trading days  "
          f"difference from AAPL: {len(diff)} days "
          f"{'OK' if len(diff) == 0 else 'MISMATCH'}")

print("\nS-window overlap with T-window:")
for ticker in TICKERS:
    s_in_t  = s_dates[ticker].issubset(t_dates[ticker])
    s_start = DATA[ticker]['S']['Date'].min().date()
    s_end   = DATA[ticker]['S']['Date'].max().date()
    t_end   = DATA[ticker]['T']['Date'].max().date()
    print(f"  {ticker}: S subset of T = {s_in_t}  "
          f"S window: {s_start} to {s_end}  "
          f"T ends: {t_end}")

# ── DU-3: Target variable — class balance analysis ───────────
# Critical for understanding the naive baseline and
# class-weight correction requirements.
print("=" * 65)
print("DU-3  CLASS BALANCE ANALYSIS")
print("=" * 65)


def compute_target(df):
    close = df['Close']
    y = (close.shift(-1) > close).astype(int)
    return y.iloc[:-1]


print("\nClass balance — Training files (_T_):")
print(f"{'Ticker':<8} {'UP_days':>9} {'DOWN_days':>11} "
      f"{'UP%':>8} {'Naive_base%':>13}")
print("-" * 55)

balance_rows = []
for ticker in TICKERS:
    y    = compute_target(DATA[ticker]['T'])
    up   = y.sum()
    down = len(y) - up
    up_p  = up / len(y) * 100
    naive = max(up_p, 100 - up_p)
    balance_rows.append({'ticker': ticker, 'file': 'T',
                         'up_pct': up_p, 'naive': naive})
    print(f"{ticker:<8} {up:>9} {down:>11} "
          f"{up_p:>7.2f}% {naive:>12.2f}%")

print("\nClass balance — Test window files (_S_):")
print(f"{'Ticker':<8} {'UP_days':>9} {'DOWN_days':>11} "
      f"{'UP%':>8} {'Naive_base%':>13} {'Regime':>10}")
print("-" * 65)

for ticker in TICKERS:
    y    = compute_target(DATA[ticker]['S'])
    up   = y.sum()
    down = len(y) - up
    up_p  = up / len(y) * 100
    naive = max(up_p, 100 - up_p)
    regime = ('Bullish' if up_p > 52
              else 'Bearish' if up_p < 48
              else 'Neutral')
    balance_rows.append({'ticker': ticker, 'file': 'S',
                         'up_pct': up_p, 'naive': naive,
                         'regime': regime})
    print(f"{ticker:<8} {up:>9} {down:>11} {up_p:>7.2f}% "
          f"{naive:>12.2f}% {regime:>10}")

print("\nYear-by-year UP fraction (all tickers, T file):")
year_rows = {}
for ticker in TICKERS:
    df = DATA[ticker]['T'].copy()
    df['y'] = compute_target(df).reindex(df.index)
    df['year'] = df['Date'].dt.year
    yf = df.groupby('year')['y'].mean()
    year_rows[ticker] = yf

year_df = pd.DataFrame(year_rows)
print(year_df.round(3).to_string())

fig, axes = plt.subplots(1, 2, figsize=(14, 5), facecolor=DARK)
fig.suptitle('DU-3  Class Balance Analysis', color=TEXT,
             fontsize=13, fontweight='bold', y=1.02)

ax = axes[0]
x  = np.arange(len(TICKERS))
w  = 0.35
train_ups = [next(r['up_pct'] for r in balance_rows
                  if r['ticker']==t and r['file']=='T')
             for t in TICKERS]
test_ups  = [next(r['up_pct'] for r in balance_rows
                  if r['ticker']==t and r['file']=='S')
             for t in TICKERS]

ax.bar(x - w/2, train_ups, width=w, label='Train (T)',
       color=ACCENT, alpha=0.85)
ax.bar(x + w/2, test_ups,  width=w, label='Test (S)',
       color=ORANGE, alpha=0.85)
ax.axhline(50, color=RED, linestyle='--', linewidth=1.2,
           label='50% reference')
ax.axhline(52, color=YELLOW, linestyle=':', linewidth=0.8,
           label='52% reference')
ax.set_xticks(x); ax.set_xticklabels(TICKERS, fontsize=11)
ax.set_ylabel('UP day fraction (%)', fontsize=10)
ax.set_title('UP% : Train vs Test window', color=TEXT, fontsize=11)
ax.legend(fontsize=8); ax.grid(True, axis='y')
ax.set_ylim(44, 60)

ax2 = axes[1]
hmap = year_df.T
im   = ax2.imshow(hmap.values, aspect='auto', cmap='RdYlGn',
                  vmin=0.42, vmax=0.62)
ax2.set_xticks(range(len(hmap.columns)))
ax2.set_xticklabels(hmap.columns, rotation=45, fontsize=8)
ax2.set_yticks(range(len(hmap.index)))
ax2.set_yticklabels(hmap.index, fontsize=10)
ax2.set_title('Year-by-year UP fraction', color=TEXT, fontsize=11)
for i in range(len(hmap.index)):
    for j in range(len(hmap.columns)):
        v = hmap.values[i, j]
        if not np.isnan(v):
            ax2.text(j, i, f'{v:.2f}', ha='center', va='center',
                     fontsize=7,
                     color='black' if 0.45 < v < 0.60 else TEXT)
plt.colorbar(im, ax=ax2, fraction=0.046)
test_yr = str(DATA['AAPL']['S']['Date'].dt.year.iloc[0])
if test_yr in [str(c) for c in hmap.columns]:
    j_idx = list(hmap.columns).index(int(test_yr))
    ax2.axvline(j_idx, color=YELLOW, linewidth=1.5, linestyle='--')

plt.tight_layout()
plt.savefig('DU3_class_balance.png', dpi=150,
            bbox_inches='tight', facecolor=DARK)
plt.show()
print("Figure saved: DU3_class_balance.png")

# ── DU-4: Feature distribution analysis ──────────────────────
print("=" * 65)
print("DU-4  FEATURE DISTRIBUTION & SCALE ANALYSIS")
print("=" * 65)

print("\nDescriptive statistics — AAPL T file (representative):")
aapl_t   = DATA['AAPL']['T'].copy()
num_cols = [c for c in FEAT_TECH if c in aapl_t.columns]
desc = aapl_t[num_cols].describe().T
desc['skewness'] = aapl_t[num_cols].skew()
desc['kurtosis'] = aapl_t[num_cols].kurt()
print(desc[['mean','std','min','max',
            'skewness','kurtosis']].round(3).to_string())

print("\nVolume statistics across all tickers (T file):")
print(f"{'Ticker':<8} {'Mean_Vol':>14} {'Std_Vol':>14} "
      f"{'Max_Vol':>14} {'Skew':>8}")
print("-" * 58)
for ticker in TICKERS:
    v = DATA[ticker]['T']['Volume']
    print(f"{ticker:<8} {v.mean():>14.0f} {v.std():>14.0f} "
          f"{v.max():>14.0f} {stats.skew(v):>8.3f}")

fig = plt.figure(figsize=(18, 10), facecolor=DARK)
fig.suptitle('DU-4  Feature Distributions — AAPL Training File',
             color=TEXT, fontsize=13, fontweight='bold')

n_feat       = len(num_cols)
cols_per_row = 4
n_rows       = int(np.ceil(n_feat / cols_per_row))

for idx, feat in enumerate(num_cols):
    ax       = fig.add_subplot(n_rows, cols_per_row, idx + 1)
    ax.set_facecolor(DARK)
    data_raw = aapl_t[feat].dropna()
    ax.hist(data_raw, bins=50, color=ACCENT, alpha=0.7, density=True)
    if feat == 'Volume':
        ax2 = ax.twinx()
        ax2.hist(np.log1p(data_raw), bins=50, color=GREEN,
                 alpha=0.4, density=True)
        ax2.set_yticks([])
        ax.set_title(f'{feat}\nskew={stats.skew(data_raw):.2f} '
                     f'[green=log1p]', color=TEXT, fontsize=8)
    else:
        ax.set_title(f'{feat}  skew={stats.skew(data_raw):.2f}',
                     color=TEXT, fontsize=8)
    ax.set_yticks([])
    ax.tick_params(labelsize=7)
    for spine in ax.spines.values():
        spine.set_edgecolor(BORDER)

plt.tight_layout()
plt.savefig('DU4_feature_distributions.png', dpi=150,
            bbox_inches='tight', facecolor=DARK)
plt.show()
print("Figure saved: DU4_feature_distributions.png")

# ── DU-5: Stationarity analysis ──────────────────────────────
# Proxy: rolling-mean coefficient of variation (252-day window).
# Higher CV = mean shifting over time = non-stationary.
# Reference: ADF formal test in appendix (DU-10).
print("=" * 65)
print("DU-5  STATIONARITY PROXY (Rolling-mean CV, 252-day window)")
print("=" * 65)
print("Interpretation: CV > 0.3 = likely non-stationary")
print("                CV < 0.1 = likely stationary\n")


def rolling_cv(series, w=252):
    means = series.rolling(w).mean().dropna()
    denom = abs(means.mean())
    return means.std() / denom if denom > 1e-10 else 0.0


stat_check_feats = ['Close', 'Volume', 'RSI', 'MACD', 'VIX', 'EMA_20']
header = f"{'Ticker':<8}" + "".join(f"{f:>10}"
                                    for f in stat_check_feats)
print(header)
print("-" * (8 + 10 * len(stat_check_feats)))

for ticker in TICKERS:
    df  = DATA[ticker]['T']
    row = f"{ticker:<8}"
    for feat in stat_check_feats:
        if feat in df.columns:
            cv   = rolling_cv(df[feat])
            flag = ' *' if cv > 0.5 else '  '
            row += f"{cv:>9.4f}{flag}"
        else:
            row += f"{'N/A':>10}"
    print(row)
print("\n* = likely non-stationary (high CV)")

fig, axes = plt.subplots(len(TICKERS), 2, figsize=(16, 14),
                          facecolor=DARK)
fig.suptitle('DU-5  Close Price & Volume Over Time — Non-Stationarity',
             color=TEXT, fontsize=13, fontweight='bold')

for idx, ticker in enumerate(TICKERS):
    df      = DATA[ticker]['T']
    s_start = DATA[ticker]['S']['Date'].min()
    s_end   = DATA[ticker]['S']['Date'].max()

    ax1 = axes[idx, 0]
    ax1.set_facecolor(DARK)
    ax1.plot(df['Date'], df['Close'], color=COLORS[idx],
             linewidth=0.8, alpha=0.9)
    ax1.axvspan(s_start, s_end, alpha=0.15, color=YELLOW,
                label='Test window')
    ax1.set_ylabel(ticker, color=COLORS[idx], fontsize=10)
    ax1.set_title('Close Price' if idx == 0 else '',
                  color=TEXT, fontsize=10)
    ax1.tick_params(labelsize=7)
    if idx == 0:
        ax1.legend(fontsize=7)

    ax2 = axes[idx, 1]
    ax2.set_facecolor(DARK)
    ax2.fill_between(df['Date'], np.log1p(df['Volume']),
                     color=ACCENT, alpha=0.5, linewidth=0)
    ax2.axvspan(s_start, s_end, alpha=0.15, color=YELLOW)
    ax2.set_title('log1p(Volume)' if idx == 0 else '',
                  color=TEXT, fontsize=10)
    ax2.tick_params(labelsize=7)

plt.tight_layout()
plt.savefig('DU5_price_volume_series.png', dpi=150,
            bbox_inches='tight', facecolor=DARK)
plt.show()
print("Figure saved: DU5_price_volume_series.png")

# ── DU-6: Feature correlation with target ────────────────────
# Measures linear predictability of each feature.
# Near-zero correlations justify non-linear deep learning models.
print("=" * 65)
print("DU-6  FEATURE CORRELATION WITH NEXT-DAY DIRECTION")
print("=" * 65)
print("Near-zero values justify deep learning over "
      "linear classifiers.\n")

corr_results = {}
for ticker in TICKERS:
    df = DATA[ticker]['T'].copy()
    df['y'] = compute_target(df).reindex(df.index)
    df = df.dropna(subset=['y'])
    num_cols = [c for c in FEAT_TECH if c in df.columns]
    corr = df[num_cols].corrwith(df['y']).abs()
    corr_results[ticker] = corr

corr_df = pd.DataFrame(corr_results)
corr_df['mean'] = corr_df.mean(axis=1)
print(corr_df.sort_values('mean', ascending=False)
      .round(5).to_string())

print("\nMaximum absolute correlation across all features:")
for ticker in TICKERS:
    max_c = corr_results[ticker].max()
    max_f = corr_results[ticker].idxmax()
    print(f"  {ticker}: max_corr = {max_c:.5f} ({max_f})")

fig, axes = plt.subplots(1, 2, figsize=(16, 7), facecolor=DARK)
fig.suptitle('DU-6  Feature Analysis', color=TEXT,
             fontsize=13, fontweight='bold')

ax = axes[0]
ax.set_facecolor(DARK)
corr_plot = corr_df.drop(columns='mean').sort_values(
    'AAPL', ascending=True)
y_pos = np.arange(len(corr_plot))
for i, ticker in enumerate(TICKERS):
    ax.barh(y_pos + i * 0.15, corr_plot[ticker],
            height=0.14, color=COLORS[i], alpha=0.85,
            label=ticker)
ax.set_yticks(y_pos + 0.3)
ax.set_yticklabels(corr_plot.index, fontsize=8)
ax.set_xlabel('|Correlation| with next-day direction', fontsize=9)
ax.set_title('Feature predictive signal (linear)',
             color=TEXT, fontsize=10)
ax.legend(fontsize=8)
ax.grid(True, axis='x')

ax2 = axes[1]
aapl_feats  = [c for c in FEAT_TECH
               if c in DATA['AAPL']['T'].columns]
corr_matrix = DATA['AAPL']['T'][aapl_feats].corr()
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
cmap = sns.diverging_palette(230, 20, as_cmap=True)
sns.heatmap(corr_matrix, mask=mask, cmap=cmap, center=0,
            vmin=-1, vmax=1, ax=ax2, annot=True, fmt='.2f',
            annot_kws={'size': 6}, linewidths=0.3,
            linecolor=BORDER, square=True,
            cbar_kws={'shrink': 0.6})
ax2.set_facecolor(DARK)
ax2.set_title('Inter-feature correlation (AAPL T file)\n'
              'Values near 1.0 = redundant features',
              color=TEXT, fontsize=10)
ax2.tick_params(labelsize=7, rotation=45)

plt.tight_layout()
plt.savefig('DU6_feature_correlation.png', dpi=150,
            bbox_inches='tight', facecolor=DARK)
plt.show()
print("Figure saved: DU6_feature_correlation.png")

# ── DU-7: Sentiment analysis ─────────────────────────────────
print("=" * 65)
print("DU-7  SENTIMENT SCORE ANALYSIS")
print("=" * 65)


def sent_analysis(ticker):
    df = DATA[ticker]['S'].copy()
    if 'Sentiment' not in df.columns:
        print(f"  {ticker}: No sentiment column found.")
        return None
    s  = df['Sentiment']
    y  = compute_target(df).reindex(df.index)
    df['y'] = y
    return {
        'ticker':           ticker,
        'mean':             s.mean(),
        'std':              s.std(),
        'min':              s.min(),
        'max':              s.max(),
        'skew':             stats.skew(s.dropna()),
        'autocorr_1d':      s.autocorr(lag=1),
        'autocorr_5d':      s.autocorr(lag=5),
        'near_zero%':       (s.abs() < 0.05).mean() * 100,
        'strong_pos%':      (s > 0.20).mean() * 100,
        'strong_neg%':      (s < -0.20).mean() * 100,
        'corr_w_direction': s.corr(df['y'].fillna(0)),
    }


sent_rows    = [sent_analysis(t) for t in TICKERS]
sent_rows    = [r for r in sent_rows if r is not None]
sent_df_out  = pd.DataFrame(sent_rows).set_index('ticker')
print(sent_df_out.round(4).to_string())

print("\nCross-ticker sentiment correlation matrix:")
sent_series = {}
for ticker in TICKERS:
    df = DATA[ticker]['S']
    if 'Sentiment' in df.columns:
        sent_series[ticker] = df.set_index('Date')['Sentiment']
sent_corr = pd.DataFrame(sent_series).corr()
print(sent_corr.round(4).to_string())
print("\nNote: r = 1.000 indicates two tickers share "
      "identical sentiment data.")

n_tickers_w_sent = len(sent_series)
fig, axes = plt.subplots(n_tickers_w_sent, 1,
                          figsize=(14, 3 * n_tickers_w_sent),
                          facecolor=DARK, sharex=True)
if n_tickers_w_sent == 1:
    axes = [axes]
fig.suptitle('DU-7  Sentiment Score Time Series — Test Window',
             color=TEXT, fontsize=13, fontweight='bold')

for idx, (ticker, series) in enumerate(sent_series.items()):
    ax  = axes[idx]
    ax.set_facecolor(DARK)
    pos = series.clip(lower=0)
    neg = series.clip(upper=0)
    ax.fill_between(series.index, pos, alpha=0.8,
                    color=GREEN, label='Positive')
    ax.fill_between(series.index, neg, alpha=0.8,
                    color=RED, label='Negative')
    ax.plot(series.index, series, color=MUTED, linewidth=0.5)
    ax.axhline(0, color=BORDER, linewidth=0.8)
    ax.set_ylabel(ticker, color=TEXT, fontsize=10)
    ax.grid(True, axis='y')
    corr_val = sent_df_out.loc[ticker, 'corr_w_direction']
    ax.text(0.02, 0.85,
            f'corr(sent, next_dir) = {corr_val:.4f}',
            transform=ax.transAxes, fontsize=8, color=YELLOW)
    if idx == 0:
        ax.legend(fontsize=8, loc='upper right')

plt.tight_layout()
plt.savefig('DU7_sentiment_series.png', dpi=150,
            bbox_inches='tight', facecolor=DARK)
plt.show()
print("Figure saved: DU7_sentiment_series.png")

# ── DU-8: Volume — log transform justification ───────────────
# Reduces skewness from 2.0–2.9 to 0.24–0.63.
# Prevents extreme z-scores from saturating LSTM input gates.
# Reference: Gu et al. (2020, RFS)
print("=" * 65)
print("DU-8  VOLUME — LOG TRANSFORM JUSTIFICATION")
print("=" * 65)

fig, axes = plt.subplots(2, len(TICKERS), figsize=(18, 7),
                          facecolor=DARK)
fig.suptitle('DU-8  Volume Distribution: Raw vs log1p Transform',
             color=TEXT, fontsize=13, fontweight='bold')

for idx, ticker in enumerate(TICKERS):
    vol     = DATA[ticker]['T']['Volume'].dropna()
    log_vol = np.log1p(vol)

    ax1 = axes[0, idx]
    ax1.set_facecolor(DARK)
    ax1.hist(vol, bins=60, color=RED, alpha=0.75, density=True)
    ax1.set_title(f'{ticker} raw\nskew={stats.skew(vol):.2f}',
                  color=TEXT, fontsize=8)
    ax1.set_yticks([])
    ax1.tick_params(labelsize=6)

    ax2 = axes[1, idx]
    ax2.set_facecolor(DARK)
    ax2.hist(log_vol, bins=60, color=GREEN, alpha=0.75,
             density=True)
    ax2.set_title(f'log1p\nskew={stats.skew(log_vol):.2f}',
                  color=TEXT, fontsize=8)
    ax2.set_yticks([])
    ax2.tick_params(labelsize=6)

plt.tight_layout()
plt.savefig('DU8_volume_transform.png', dpi=150,
            bbox_inches='tight', facecolor=DARK)
plt.show()
print("Figure saved: DU8_volume_transform.png")

# ── DU-9: Market regime identification ───────────────────────
print("=" * 65)
print("DU-9  MARKET REGIME IN TEST WINDOW (2021-09 to 2022-09)")
print("=" * 65)

regime_rows = []
for ticker in TICKERS:
    df      = DATA[ticker]['T'].copy()
    s_start = DATA[ticker]['S']['Date'].min()
    s_end   = DATA[ticker]['S']['Date'].max()
    test_df = df[(df['Date'] >= s_start) &
                 (df['Date'] <= s_end)]
    if len(test_df) == 0:
        continue
    close        = test_df['Close']
    total_return = (close.iloc[-1] / close.iloc[0] - 1) * 100
    max_dd       = ((close / close.cummax()) - 1).min() * 100
    vol_ann      = close.pct_change().std() * np.sqrt(252) * 100
    vix_mean     = (test_df['VIX'].mean()
                    if 'VIX' in test_df.columns else None)
    regime_rows.append({
        'Ticker':        ticker,
        'Return_%':      round(total_return, 2),
        'MaxDrawdown_%': round(max_dd, 2),
        'Ann_Vol_%':     round(vol_ann, 2),
        'VIX_mean':      round(vix_mean, 2) if vix_mean else None,
        'Regime':        ('Bear'     if total_return < -10
                          else 'Bull' if total_return > 10
                          else 'Sideways'),
    })

regime_df = pd.DataFrame(regime_rows)
print(regime_df.to_string(index=False))
print("\nNote: This window spans the 2022 tech drawdown "
      "driven by Fed rate hikes.")
print("A model trained on 2015-2021 bull data faces "
      "genuine regime shift here.")

fig, axes = plt.subplots(1, len(TICKERS), figsize=(18, 4),
                          facecolor=DARK)
fig.suptitle('DU-9  Test Window Price Performance '
             '(Normalised to 100)',
             color=TEXT, fontsize=13, fontweight='bold')

for idx, ticker in enumerate(TICKERS):
    df      = DATA[ticker]['T'].copy()
    s_start = DATA[ticker]['S']['Date'].min()
    s_end   = DATA[ticker]['S']['Date'].max()
    test_df = df[(df['Date'] >= s_start) &
                 (df['Date'] <= s_end)].copy()
    test_df['norm'] = (test_df['Close']
                       / test_df['Close'].iloc[0] * 100)
    ax    = axes[idx]
    ax.set_facecolor(DARK)
    ret   = test_df['norm'].iloc[-1] - 100
    color = GREEN if ret >= 0 else RED
    ax.fill_between(test_df['Date'], 100, test_df['norm'],
                    alpha=0.3, color=color)
    ax.plot(test_df['Date'], test_df['norm'],
            color=color, linewidth=1.2)
    ax.axhline(100, color=BORDER, linewidth=0.8,
               linestyle='--')
    ax.set_title(f'{ticker}\n{ret:+.1f}%',
                 color=TEXT, fontsize=10)
    ax.tick_params(labelsize=6, rotation=30)
    ax.grid(True, axis='y')

plt.tight_layout()
plt.savefig('DU9_test_window_performance.png', dpi=150,
            bbox_inches='tight', facecolor=DARK)
plt.show()
print("Figure saved: DU9_test_window_performance.png")

# ── DU-10: Augmented Dickey-Fuller stationarity test ─────────
# Formal stationarity test for thesis appendix.
# H0: unit root present (non-stationary)
# p < 0.05 = reject H0 = stationary
# Reference: Dickey & Fuller (1979)
print("=" * 65)
print("DU-10  AUGMENTED DICKEY-FULLER STATIONARITY TESTS")
print("=" * 65)
print("H0: series has a unit root (non-stationary)")
print("p < 0.05: reject H0 -> stationary\n")

try:
    from statsmodels.tsa.stattools import adfuller

    adf_feats = ['Close', 'Volume', 'RSI', 'MACD',
                 'VIX', 'EMA_20']
    header = (f"{'Ticker':<8} {'Feature':<12} "
              f"{'ADF_stat':>10} {'p-value':>10} "
              f"{'Stationary?':>14}")
    print(header)
    print("-" * 58)

    for ticker in TICKERS:
        df = DATA[ticker]['T']
        for feat in adf_feats:
            if feat not in df.columns:
                continue
            series = df[feat].dropna()
            try:
                result   = adfuller(series, autolag='AIC',
                                    maxlag=20)
                adf_stat = result[0]
                p_val    = result[1]
                is_stat  = 'YES' if p_val < 0.05 else 'NO (*)'
                print(f"{ticker:<8} {feat:<12} "
                      f"{adf_stat:>10.4f} {p_val:>10.4f} "
                      f"{is_stat:>14}")
            except Exception as e:
                print(f"{ticker:<8} {feat:<12} ERROR: {e}")

except ImportError:
    print("statsmodels not installed. "
          "Run: pip install statsmodels")
```

# Data Preparation

```python
# ============================================================
# Phase   : CRISP-DM — Data Preparation
#
# EXPERIMENTAL CONDITIONS:
#   C1 — Raw OHLCV only          (5 features)
#   C2 — Raw + Technical         (16 features)
#   C3 — Raw + Technical + Sent  (17 features)
#
# PREREQUISITE:
#   DATA dict must be in memory from data_understanding.py.
#   In a fresh session: re-run data_understanding.py first,
#   then run this file.
# ============================================================

import os
import pickle
import random
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats as scipy_stats
from sklearn.preprocessing import StandardScaler
from sklearn.utils.class_weight import compute_class_weight
import tensorflow as tf

# ── Reproducibility ─────────────────────────────────────────
SEED = 42
os.environ['PYTHONHASHSEED'] = str(SEED)
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# ── Plot theme ───────────────────────────────────────────────
DARK   = '#0d1117'; GRID = '#21262d'; BORDER = '#30363d'
TEXT   = '#c9d1d9'; MUTED = '#8b949e'
ACCENT = '#58a6ff'; GREEN = '#3fb950'; RED = '#f85149'
YELLOW = '#e3b341'; PURPLE = '#bc8cff'; ORANGE = '#f0883e'
COLORS = [ACCENT, GREEN, RED, YELLOW, PURPLE]

plt.rcParams.update({
    'figure.facecolor': DARK, 'axes.facecolor':  DARK,
    'axes.edgecolor':   BORDER,'axes.labelcolor': TEXT,
    'xtick.color':      MUTED, 'ytick.color':     MUTED,
    'text.color':       TEXT,  'grid.color':      GRID,
    'grid.linestyle':   '--',  'grid.alpha':      0.5,
    'font.family':      'monospace',
    'legend.facecolor': '#161b22',
    'legend.edgecolor': BORDER,
    'figure.dpi':       120,
})

# ── Experimental configuration ───────────────────────────────
CFG = dict(
    tickers    = ['AAPL', 'AMZN', 'GOOG', 'MSFT', 'TSLA'],
    tickers_c3 = ['AAPL', 'AMZN', 'GOOG', 'TSLA'],
    look_back  = 20,
    val_frac   = 0.20,
    test_start = '2021-09-30',
    test_end   = '2022-09-29',
    save_path  = '/content/',
)

# ── Feature sets ─────────────────────────────────────────────
FEAT_C1 = ['Open', 'High', 'Low', 'Close', 'Volume']

FEAT_C2 = ['Open', 'High', 'Low', 'Close', 'Volume',
           'EMA_20', 'EMA_50', 'RSI', 'RSI_MA',
           'Histogram', 'BB_Basis', 'BB_Upper', 'BB_Lower',
           'MACD', 'Signal', 'VIX']

FEAT_C3 = FEAT_C2 + ['Sentiment']

FEATURE_SETS = {'C1': FEAT_C1, 'C2': FEAT_C2, 'C3': FEAT_C3}

print("Configuration loaded.")
print(f"  C1: {len(FEAT_C1)} features — {FEAT_C1}")
print(f"  C2: {len(FEAT_C2)} features")
print(f"  C3: {len(FEAT_C3)} features  (MSFT excluded)")
print(f"  look_back  : {CFG['look_back']} days")
print(f"  val_frac   : {CFG['val_frac']*100:.0f}% of train rows")
print(f"  test window: {CFG['test_start']} to {CFG['test_end']}")


# ── Step 1 — Sentiment imputation ────────────────────────────
# GOOG(S) has 14 missing Sentiment values (confirmed in DU-1).
# Forward-fill within the (S) file only — no cross-boundary fill.
# Justification: missing = no tweets collected that day, not
# neutral sentiment. ffill is causally valid (uses only t-1).

def impute_sentiment(df, ticker):
    if 'Sentiment' not in df.columns:
        return df
    before = df['Sentiment'].isnull().sum()
    if before == 0:
        return df
    df = df.copy()
    df['Sentiment'] = df['Sentiment'].ffill()
    df['Sentiment'] = df['Sentiment'].bfill()
    after = df['Sentiment'].isnull().sum()
    print(f"  [{ticker}(S)] Sentiment NaN: {before} -> {after}  "
          f"{'OK' if after == 0 else 'PROBLEM'}")
    assert after == 0, \
        f"Imputation failed: {after} NaN remain in {ticker} Sentiment"
    return df


print("Step 1 — Sentiment imputation:")
for ticker in CFG['tickers']:
    DATA[ticker]['S'] = impute_sentiment(DATA[ticker]['S'], ticker)

total_missing = sum(
    DATA[t]['S'].isnull().sum().sum() for t in CFG['tickers'])
print(f"\nTotal missing across all (S) files after imputation: "
      f"{total_missing}  {'OK' if total_missing == 0 else 'PROBLEM'}")


# ── Step 2 — Target construction ─────────────────────────────
# y[t] = 1  if Close[t+1] > Close[t]  (UP)
# y[t] = 0  if Close[t+1] <= Close[t] (DOWN / flat)
# Built from raw Close BEFORE any transform or scaling.
# Last row dropped — no t+1 available for the final trading day.
# Close[t+1] is used for label construction only, never as a feature.

def add_target(df):
    df = df.copy()
    df['y'] = (df['Close'].shift(-1) > df['Close']).astype(np.int32)
    df = df.iloc[:-1].reset_index(drop=True)
    return df


print("Step 2 — Target construction:")
for ticker in CFG['tickers']:
    DATA[ticker]['T'] = add_target(DATA[ticker]['T'])
    DATA[ticker]['S'] = add_target(DATA[ticker]['S'])

print("\nTarget distribution after construction:")
print(f"{'Ticker':<8} {'(T) UP%':>9} {'(T) rows':>10} "
      f"{'(S) UP%':>9} {'(S) rows':>10}")
print("-" * 52)
for ticker in CFG['tickers']:
    t_up = DATA[ticker]['T']['y'].mean() * 100
    s_up = DATA[ticker]['S']['y'].mean() * 100
    print(f"{ticker:<8} {t_up:>8.2f}% {len(DATA[ticker]['T']):>10} "
          f"{s_up:>8.2f}% {len(DATA[ticker]['S']):>10}")


# ── Step 3 — Temporal split ───────────────────────────────────
# Train     : (T) rows strictly BEFORE test_start
# Validation: last val_frac of training rows (temporal, no shuffle)
# Test      : fixed 252-day (S) window — same for all conditions
#
# No shuffling at any stage. Stock forecasting is a temporal
# prediction problem; shuffling constitutes look-ahead bias.
# Reference: Prado (2018) Ch.7; Gu et al. (2020, RFS)

def temporal_split(df_T, df_S, val_frac=None):
    if val_frac is None:
        val_frac = CFG['val_frac']

    test_start    = pd.Timestamp(CFG['test_start'])
    df_train_full = df_T[df_T['Date'] < test_start].copy()
    df_test       = df_S.copy()

    # Leakage assertion — must pass before scaler is fitted
    train_max = df_train_full['Date'].max()
    test_min  = df_test['Date'].min()
    assert train_max < test_min, (
        f"LEAKAGE: train ends {train_max}, test starts {test_min}")

    n_val    = int(len(df_train_full) * val_frac)
    df_val   = df_train_full.iloc[-n_val:].copy()
    df_train = df_train_full.iloc[:-n_val].copy()

    info = {
        'train_start': df_train['Date'].min().date(),
        'train_end':   df_train['Date'].max().date(),
        'val_start':   df_val['Date'].min().date(),
        'val_end':     df_val['Date'].max().date(),
        'test_start':  df_test['Date'].min().date(),
        'test_end':    df_test['Date'].max().date(),
        'n_train':     len(df_train),
        'n_val':       len(df_val),
        'n_test':      len(df_test),
    }
    return df_train, df_val, df_test, info


print("Step 3 — Temporal split boundaries:")
SPLITS = {}
for ticker in CFG['tickers']:
    _, _, _, info = temporal_split(
        DATA[ticker]['T'], DATA[ticker]['S'])
    SPLITS[ticker] = info
    print(f"\n  {ticker}:")
    print(f"    Train : {info['train_start']} to "
          f"{info['train_end']}  ({info['n_train']} rows)")
    print(f"    Val   : {info['val_start']} to "
          f"{info['val_end']}  ({info['n_val']} rows)")
    print(f"    Test  : {info['test_start']} to "
          f"{info['test_end']}  ({info['n_test']} rows)")


# ── Step 4 — Volume log1p transform ──────────────────────────
# Applied BEFORE scaling. Deterministic — no fit required.
# Reduces volume skewness from 2.0–2.9 to 0.24–0.63 (see DU-8).
# Prevents extreme z-scores from saturating LSTM input gates.
# Reference: Gu et al. (2020, RFS); Hochreiter & Schmidhuber (1997)

def apply_log_volume(df):
    df = df.copy()
    if 'Volume' in df.columns:
        df['Volume'] = np.log1p(df['Volume'])
    return df


print("Step 4 — log1p(Volume) transform:")
for ticker in CFG['tickers']:
    DATA[ticker]['T'] = apply_log_volume(DATA[ticker]['T'])
    DATA[ticker]['S'] = apply_log_volume(DATA[ticker]['S'])

original_skews = {
    'AAPL': 2.00, 'AMZN': 2.48,
    'GOOG': 2.79, 'MSFT': 2.71, 'TSLA': 2.91}
print(f"\n{'Ticker':<8} {'Skew_before':>13} {'Skew_after':>12}")
print("-" * 36)
for ticker in CFG['tickers']:
    skew_after = scipy_stats.skew(DATA[ticker]['T']['Volume'])
    print(f"{ticker:<8} {original_skews[ticker]:>13.2f} "
          f"{skew_after:>12.4f}")


# ── Steps 5–7 — Scaling, sequence construction, output arrays

def make_sequences(X_scaled, y, look_back):
    """
    Sliding window sequence builder.
    X[i] = rows i to i+look_back-1   shape: (look_back, n_features)
    y[i] = label at row i+look_back
    Causal: no future information enters the input window.
    Reference: Lu et al. (2020); Nelson et al. (2017)
    """
    Xs, ys = [], []
    for i in range(len(X_scaled) - look_back):
        Xs.append(X_scaled[i: i + look_back])
        ys.append(y[i + look_back])
    return (np.array(Xs, dtype=np.float32),
            np.array(ys, dtype=np.int32))


def prepare_condition(ticker, condition):
    """
    Full pipeline: split -> inject C3 sentiment -> scale -> sequences.

    C3 two-stage scaler design:
      Stage A — non-Sentiment features scaled using train rows only.
      Stage B — Sentiment scaled using (S) file real values, so the
                scaler's mean/variance reflects actual test-window
                sentiment rather than the training zero-fill.
      Stage C — both scalers merged into one unified StandardScaler
                for consistent downstream inference.

    Returns
    -------
    arrays : dict — X/y for train, val, test + class_weights
    scaler : unified fitted StandardScaler
    info   : metadata dict for thesis summary table
    """
    features  = FEATURE_SETS[condition]
    is_c3     = (condition == 'C3')
    look_back = CFG['look_back']

    df_tr, df_va, df_te, split_info = temporal_split(
        DATA[ticker]['T'], DATA[ticker]['S'])

    # C3: inject Sentiment = 0.0 into (T)-sourced train/val rows.
    # The model sees a neutral prior during training and learns
    # to weight the real signal at test time.
    if is_c3:
        df_tr = df_tr.copy()
        df_va = df_va.copy()
        if 'Sentiment' not in df_tr.columns:
            df_tr['Sentiment'] = 0.0
        if 'Sentiment' not in df_va.columns:
            df_va['Sentiment'] = 0.0
        assert 'Sentiment' in df_te.columns, \
            f"[{ticker} C3] (S) file missing Sentiment column"

    # Validate all required features are present
    for sname, df in [('train', df_tr), ('val', df_va),
                      ('test', df_te)]:
        missing = [f for f in features if f not in df.columns]
        if missing:
            raise ValueError(
                f"[{ticker} {condition}] Missing in {sname}: {missing}")

    # Extract raw arrays
    X_tr_raw = df_tr[features].values.astype(np.float32)
    X_va_raw = df_va[features].values.astype(np.float32)
    X_te_raw = df_te[features].values.astype(np.float32)
    y_tr     = df_tr['y'].values.astype(np.int32)
    y_va     = df_va['y'].values.astype(np.int32)
    y_te     = df_te['y'].values.astype(np.int32)

    # Scaling
    if not is_c3:
        scaler  = StandardScaler()
        X_tr_sc = scaler.fit_transform(X_tr_raw)
        X_va_sc = scaler.transform(X_va_raw)
        X_te_sc = scaler.transform(X_te_raw)

    else:
        sent_idx     = features.index('Sentiment')
        non_sent_idx = [i for i in range(len(features))
                        if i != sent_idx]

        sc_main = StandardScaler()
        sc_main.fit(X_tr_raw[:, non_sent_idx])

        sc_sent = StandardScaler()
        sc_sent.fit(df_te[['Sentiment']].values.astype(np.float32))

        n      = len(features)
        mean_  = np.zeros(n, dtype=np.float64)
        scale_ = np.ones(n,  dtype=np.float64)
        var_   = np.ones(n,  dtype=np.float64)

        mean_[non_sent_idx]  = sc_main.mean_
        scale_[non_sent_idx] = sc_main.scale_
        var_[non_sent_idx]   = sc_main.var_

        mean_[sent_idx]  = float(sc_sent.mean_[0])
        scale_[sent_idx] = float(sc_sent.scale_[0])
        var_[sent_idx]   = float(sc_sent.var_[0])

        scaler = StandardScaler()
        scaler.mean_           = mean_
        scaler.scale_          = scale_
        scaler.var_            = var_
        scaler.n_features_in_  = n
        scaler.n_samples_seen_ = len(X_tr_raw)

        X_tr_sc = scaler.transform(X_tr_raw)
        X_va_sc = scaler.transform(X_va_raw)
        X_te_sc = scaler.transform(X_te_raw)

        s_tr = X_tr_sc[:, sent_idx]
        s_te = X_te_sc[:, sent_idx]
        print(f"    [{ticker} C3] Sentiment z-scores: "
              f"train mean={s_tr.mean():+.4f} std={s_tr.std():.6f}  "
              f"test mean={s_te.mean():+.4f} std={s_te.std():.4f}")

    # Sequence construction
    X_train, y_train = make_sequences(X_tr_sc, y_tr, look_back)
    X_val,   y_val   = make_sequences(X_va_sc, y_va, look_back)
    X_test,  y_test  = make_sequences(X_te_sc, y_te, look_back)

    # Class weights correct for the bull-bias in training distribution
    cw = compute_class_weight('balanced',
                              classes=np.array([0, 1]),
                              y=y_train)
    class_weights = {0: float(cw[0]), 1: float(cw[1])}

    arrays = dict(
        X_train=X_train, y_train=y_train,
        X_val=X_val,     y_val=y_val,
        X_test=X_test,   y_test=y_test,
        class_weights=class_weights,
    )

    info = {
        **split_info,
        'ticker':        ticker,
        'condition':     condition,
        'n_features':    len(features),
        'features':      features,
        'X_train_shape': X_train.shape,
        'X_val_shape':   X_val.shape,
        'X_test_shape':  X_test.shape,
        'class_weights': class_weights,
        'y_train_up%':   float(y_train.mean() * 100),
        'y_val_up%':     float(y_val.mean()   * 100),
        'y_test_up%':    float(y_test.mean()  * 100),
    }

    return arrays, scaler, info


print("Running sanity check on AAPL C3...")
_arr, _sc, _info = prepare_condition('AAPL', 'C3')
assert _arr['X_train'].shape[2] == 17, "C3 should have 17 features"
assert _arr['X_test'].shape[2]  == 17
assert np.isnan(_arr['X_test']).sum() == 0, "NaN found in X_test"
print(f"  X_train: {_arr['X_train'].shape}  "
      f"X_test: {_arr['X_test'].shape}  NaN: 0")
print("  Sanity check PASSED")


# ── Full preparation: all tickers × all conditions ───────────
PREPARED  = {}
SCALERS   = {}
PREP_INFO = []

print("Running full preparation pipeline...\n")

for ticker in CFG['tickers']:
    PREPARED[ticker] = {}
    SCALERS[ticker]  = {}

    for condition in ['C1', 'C2', 'C3']:

        if condition == 'C3' and ticker not in CFG['tickers_c3']:
            print(f"  [{ticker} C3] SKIPPED  "
                  f"(duplicate Sentiment with AMZN, r=1.000)")
            continue

        arrays, scaler, info = prepare_condition(ticker, condition)
        PREPARED[ticker][condition] = arrays
        SCALERS[ticker][condition]  = scaler
        PREP_INFO.append(info)

        cw = info['class_weights']
        print(f"  [{ticker} {condition}]  "
              f"train={arrays['X_train'].shape}  "
              f"val={arrays['X_val'].shape}  "
              f"test={arrays['X_test'].shape}  "
              f"cw={{DOWN:{cw[0]:.3f} UP:{cw[1]:.3f}}}")

print("\nPipeline complete.")


# ── Verification checks ───────────────────────────────────────
print("=" * 65)
print("VERIFICATION CHECKS")
print("=" * 65)

all_passed = True

for ticker in CFG['tickers']:
    for condition in ['C1', 'C2', 'C3']:

        if condition not in PREPARED.get(ticker, {}):
            continue

        arrays = PREPARED[ticker][condition]
        info   = next(i for i in PREP_INFO
                      if i['ticker']    == ticker
                      and i['condition'] == condition)
        n_feat = info['n_features']
        lb     = CFG['look_back']
        tag    = f"[{ticker} {condition}]"
        ok     = True

        for name in ['X_train','X_val','X_test',
                     'y_train','y_val','y_test']:
            n_nan = np.isnan(arrays[name].astype(float)).sum()
            if n_nan > 0:
                print(f"  FAIL {tag} {name}: {n_nan} NaN")
                ok = False

        for name in ['X_train','X_val','X_test']:
            s = arrays[name].shape
            if s[1] != lb or s[2] != n_feat:
                print(f"  FAIL {tag} {name} shape {s} "
                      f"expected (*,{lb},{n_feat})")
                ok = False

        for xn, yn in [('X_train','y_train'),('X_val','y_val'),
                       ('X_test','y_test')]:
            if len(arrays[xn]) != len(arrays[yn]):
                print(f"  FAIL {tag} {xn}/{yn} length mismatch")
                ok = False

        for name in ['y_train','y_val','y_test']:
            u = set(np.unique(arrays[name]).tolist())
            if not u.issubset({0, 1}):
                print(f"  FAIL {tag} {name} not binary: {u}")
                ok = False

        if not hasattr(SCALERS[ticker][condition], 'mean_'):
            print(f"  FAIL {tag} scaler not fitted")
            ok = False

        if arrays['X_test'].std() < 1e-6:
            print(f"  FAIL {tag} X_test near-zero variance")
            ok = False

        if condition == 'C3':
            si = info['features'].index('Sentiment')
            sv = arrays['X_test'][:, :, si].std()
            if sv < 0.1:
                print(f"  FAIL {tag} test Sentiment std={sv:.4f}")
                ok = False

        if ok:
            print(f"  PASS {tag}")
        else:
            all_passed = False

print()
if all_passed:
    print("All verification checks PASSED.")
    print("Data is ready for the Modelling phase.")
else:
    print("Some checks FAILED — review above before modelling.")


# ── Summary table ─────────────────────────────────────────────
print("\n" + "=" * 65)
print("PREPARED DATASET SUMMARY")
print("=" * 65)

rows = []
for info in PREP_INFO:
    rows.append({
        'Ticker':     info['ticker'],
        'Cond':       info['condition'],
        'Feats':      info['n_features'],
        'Train_seqs': info['X_train_shape'][0],
        'Val_seqs':   info['X_val_shape'][0],
        'Test_seqs':  info['X_test_shape'][0],
        'Train_UP%':  round(info['y_train_up%'], 2),
        'Test_UP%':   round(info['y_test_up%'],  2),
        'CW_DOWN':    round(info['class_weights'][0], 3),
        'CW_UP':      round(info['class_weights'][1], 3),
    })

print(pd.DataFrame(rows).to_string(index=False))
print("""
Notes:
  *_seqs   = file rows minus look_back (20) warm-up period
  Test_seqs should be 232 for all tickers (252 rows - 20)
  CW_DOWN/UP = class weights applied during training
  MSFT absent from C3 — Sentiment identical to AMZN (r=1.000)
""")


# ── Figure: temporal split boundaries ────────────────────────
fig, axes = plt.subplots(len(CFG['tickers']), 1,
                          figsize=(15, 11), facecolor=DARK,
                          sharex=True)
fig.suptitle('DP — Train / Validation / Test Boundaries',
             color=TEXT, fontsize=13, fontweight='bold')

for idx, ticker in enumerate(CFG['tickers']):
    ax   = axes[idx]
    ax.set_facecolor(DARK)
    df   = DATA[ticker]['T']
    info = SPLITS[ticker]
    ax.plot(df['Date'], df['Close'], color=COLORS[idx],
            linewidth=0.7, alpha=0.85)
    ax.axvspan(pd.Timestamp(info['val_start']),
               pd.Timestamp(info['val_end']),
               alpha=0.22, color=YELLOW, label='Validation')
    ax.axvspan(pd.Timestamp(CFG['test_start']),
               pd.Timestamp(CFG['test_end']),
               alpha=0.22, color=RED, label='Test window')
    ax.set_ylabel(ticker, color=COLORS[idx], fontsize=10)
    ax.grid(True, axis='y')
    ax.tick_params(labelsize=7)
    if idx == 0:
        ax.legend(fontsize=8, loc='upper left')
    ax.text(0.01, 0.78,
            f"Train={info['n_train']}  Val={info['n_val']}  "
            f"Test={info['n_test']}",
            transform=ax.transAxes, fontsize=7, color=MUTED)

plt.tight_layout()
plt.savefig('DP_splits.png', dpi=150,
            bbox_inches='tight', facecolor=DARK)
plt.show()
print("Figure saved: DP_splits.png")


# ── Figure: scaling effect ────────────────────────────────────
fig, axes = plt.subplots(2, 1, figsize=(15, 9), facecolor=DARK)
fig.suptitle('DP — Feature Scale Before and After StandardScaler '
             '(AAPL, Condition C2)',
             color=TEXT, fontsize=13, fontweight='bold')

df_tr_raw, _, _, _ = temporal_split(DATA['AAPL']['T'],
                                     DATA['AAPL']['S'])
ax1 = axes[0]
ax1.set_facecolor(DARK)
for i, feat in enumerate(FEAT_C2):
    if feat in df_tr_raw.columns:
        ax1.plot(range(len(df_tr_raw)), df_tr_raw[feat].values,
                 alpha=0.55, linewidth=0.5,
                 label=feat if i < 6 else None)
ax1.set_title('Raw features — extreme scale differences',
              color=TEXT, fontsize=10)
ax1.set_ylabel('Raw value', fontsize=9)
ax1.legend(fontsize=7, ncol=6, loc='upper left')
ax1.grid(True)

ax2 = axes[1]
ax2.set_facecolor(DARK)
X_sc   = PREPARED['AAPL']['C2']['X_train']
n_show = min(600, len(X_sc))
for i, feat in enumerate(FEAT_C2[:8]):
    ax2.plot(range(n_show), X_sc[:n_show, -1, i],
             alpha=0.55, linewidth=0.5, label=feat)
ax2.axhline(0,  color=BORDER, linewidth=0.8)
ax2.axhline(3,  color=RED, linewidth=0.5, linestyle=':', alpha=0.5)
ax2.axhline(-3, color=RED, linewidth=0.5, linestyle=':', alpha=0.5)
ax2.set_title('After StandardScaler — all features on z-score scale',
              color=TEXT, fontsize=10)
ax2.set_ylabel('z-score', fontsize=9)
ax2.legend(fontsize=7, ncol=4, loc='upper left')
ax2.grid(True)

plt.tight_layout()
plt.savefig('DP_scaling.png', dpi=150,
            bbox_inches='tight', facecolor=DARK)
plt.show()
print("Figure saved: DP_scaling.png")


# ── Figure: sliding window construction ──────────────────────
fig, ax = plt.subplots(figsize=(14, 5), facecolor=DARK)
ax.set_facecolor(DARK)
ax.set_xlim(-0.5, 27)
ax.set_ylim(-1.2, 5.2)
ax.axis('off')
ax.set_title('DP — Sliding Window Sequence Construction '
             '(look_back = 20)',
             color=TEXT, fontsize=12, fontweight='bold')

for t in range(25):
    fc  = ACCENT if t < 20 else (GREEN if t == 20 else MUTED)
    alp = 0.85 if t <= 20 else 0.25
    lw  = 1.5 if t == 20 else 0.8
    ax.add_patch(plt.Rectangle(
        (t * 1.02, 1.6), 0.88, 1.5,
        linewidth=lw, edgecolor=BORDER, facecolor=fc, alpha=alp))
    if t <= 20:
        ax.text(t * 1.02 + 0.44, 2.35, f't{t}',
                ha='center', va='center',
                fontsize=6.5, color=DARK)

ax.annotate('', xy=(19.9 * 1.02 + 0.88, 3.4),
            xytext=(0, 3.4),
            arrowprops=dict(arrowstyle='<->', color=YELLOW, lw=1.5))
ax.text(10.2, 3.7,
        'X[0] — window of 20 timesteps · shape (20, n_features)',
        ha='center', color=YELLOW, fontsize=9)
ax.annotate('', xy=(20 * 1.02 + 0.88, 1.3),
            xytext=(20 * 1.02 + 0.44, 1.3),
            arrowprops=dict(arrowstyle='->', color=GREEN, lw=1.5))
ax.text(20 * 1.02 + 0.44, 0.95, 'y[0] = direction at t20',
        ha='center', color=GREEN, fontsize=8)

for t in range(1, 22):
    fc = PURPLE if t < 21 else GREEN
    ax.add_patch(plt.Rectangle(
        (t * 1.02, 0.0), 0.88, 1.0,
        linewidth=0.8, edgecolor=BORDER, facecolor=fc, alpha=0.45))

ax.annotate('', xy=(21 * 1.02 + 0.88, 0.45),
            xytext=(1 * 1.02, 0.45),
            arrowprops=dict(arrowstyle='<->', color=PURPLE, lw=1.0))
ax.text(11.5, -0.6,
        'X[1] — shifted by 1 day  →  y[1] = direction at t21',
        ha='center', color=PURPLE, fontsize=8)
ax.text(0, 4.9,
        'Each sequence = (look_back × n_features) matrix. '
        'Label is strictly one step ahead.',
        fontsize=9, color=TEXT)

plt.tight_layout()
plt.savefig('DP_sequence.png', dpi=150,
            bbox_inches='tight', facecolor=DARK)
plt.show()
print("Figure saved: DP_sequence.png")


# ── Figure: C3 Sentiment z-score contrast ────────────────────
from scipy.stats import gaussian_kde

fig, axes = plt.subplots(1, len(CFG['tickers_c3']),
                          figsize=(16, 5), facecolor=DARK)
fig.suptitle(
    'DP — C3 Sentiment Feature: Training Zero-fill vs Real Test Values\n'
    '(Sentiment scaled using S-file statistics; '
    'training value is constant)',
    color=TEXT, fontsize=12, fontweight='bold')

for idx, ticker in enumerate(CFG['tickers_c3']):
    ax      = axes[idx]
    ax.set_facecolor(DARK)
    si      = FEATURE_SETS['C3'].index('Sentiment')
    arrays  = PREPARED[ticker]['C3']
    s_train = arrays['X_train'][:, -1, si]
    s_test  = arrays['X_test'][:,  -1, si]
    train_val = float(s_train[0])
    test_min  = float(s_test.min())
    test_max  = float(s_test.max())
    pad    = (test_max - test_min) * 0.3
    x_lo   = min(train_val - 0.4, test_min - pad)
    x_hi   = max(train_val + 0.4, test_max + pad)

    ax.hist(s_test, bins=25, color=ACCENT, alpha=0.65,
            density=True, label='Test (real Sentiment)')
    if s_test.std() > 0.01:
        kde_x = np.linspace(x_lo, x_hi, 300)
        kde_y = gaussian_kde(s_test)(kde_x)
        ax.plot(kde_x, kde_y, color=ACCENT, linewidth=1.5, alpha=0.9)
    ax.axvline(train_val, color=YELLOW, linewidth=1.8,
               linestyle='--',
               label=f'Train constant z={train_val:.3f}')
    ax.set_title(
        f'{ticker}\n'
        f'Train: z = {train_val:.3f} (constant)\n'
        f'Test:  mean={s_test.mean():+.3f}  std={s_test.std():.3f}',
        color=TEXT, fontsize=9)
    ax.set_xlabel('Scaled z-score', fontsize=8)
    ax.set_xlim(x_lo, x_hi)
    ax.set_yticks([])
    ax.grid(True, axis='x', alpha=0.4)
    ax.axvline(0, color=MUTED, linewidth=0.8, alpha=0.5)
    if idx == 0:
        ax.legend(fontsize=8, loc='upper right')

plt.tight_layout()
plt.savefig('DP_sentiment_scaling.png', dpi=150,
            bbox_inches='tight', facecolor=DARK)
plt.show()
print("Figure saved: DP_sentiment_scaling.png")

print()
print("Sentiment z-score summary (train constant vs test distribution):")
for ticker in CFG['tickers_c3']:
    si      = FEATURE_SETS['C3'].index('Sentiment')
    s_train = PREPARED[ticker]['C3']['X_train'][:, -1, si]
    s_test  = PREPARED[ticker]['C3']['X_test'][:,  -1, si]
    gap     = s_test.mean() - float(s_train[0])
    print(f"  {ticker}: train z={float(s_train[0]):+.3f}  "
          f"test mean={s_test.mean():+.3f}  "
          f"gap={gap:+.3f}  "
          f"test std={s_test.std():.3f}")


# ── Persistence ───────────────────────────────────────────────
def save_prepared(path=None):
    if path is None:
        path = CFG['save_path']
    os.makedirs(path, exist_ok=True)
    with open(os.path.join(path, 'PREPARED.pkl'),  'wb') as f:
        pickle.dump(PREPARED,  f, protocol=4)
    with open(os.path.join(path, 'SCALERS.pkl'),   'wb') as f:
        pickle.dump(SCALERS,   f, protocol=4)
    with open(os.path.join(path, 'PREP_INFO.pkl'), 'wb') as f:
        pickle.dump(PREP_INFO, f, protocol=4)
    print(f"Saved: PREPARED.pkl  SCALERS.pkl  PREP_INFO.pkl  "
          f"-> {path}")


def load_prepared(path=None):
    if path is None:
        path = CFG['save_path']
    with open(os.path.join(path, 'PREPARED.pkl'),  'rb') as f:
        prepared  = pickle.load(f)
    with open(os.path.join(path, 'SCALERS.pkl'),   'rb') as f:
        scalers   = pickle.load(f)
    with open(os.path.join(path, 'PREP_INFO.pkl'), 'rb') as f:
        prep_info = pickle.load(f)
    print(f"Loaded from {path}")
    return prepared, scalers, prep_info


save_prepared()

print()
print("=" * 65)
print("DATA PREPARATION COMPLETE")
print("=" * 65)
print(f"""
PREPARED[ticker][condition] — numpy arrays ready for modelling
SCALERS[ticker][condition]  — fitted scalers for inference
PREP_INFO                   — metadata for thesis summary table

Conditions prepared:
  C1 — 5 tickers x  5 features  (OHLCV raw)
  C2 — 5 tickers x 16 features  (OHLCV + technicals)
  C3 — 4 tickers x 17 features  (OHLCV + technicals + Sentiment)

Total model runs in Modelling phase:
  6 architectures x (C1:5 + C2:5 + C3:4) = 84 models

To reload in a fresh session:
  PREPARED, SCALERS, PREP_INFO = load_prepared('/content/')
""")
```

# Modelling

```python
# ============================================================
# Phase   : CRISP-DM — Modelling
#
# 6 architectures x 14 ticker-condition combinations = 84 models
#
# Classical ML (flattened sequence input):
#   M1 — Logistic Regression
#   M2 — Random Forest
#   M3 — XGBoost
#
# Deep Learning (3D sequence input):
#   M4 — CNN-only
#   M5 — LSTM-only
#   M6 — CNN-LSTM  (proposed hybrid)
#
# Conditions:
#   C1 — Raw OHLCV         (5 features)   5 tickers
#   C2 — Raw + Technical   (16 features)  5 tickers
#   C3 — Technical + Sent  (17 features)  4 tickers (MSFT excluded)
#
# PREREQUISITE:
#   PREPARED, SCALERS, PREP_INFO must be in memory or reloaded
#   via load_prepared() from data_preparation.py output.
# ============================================================

import os, pickle, random, warnings, time
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from scipy import stats as scipy_stats
from itertools import product

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, callbacks, regularizers
from tensorflow.keras.optimizers import Adam

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score, roc_auc_score, f1_score,
    average_precision_score, confusion_matrix,
    classification_report, roc_curve
)
import xgboost as xgb

# ── Reproducibility ──────────────────────────────────────────
SEED = 42
os.environ['PYTHONHASHSEED']       = str(SEED)
os.environ['TF_DETERMINISTIC_OPS'] = '1'
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# ── Plot theme ───────────────────────────────────────────────
DARK   = '#0d1117'; GRID = '#21262d'; BORDER = '#30363d'
TEXT   = '#c9d1d9'; MUTED = '#8b949e'
ACCENT = '#58a6ff'; GREEN = '#3fb950'; RED    = '#f85149'
YELLOW = '#e3b341'; PURPLE = '#bc8cff'; ORANGE = '#f0883e'
TEAL   = '#39d353'
COLORS = [ACCENT, GREEN, RED, YELLOW, PURPLE]

plt.rcParams.update({
    'figure.facecolor': DARK, 'axes.facecolor':  DARK,
    'axes.edgecolor':   BORDER,'axes.labelcolor': TEXT,
    'xtick.color':      MUTED, 'ytick.color':     MUTED,
    'text.color':       TEXT,  'grid.color':      GRID,
    'grid.linestyle':   '--',  'grid.alpha':      0.5,
    'font.family':      'monospace',
    'legend.facecolor': '#161b22',
    'legend.edgecolor': BORDER,
    'figure.dpi':       120,
})

print(f"TensorFlow : {tf.__version__}")
print(f"Keras      : {keras.__version__}")

# ── Reload prepared data ─────────────────────────────────────
SAVE_PATH = '/content/'

def load_prepared(path=SAVE_PATH):
    with open(os.path.join(path, 'PREPARED.pkl'),  'rb') as f:
        prepared  = pickle.load(f)
    with open(os.path.join(path, 'SCALERS.pkl'),   'rb') as f:
        scalers   = pickle.load(f)
    with open(os.path.join(path, 'PREP_INFO.pkl'), 'rb') as f:
        prep_info = pickle.load(f)
    print(f"Loaded PREPARED, SCALERS, PREP_INFO from {path}")
    return prepared, scalers, prep_info

try:
    _ = PREPARED
    print("PREPARED already in memory — skipping reload.")
except NameError:
    PREPARED, SCALERS, PREP_INFO = load_prepared()

# ── Experiment configuration ─────────────────────────────────
CFG = dict(
    tickers    = ['AAPL', 'AMZN', 'GOOG', 'MSFT', 'TSLA'],
    tickers_c3 = ['AAPL', 'AMZN', 'GOOG', 'TSLA'],
    conditions = ['C1', 'C2', 'C3'],
    look_back     = 20,
    cnn_filters   = 64,
    cnn_kernel    = 3,
    lstm_units    = 128,
    dropout       = 0.3,
    l2_reg        = 1e-4,
    learning_rate = 1e-3,
    batch_size    = 32,
    epochs        = 100,
    patience      = 15,
    rf_trees      = 200,
    rf_depth      = 8,
    xgb_trees     = 200,
    xgb_depth     = 4,
    xgb_lr        = 0.05,
    save_path     = '/content/',
)

MODEL_NAMES = [
    'Logistic Regression',
    'Random Forest',
    'XGBoost',
    'CNN-only',
    'LSTM-only',
    'CNN-LSTM',
]

print("\nConfiguration set.")
print(f"  DL: filters={CFG['cnn_filters']} kernel={CFG['cnn_kernel']} "
      f"lstm={CFG['lstm_units']} dropout={CFG['dropout']} "
      f"l2={CFG['l2_reg']} lr={CFG['learning_rate']}")
print(f"  Training: epochs={CFG['epochs']} patience={CFG['patience']} "
      f"batch={CFG['batch_size']}")


# ── Model builders ────────────────────────────────────────────
#
# CNN-LSTM ARCHITECTURE:
#   Conv1D (causal padding): extracts local temporal patterns from
#     the look_back window. Causal padding ensures no future
#     timestep is visible to earlier positions.
#     Reference: Sezer & Ozbayoglu (2018)
#   MaxPooling1D: reduces temporal dimensionality, prevents the
#     LSTM from over-fitting to fine-grained noise.
#   LSTM: captures long-range sequential dependencies across the
#     compressed representation.
#     Reference: Fischer & Krauss (2018)
#   BatchNormalization: stabilises training, mild regulariser.
#     Reference: Ioffe & Szegedy (2015)
#   Dropout: prevents co-adaptation of hidden units.
#     Reference: Srivastava et al. (2014)
#   L2 weight decay: additional regularisation for small datasets.
#     Reference: Goodfellow et al. (2016) Ch.7
#   Sigmoid output: binary classification probability.
#
# CNN-ONLY ABLATION:
#   LSTM replaced by GlobalAveragePooling1D. Isolates the
#   contribution of sequential memory.
#   Reference: Lu et al. (2020) Table 3 ablation design.
#
# LSTM-ONLY ABLATION:
#   Conv1D and MaxPooling removed. Sequence fed directly to LSTM.
#   Isolates the contribution of local feature extraction.

def build_cnn_lstm(input_shape, name='CNN-LSTM'):
    reg = regularizers.l2(CFG['l2_reg'])
    inp = layers.Input(shape=input_shape, name='input')
    x   = layers.Conv1D(
              filters=CFG['cnn_filters'],
              kernel_size=CFG['cnn_kernel'],
              padding='causal', activation='relu',
              kernel_regularizer=reg, name='conv1d')(inp)
    x   = layers.MaxPooling1D(pool_size=2, name='maxpool')(x)
    x   = layers.Dropout(CFG['dropout'], name='drop_cnn')(x)
    x   = layers.LSTM(
              CFG['lstm_units'],
              kernel_regularizer=reg,
              recurrent_regularizer=reg, name='lstm')(x)
    x   = layers.BatchNormalization(name='bn')(x)
    x   = layers.Dropout(CFG['dropout'], name='drop_lstm')(x)
    x   = layers.Dense(32, activation='relu',
                       kernel_regularizer=reg, name='dense')(x)
    out = layers.Dense(1, activation='sigmoid', name='output')(x)
    model = keras.Model(inp, out, name=name)
    model.compile(
        optimizer=Adam(CFG['learning_rate']),
        loss='binary_crossentropy',
        metrics=['accuracy'])
    return model


def build_cnn_only(input_shape, name='CNN-only'):
    reg = regularizers.l2(CFG['l2_reg'])
    inp = layers.Input(shape=input_shape, name='input')
    x   = layers.Conv1D(
              filters=CFG['cnn_filters'],
              kernel_size=CFG['cnn_kernel'],
              padding='causal', activation='relu',
              kernel_regularizer=reg, name='conv1d')(inp)
    x   = layers.MaxPooling1D(pool_size=2, name='maxpool')(x)
    x   = layers.Dropout(CFG['dropout'], name='drop_cnn')(x)
    x   = layers.GlobalAveragePooling1D(name='gap')(x)
    x   = layers.BatchNormalization(name='bn')(x)
    x   = layers.Dropout(CFG['dropout'], name='drop_dense')(x)
    x   = layers.Dense(32, activation='relu',
                       kernel_regularizer=reg, name='dense')(x)
    out = layers.Dense(1, activation='sigmoid', name='output')(x)
    model = keras.Model(inp, out, name=name)
    model.compile(
        optimizer=Adam(CFG['learning_rate']),
        loss='binary_crossentropy',
        metrics=['accuracy'])
    return model


def build_lstm_only(input_shape, name='LSTM-only'):
    reg = regularizers.l2(CFG['l2_reg'])
    inp = layers.Input(shape=input_shape, name='input')
    x   = layers.LSTM(
              CFG['lstm_units'],
              kernel_regularizer=reg,
              recurrent_regularizer=reg, name='lstm')(inp)
    x   = layers.BatchNormalization(name='bn')(x)
    x   = layers.Dropout(CFG['dropout'], name='drop')(x)
    x   = layers.Dense(32, activation='relu',
                       kernel_regularizer=reg, name='dense')(x)
    out = layers.Dense(1, activation='sigmoid', name='output')(x)
    model = keras.Model(inp, out, name=name)
    model.compile(
        optimizer=Adam(CFG['learning_rate']),
        loss='binary_crossentropy',
        metrics=['accuracy'])
    return model


print("Model builders defined.")
print("Quick shape test (AAPL C2):")
_shape = PREPARED['AAPL']['C2']['X_train'].shape[1:]
for builder, nm in [(build_cnn_lstm,  'CNN-LSTM'),
                    (build_cnn_only,  'CNN-only'),
                    (build_lstm_only, 'LSTM-only')]:
    m = builder(_shape, nm)
    n = m.count_params()
    print(f"  {nm:<12} input={_shape}  params={n:,}")
del m


# ── Training utilities ────────────────────────────────────────

def train_dl_model(model, arrays, ticker, model_name):
    """
    Train a Keras model with early stopping and LR reduction.

    Early stopping monitors val_loss with restore_best_weights=True,
    preventing over-fitting without requiring a fixed epoch count.
    Reference: Prechelt (1998)

    ReduceLROnPlateau halves the learning rate when val_loss
    stagnates for 7 consecutive epochs.
    """
    X_tr = arrays['X_train']
    y_tr = arrays['y_train']
    X_va = arrays['X_val']
    y_va = arrays['y_val']
    cw   = arrays['class_weights']

    cb_list = [
        callbacks.EarlyStopping(
            monitor='val_loss',
            patience=CFG['patience'],
            restore_best_weights=True,
            verbose=0),
        callbacks.ReduceLROnPlateau(
            monitor='val_loss',
            factor=0.5,
            patience=7,
            min_lr=1e-6,
            verbose=0),
    ]

    t0      = time.time()
    history = model.fit(
        X_tr, y_tr,
        validation_data=(X_va, y_va),
        epochs=CFG['epochs'],
        batch_size=CFG['batch_size'],
        class_weight=cw,
        callbacks=cb_list,
        verbose=0)
    elapsed = time.time() - t0
    best_ep = int(np.argmin(history.history['val_loss'])) + 1
    val_acc = history.history['val_accuracy'][best_ep - 1]
    print(f"    [{ticker} {model_name}] "
          f"best_epoch={best_ep:3d}  "
          f"val_loss={min(history.history['val_loss']):.4f}  "
          f"val_acc={val_acc:.4f}  "
          f"time={elapsed:.0f}s")
    return model, history


def evaluate_model(y_true, y_prob, ticker, condition, model_name):
    """
    Compute evaluation metrics for one model on the test set.

    Metrics reported:
      accuracy       — directional correctness (primary metric)
      roc_auc        — threshold-independent discriminative power
      f1_weighted    — harmonic mean of precision/recall (weighted)
      avg_precision  — area under precision-recall curve
    """
    y_pred = (y_prob >= 0.5).astype(int)
    acc    = accuracy_score(y_true, y_pred)
    try:
        auc = roc_auc_score(y_true, y_prob)
    except Exception:
        auc = 0.5
    f1  = f1_score(y_true, y_pred,
                   average='weighted', zero_division=0)
    ap  = average_precision_score(y_true, y_prob)
    cm  = confusion_matrix(y_true, y_pred)
    return {
        'ticker':        ticker,
        'condition':     condition,
        'model':         model_name,
        'accuracy':      round(acc,  4),
        'roc_auc':       round(auc,  4),
        'f1_weighted':   round(f1,   4),
        'avg_precision': round(ap,   4),
        'y_true':        y_true,
        'y_prob':        y_prob,
        'y_pred':        y_pred,
        'conf_matrix':   cm,
    }


def train_classical(arrays, ticker, condition):
    """
    Train LR, RF, and XGBoost on flattened sequences.

    Classical models require 2D input (samples, features).
    The 3D sequence (samples, look_back, n_features) is flattened
    to (samples, look_back × n_features).
    This is standard practice when comparing sequence models
    against classical methods on identical features.
    Reference: Fischer & Krauss (2018); Gu et al. (2020)
    """
    X_tr_f = arrays['X_train'].reshape(len(arrays['X_train']), -1)
    X_te_f = arrays['X_test'].reshape(len(arrays['X_test']),   -1)
    y_tr   = arrays['y_train']
    y_te   = arrays['y_test']
    n_neg  = (y_tr == 0).sum()
    n_pos  = (y_tr == 1).sum()
    scale_pos = n_neg / n_pos if n_pos > 0 else 1.0

    results = []
    models  = {}

    # Logistic Regression — linear baseline.
    # Expected to perform near majority-class baseline because
    # no feature has meaningful linear predictive power
    # (max |corr| = 0.045, DU-6). C=0.1 applies L2 regularisation.
    lr = LogisticRegression(
            max_iter=2000, C=0.1,
            class_weight='balanced',
            random_state=SEED, n_jobs=-1)
    lr.fit(X_tr_f, y_tr)
    prob = lr.predict_proba(X_te_f)[:, 1]
    results.append(evaluate_model(
        y_te, prob, ticker, condition, 'Logistic Regression'))
    models['Logistic Regression'] = lr

    # Random Forest — non-linear ensemble baseline.
    # Reference: Khaidem et al. (2016)
    rf = RandomForestClassifier(
            n_estimators=CFG['rf_trees'],
            max_depth=CFG['rf_depth'],
            class_weight='balanced',
            random_state=SEED, n_jobs=-1)
    rf.fit(X_tr_f, y_tr)
    prob = rf.predict_proba(X_te_f)[:, 1]
    results.append(evaluate_model(
        y_te, prob, ticker, condition, 'Random Forest'))
    models['Random Forest'] = rf

    # XGBoost — gradient boosting baseline.
    # Reference: Chen & Guestrin (2016)
    xgb_clf = xgb.XGBClassifier(
            n_estimators=CFG['xgb_trees'],
            max_depth=CFG['xgb_depth'],
            learning_rate=CFG['xgb_lr'],
            scale_pos_weight=scale_pos,
            eval_metric='logloss',
            use_label_encoder=False,
            random_state=SEED,
            verbosity=0)
    xgb_clf.fit(X_tr_f, y_tr)
    prob = xgb_clf.predict_proba(X_te_f)[:, 1]
    results.append(evaluate_model(
        y_te, prob, ticker, condition, 'XGBoost'))
    models['XGBoost'] = xgb_clf

    return results, models


print("Training utilities defined.")


# ── Main training loop ────────────────────────────────────────
RESULTS   = []
HISTORIES = {}
MODELS    = {}

print("=" * 65)
print("TRAINING LOOP — 84 models")
print("=" * 65)

for ticker in CFG['tickers']:
    print(f"\n{'─'*55}")
    print(f"  TICKER: {ticker}")
    print(f"{'─'*55}")

    for condition in CFG['conditions']:

        if condition == 'C3' and ticker not in CFG['tickers_c3']:
            print(f"\n  [{ticker} {condition}] SKIPPED")
            continue

        arrays    = PREPARED[ticker][condition]
        y_te      = arrays['y_test']
        n_feat    = arrays['X_train'].shape[2]
        inp_shape = (CFG['look_back'], n_feat)

        print(f"\n  Condition {condition}  "
              f"({n_feat} features, {len(y_te)} test sequences)")

        print("  Classical ML:")
        classical_results, classical_models = train_classical(
            arrays, ticker, condition)
        RESULTS.extend(classical_results)
        MODELS.setdefault(ticker, {})\
              .setdefault(condition, {})\
              .update(classical_models)

        for r in classical_results:
            print(f"    [{r['model']:<22}] "
                  f"acc={r['accuracy']:.4f}  "
                  f"auc={r['roc_auc']:.4f}  "
                  f"f1={r['f1_weighted']:.4f}")

        print("  Deep Learning:")
        for builder, model_name in [
            (build_cnn_only,  'CNN-only'),
            (build_lstm_only, 'LSTM-only'),
            (build_cnn_lstm,  'CNN-LSTM'),
        ]:
            tf.random.set_seed(SEED)
            np.random.seed(SEED)
            model = builder(inp_shape, model_name)
            model, hist = train_dl_model(
                model, arrays, ticker, model_name)
            y_prob = model.predict(
                arrays['X_test'], verbose=0).ravel()
            result = evaluate_model(
                y_te, y_prob, ticker, condition, model_name)
            RESULTS.append(result)
            key = f"{ticker}_{condition}_{model_name}"
            HISTORIES[key] = hist
            MODELS.setdefault(ticker, {})\
                  .setdefault(condition, {})[model_name] = model
            print(f"    [{model_name:<22}] "
                  f"acc={result['accuracy']:.4f}  "
                  f"auc={result['roc_auc']:.4f}  "
                  f"f1={result['f1_weighted']:.4f}")

    interim_df = pd.DataFrame([
        {k: v for k, v in r.items()
         if k not in ('y_true','y_prob','y_pred','conf_matrix')}
        for r in RESULTS])
    interim_df.to_csv(
        os.path.join(CFG['save_path'],
                     f'results_interim_{ticker}.csv'),
        index=False)
    print(f"\n  Saved interim results after {ticker}.")

print("\n" + "=" * 65)
print("All 84 models trained.")
print("=" * 65)


# ── Results DataFrame ─────────────────────────────────────────
RESULTS_DF = pd.DataFrame([
    {k: v for k, v in r.items()
     if k not in ('y_true','y_prob','y_pred','conf_matrix')}
    for r in RESULTS
])

RESULTS_DF.to_csv(
    os.path.join(CFG['save_path'], 'results_df.csv'), index=False)

print("RESULTS_DF shape:", RESULTS_DF.shape)
print("\nColumn list:", list(RESULTS_DF.columns))
print("\nUnique models:", RESULTS_DF['model'].unique().tolist())
print("Unique conditions:", RESULTS_DF['condition'].unique().tolist())
print("Unique tickers:", RESULTS_DF['ticker'].unique().tolist())


# ── Results summary tables ────────────────────────────────────
print("=" * 65)
print("RESULTS — DIRECTIONAL ACCURACY BY MODEL AND CONDITION")
print("=" * 65)

for condition in CFG['conditions']:
    sub = RESULTS_DF[RESULTS_DF['condition'] == condition]
    if sub.empty:
        continue
    pivot = sub.pivot_table(
        index='model', columns='ticker',
        values='accuracy', aggfunc='first')
    pivot['Mean'] = pivot.mean(axis=1)
    pivot = pivot.sort_values('Mean', ascending=False)
    print(f"\n  Condition {condition}:")
    print(pivot.round(4).to_string())

print("\n" + "=" * 65)
print("RESULTS — MEAN ACCURACY ACROSS TICKERS (all conditions)")
print("=" * 65)
mean_acc = RESULTS_DF.groupby(['condition','model'])['accuracy']\
                     .mean().unstack('condition').round(4)
mean_acc['Overall'] = mean_acc.mean(axis=1)
print(mean_acc.sort_values('Overall', ascending=False).to_string())

print("\n" + "=" * 65)
print("RESULTS — ROC-AUC BY MODEL AND CONDITION")
print("=" * 65)
for condition in CFG['conditions']:
    sub = RESULTS_DF[RESULTS_DF['condition'] == condition]
    if sub.empty:
        continue
    pivot = sub.pivot_table(
        index='model', columns='ticker',
        values='roc_auc', aggfunc='first')
    pivot['Mean'] = pivot.mean(axis=1)
    pivot = pivot.sort_values('Mean', ascending=False)
    print(f"\n  Condition {condition}:")
    print(pivot.round(4).to_string())


# ── Diebold-Mariano tests ─────────────────────────────────────
# Tests whether two forecasters have equal predictive accuracy.
# H0: equal predictive accuracy (0-1 loss function).
# Harvey et al. (1997) small-sample correction applied.
# Reference: Diebold & Mariano (1995, JBES)
#            Harvey, Leybourne & Newbold (1997, IJF)

def dm_test(y_true, prob1, prob2, h=1):
    """
    Diebold-Mariano test for equal predictive accuracy.
    y_true : binary ground truth
    prob1  : probability scores from model 1 (baseline)
    prob2  : probability scores from model 2 (proposed)
    h      : forecast horizon (1 for one-step-ahead)
    Negative DM stat: model 1 has lower loss (baseline better).
    """
    pred1  = (prob1 >= 0.5).astype(int)
    pred2  = (prob2 >= 0.5).astype(int)
    e1     = (pred1 != y_true).astype(float)
    e2     = (pred2 != y_true).astype(float)
    d      = e1 - e2
    T      = len(d)
    d_bar  = d.mean()
    gamma0 = np.var(d, ddof=1)
    if gamma0 < 1e-12:
        return 0.0, 1.0
    hac_factor = np.sqrt((T + 1 - 2*h + h*(h-1)/T) / T)
    dm_raw = d_bar / np.sqrt(gamma0 / T)
    dm_hc  = dm_raw * hac_factor
    p_val  = 2 * scipy_stats.t.sf(abs(dm_hc), df=T-1)
    return round(dm_hc, 4), round(p_val, 4)


print("=" * 65)
print("DIEBOLD-MARIANO TESTS")
print("H0: equal predictive accuracy vs CNN-LSTM")
print("Negative DM stat: baseline has higher loss (CNN-LSTM better)")
print("=" * 65)

DM_RESULTS = []
baselines  = ['Logistic Regression', 'Random Forest',
               'XGBoost', 'CNN-only', 'LSTM-only']

for condition in CFG['conditions']:
    print(f"\n  Condition {condition}:")
    tickers_this = (CFG['tickers_c3']
                    if condition == 'C3'
                    else CFG['tickers'])

    for ticker in tickers_this:
        cnn_lstm_r = next(
            (r for r in RESULTS
             if r['ticker']    == ticker
             and r['condition'] == condition
             and r['model']     == 'CNN-LSTM'), None)
        if cnn_lstm_r is None:
            continue

        print(f"\n    {ticker}:")
        print(f"    {'Baseline':<24} {'DM_stat':>9} "
              f"{'p-value':>9} {'Sig(5%)':>9} "
              f"{'CNN-LSTM_acc':>13} {'Base_acc':>10}")
        print(f"    {'-'*70}")

        for baseline in baselines:
            base_r = next(
                (r for r in RESULTS
                 if r['ticker']    == ticker
                 and r['condition'] == condition
                 and r['model']     == baseline), None)
            if base_r is None:
                continue
            n  = min(len(cnn_lstm_r['y_true']),
                     len(base_r['y_true']))
            dm, pv = dm_test(
                cnn_lstm_r['y_true'][:n],
                base_r['y_prob'][:n],
                cnn_lstm_r['y_prob'][:n])
            sig = "YES" if pv < 0.05 else ("~" if pv < 0.10 else "no")
            print(f"    {baseline:<24} {dm:>9.4f} "
                  f"{pv:>9.4f} {sig:>9} "
                  f"{cnn_lstm_r['accuracy']:>13.4f} "
                  f"{base_r['accuracy']:>10.4f}")
            DM_RESULTS.append({
                'ticker':       ticker,
                'condition':    condition,
                'baseline':     baseline,
                'dm_stat':      dm,
                'p_value':      pv,
                'significant':  pv < 0.05,
                'cnn_lstm_acc': cnn_lstm_r['accuracy'],
                'base_acc':     base_r['accuracy'],
            })

DM_DF = pd.DataFrame(DM_RESULTS)
DM_DF.to_csv(
    os.path.join(CFG['save_path'], 'dm_results.csv'), index=False)
print("\nDM results saved: dm_results.csv")

print("\nDM test significance summary:")
print(f"  Total tests run    : {len(DM_DF)}")
print(f"  p < 0.05  (sig)    : {(DM_DF['p_value'] < 0.05).sum()}")
print(f"  p < 0.10  (trend)  : {(DM_DF['p_value'] < 0.10).sum()}")
print(f"  CNN-LSTM better acc: "
      f"{(DM_DF['cnn_lstm_acc'] > DM_DF['base_acc']).sum()} "
      f"/ {len(DM_DF)}")


# ── Figure 1: Training histories ─────────────────────────────
print("Generating Figure 1: Training histories...")

for condition in CFG['conditions']:
    tickers_this = (CFG['tickers_c3']
                    if condition == 'C3'
                    else CFG['tickers'])
    n = len(tickers_this)

    fig, axes = plt.subplots(
        2, n, figsize=(4*n, 8), facecolor=DARK)
    fig.suptitle(
        f'Training History — Condition {condition}',
        color=TEXT, fontsize=13, fontweight='bold')

    dl_models = ['CNN-only', 'LSTM-only', 'CNN-LSTM']
    dl_colors = [YELLOW, ORANGE, ACCENT]

    for idx, ticker in enumerate(tickers_this):
        ax_loss = axes[0, idx]
        ax_acc  = axes[1, idx]
        for ax in [ax_loss, ax_acc]:
            ax.set_facecolor(DARK)
        for model_name, col in zip(dl_models, dl_colors):
            key  = f"{ticker}_{condition}_{model_name}"
            hist = HISTORIES.get(key)
            if hist is None:
                continue
            ep = range(1, len(hist.history['val_loss']) + 1)
            ax_loss.plot(ep, hist.history['val_loss'],
                         color=col, linewidth=1.2,
                         label=model_name)
            ax_acc.plot(ep, hist.history['val_accuracy'],
                        color=col, linewidth=1.2,
                        label=model_name)
        ax_loss.set_title(ticker, color=TEXT, fontsize=10)
        ax_loss.set_ylabel('val_loss' if idx==0 else '', fontsize=8)
        ax_acc.set_ylabel('val_accuracy' if idx==0 else '', fontsize=8)
        ax_acc.set_xlabel('Epoch', fontsize=8)
        for ax in [ax_loss, ax_acc]:
            ax.grid(True, alpha=0.3)
            ax.tick_params(labelsize=7)
        if idx == 0:
            ax_loss.legend(fontsize=7)

    plt.tight_layout()
    fname = f'FIG1_training_history_{condition}.png'
    plt.savefig(fname, dpi=150,
                bbox_inches='tight', facecolor=DARK)
    plt.show()
    print(f"  Saved: {fname}")


# ── Figure 2: Accuracy comparison bar charts ──────────────────
MAJORITY_CLASS = {
    'AAPL': 0.5020, 'AMZN': 0.5139,
    'GOOG': 0.5100, 'MSFT': 0.5179, 'TSLA': 0.5299}

MODEL_COLORS = {
    'Logistic Regression': '#6e7681',
    'Random Forest':       '#8b949e',
    'XGBoost':             '#c9d1d9',
    'CNN-only':            YELLOW,
    'LSTM-only':           ORANGE,
    'CNN-LSTM':            ACCENT,
}

print("Generating Figure 2: Accuracy comparison...")

for condition in CFG['conditions']:
    tickers_this = (CFG['tickers_c3']
                    if condition == 'C3'
                    else CFG['tickers'])
    sub = RESULTS_DF[RESULTS_DF['condition'] == condition]
    n_t = len(tickers_this)
    n_m = len(MODEL_NAMES)
    w   = 0.12
    x   = np.arange(n_t)

    fig, ax = plt.subplots(figsize=(3*n_t + 4, 6), facecolor=DARK)
    ax.set_facecolor(DARK)

    for i, model_name in enumerate(MODEL_NAMES):
        accs = []
        for ticker in tickers_this:
            row = sub[(sub['ticker']==ticker) &
                      (sub['model']==model_name)]
            accs.append(float(row['accuracy'].values[0])
                        if len(row) > 0 else 0.0)
        ax.bar(x + i*w, accs, width=w,
               color=MODEL_COLORS[model_name],
               alpha=0.88, label=model_name)

    ax.axhline(0.50,  color=RED,    linestyle='--', linewidth=1.2,
               label='50% random', alpha=0.8)
    ax.axhline(0.532, color=PURPLE, linestyle=':',  linewidth=1.0,
               label='Fischer & Krauss (2018) 53.2%', alpha=0.7)
    ax.axhline(0.559, color=TEAL,   linestyle=':',  linewidth=1.0,
               label='Lu et al. (2020) 55.9%', alpha=0.7)

    for j, ticker in enumerate(tickers_this):
        mc = MAJORITY_CLASS.get(ticker, 0.50)
        ax.plot(x[j] + (n_m-1)*w/2, mc,
                marker='_', markersize=20,
                color=YELLOW, markeredgewidth=2, zorder=5)

    ax.set_xticks(x + w*(n_m-1)/2)
    ax.set_xticklabels(tickers_this, fontsize=11)
    ax.set_ylabel('Directional Accuracy', fontsize=10)
    ax.set_title(
        f'Model Comparison — Directional Accuracy | '
        f'Condition {condition}\n'
        f'Yellow bar = majority-class baseline per ticker',
        color=TEXT, fontsize=11, fontweight='bold')
    ax.legend(loc='lower right', fontsize=7, ncol=3)
    ax.set_ylim(0.38, 0.70)
    ax.grid(True, axis='y', alpha=0.3)

    plt.tight_layout()
    fname = f'FIG2_accuracy_{condition}.png'
    plt.savefig(fname, dpi=150,
                bbox_inches='tight', facecolor=DARK)
    plt.show()
    print(f"  Saved: {fname}")


# ── Figure 3: ROC curves ──────────────────────────────────────
print("Generating Figure 3: ROC curves...")

for condition in CFG['conditions']:
    tickers_this = (CFG['tickers_c3']
                    if condition == 'C3'
                    else CFG['tickers'])
    n = len(tickers_this)

    fig, axes = plt.subplots(1, n,
                              figsize=(4*n, 4), facecolor=DARK)
    if n == 1:
        axes = [axes]
    fig.suptitle(
        f'ROC Curves — DL Models | Condition {condition}',
        color=TEXT, fontsize=12, fontweight='bold')

    dl_models_roc = ['CNN-only', 'LSTM-only', 'CNN-LSTM']
    dl_colors_roc = [YELLOW, ORANGE, ACCENT]

    for idx, ticker in enumerate(tickers_this):
        ax = axes[idx]
        ax.set_facecolor(DARK)
        ax.plot([0,1],[0,1], color=BORDER,
                linestyle='--', linewidth=0.8)
        for model_name, col in zip(dl_models_roc, dl_colors_roc):
            r = next(
                (x for x in RESULTS
                 if x['ticker']    == ticker
                 and x['condition'] == condition
                 and x['model']     == model_name), None)
            if r is None:
                continue
            fpr, tpr, _ = roc_curve(r['y_true'], r['y_prob'])
            ax.plot(fpr, tpr, color=col, linewidth=1.5,
                    label=f"{model_name} ({r['roc_auc']:.3f})")
        ax.set_title(ticker, color=TEXT, fontsize=10)
        ax.set_xlabel('FPR', fontsize=8)
        if idx == 0:
            ax.set_ylabel('TPR', fontsize=8)
        ax.legend(fontsize=7, loc='lower right')
        ax.grid(True, alpha=0.3)
        ax.tick_params(labelsize=7)

    plt.tight_layout()
    fname = f'FIG3_roc_{condition}.png'
    plt.savefig(fname, dpi=150,
                bbox_inches='tight', facecolor=DARK)
    plt.show()
    print(f"  Saved: {fname}")


# ── Figure 4: Confusion matrices ─────────────────────────────
print("Generating Figure 4: Confusion matrices...")

from matplotlib.colors import LinearSegmentedColormap
cmap_dark = LinearSegmentedColormap.from_list(
    'dark_blue', [DARK, ACCENT])

for condition in CFG['conditions']:
    tickers_this = (CFG['tickers_c3']
                    if condition == 'C3'
                    else CFG['tickers'])
    n = len(tickers_this)

    fig, axes = plt.subplots(1, n,
                              figsize=(3.5*n, 4), facecolor=DARK)
    if n == 1:
        axes = [axes]
    fig.suptitle(
        f'Confusion Matrices — CNN-LSTM | Condition {condition}',
        color=TEXT, fontsize=12, fontweight='bold')

    for idx, ticker in enumerate(tickers_this):
        r = next(
            (x for x in RESULTS
             if x['ticker']    == ticker
             and x['condition'] == condition
             and x['model']     == 'CNN-LSTM'), None)
        if r is None:
            continue
        cm  = r['conf_matrix']
        ax  = axes[idx]
        ax.set_facecolor(DARK)
        ax.imshow(cm, cmap=cmap_dark)
        ax.set_title(f'{ticker}\nacc={r["accuracy"]:.4f}',
                     color=TEXT, fontsize=10)
        ax.set_xticks([0,1]); ax.set_yticks([0,1])
        ax.set_xticklabels(['DOWN','UP'], fontsize=9)
        ax.set_yticklabels(['DOWN','UP'], fontsize=9)
        if idx == 0:
            ax.set_ylabel('Actual', fontsize=9)
        ax.set_xlabel('Predicted', fontsize=9)
        for i in range(2):
            for j in range(2):
                ax.text(j, i, str(cm[i,j]),
                        ha='center', va='center',
                        fontsize=13, fontweight='bold',
                        color=TEXT)

    plt.tight_layout()
    fname = f'FIG4_confusion_{condition}.png'
    plt.savefig(fname, dpi=150,
                bbox_inches='tight', facecolor=DARK)
    plt.show()
    print(f"  Saved: {fname}")


# ── Figure 5: Ablation study ──────────────────────────────────
print("Generating Figure 5: Ablation study...")

fig, axes = plt.subplots(1, 2, figsize=(14, 6), facecolor=DARK)
fig.suptitle(
    'Ablation Study — Marginal Value of Feature Engineering\n'
    'CNN-LSTM accuracy across conditions',
    color=TEXT, fontsize=12, fontweight='bold')

ax1          = axes[0]
ax1.set_facecolor(DARK)
cnn_lstm_sub = RESULTS_DF[RESULTS_DF['model'] == 'CNN-LSTM']
conditions   = ['C1', 'C2', 'C3']
cond_colors  = [MUTED, ACCENT, GREEN]
cond_labels  = ['C1 (raw OHLCV)', 'C2 (+Technicals)', 'C3 (+Sentiment)']
c3_tickers   = CFG['tickers_c3']
x = np.arange(len(c3_tickers))
w = 0.25

for i, (cond, col, lbl) in enumerate(
        zip(conditions, cond_colors, cond_labels)):
    accs = []
    for ticker in c3_tickers:
        row = cnn_lstm_sub[
            (cnn_lstm_sub['ticker']==ticker) &
            (cnn_lstm_sub['condition']==cond)]
        accs.append(float(row['accuracy'].values[0])
                    if len(row) > 0 else 0.0)
    ax1.bar(x + i*w, accs, width=w, color=col,
            alpha=0.85, label=lbl)

ax1.axhline(0.50, color=RED, linestyle='--',
            linewidth=1.0, alpha=0.7)
ax1.set_xticks(x + w)
ax1.set_xticklabels(c3_tickers, fontsize=11)
ax1.set_ylabel('Directional Accuracy', fontsize=10)
ax1.set_title('CNN-LSTM: C1 vs C2 vs C3\n'
              '(tickers in C3 only)',
              color=TEXT, fontsize=10)
ax1.legend(fontsize=9)
ax1.set_ylim(0.38, 0.70)
ax1.grid(True, axis='y', alpha=0.3)

ax2 = axes[1]
ax2.set_facecolor(DARK)
mean_by_cond = RESULTS_DF.groupby(
    ['model','condition'])['accuracy'].mean().unstack('condition')
model_order  = ['Logistic Regression','Random Forest','XGBoost',
                'CNN-only','LSTM-only','CNN-LSTM']
mean_by_cond = mean_by_cond.reindex(model_order)
y_pos = np.arange(len(model_order))
bar_h = 0.25

for i, (cond, col, lbl) in enumerate(
        zip(conditions, cond_colors, cond_labels)):
    if cond not in mean_by_cond.columns:
        continue
    vals = mean_by_cond[cond].fillna(0).values
    ax2.barh(y_pos + i*bar_h, vals, height=bar_h,
             color=col, alpha=0.85, label=lbl)

ax2.axvline(0.50, color=RED, linestyle='--',
            linewidth=1.0, alpha=0.7)
ax2.set_yticks(y_pos + bar_h)
ax2.set_yticklabels(model_order, fontsize=9)
ax2.set_xlabel('Mean Accuracy (across tickers)', fontsize=10)
ax2.set_title('All models: mean accuracy per condition',
              color=TEXT, fontsize=10)
ax2.legend(fontsize=9, loc='lower right')
ax2.set_xlim(0.38, 0.70)
ax2.grid(True, axis='x', alpha=0.3)

plt.tight_layout()
plt.savefig('FIG5_ablation.png', dpi=150,
            bbox_inches='tight', facecolor=DARK)
plt.show()
print("Saved: FIG5_ablation.png")


# ── Thesis results summary table ─────────────────────────────
print()
print("=" * 65)
print("THESIS RESULTS SUMMARY TABLE")
print("=" * 65)

summary = RESULTS_DF.groupby(['condition','model']).agg(
    Acc_mean  = ('accuracy',      'mean'),
    Acc_std   = ('accuracy',      'std'),
    AUC_mean  = ('roc_auc',       'mean'),
    F1_mean   = ('f1_weighted',   'mean'),
    AP_mean   = ('avg_precision', 'mean'),
).round(4)
print(summary.to_string())

print("\nBenchmark comparison:")
cnn_lstm_mean = RESULTS_DF[
    RESULTS_DF['model']=='CNN-LSTM'].groupby(
    'condition')['accuracy'].mean()
print(f"  Fischer & Krauss (2018) LSTM      : 53.20%")
print(f"  Lu et al. (2020) CNN-LSTM         : 55.90%")
for cond in ['C1','C2','C3']:
    if cond in cnn_lstm_mean.index:
        print(f"  This study CNN-LSTM ({cond})        "
              f": {cnn_lstm_mean[cond]*100:.2f}%")

print()
print("DM test significant results (p < 0.05):")
sig = DM_DF[DM_DF['p_value'] < 0.05]
if len(sig) > 0:
    print(sig[['ticker','condition','baseline',
               'dm_stat','p_value',
               'cnn_lstm_acc','base_acc']].to_string(index=False))
else:
    print("  No significant results at p < 0.05")
    print("  Results at p < 0.10:")
    trend = DM_DF[DM_DF['p_value'] < 0.10]
    if len(trend) > 0:
        print(trend[['ticker','condition','baseline',
                     'dm_stat','p_value',
                     'cnn_lstm_acc','base_acc']].to_string(
                         index=False))
    else:
        print("  No results at p < 0.10 either.")

print()
print("=" * 65)
print("INTERPRETATION GUIDE FOR THESIS CHAPTER 4")
print("=" * 65)
print("""
CNN-LSTM vs benchmarks:
  Compare CNN-LSTM mean accuracy in C2 against Fischer & Krauss
  (2018) 53.2% and Lu et al. (2020) 55.9%. Even if below, you
  tested on a bear-market regime — harder than their evaluation.

Ablation (C1 -> C2 -> C3):
  C2 > C1: technical indicators add value over raw OHLCV.
  C3 > C2: sentiment adds value over technicals.
  C3 < C2: sentiment adds noise (also a valid, publishable finding).

DM test:
  p < 0.05 for any ticker = statistically significant improvement.
  No significance = models are statistically equivalent —
  common in financial forecasting. Report this honestly.
  The DM test is conservative with n=231 test observations.

Architecture ablation:
  CNN-LSTM > CNN-only: sequential memory (LSTM) adds value.
  CNN-LSTM > LSTM-only: local feature extraction (CNN) adds value.
  If CNN-LSTM < both: hybrid does not improve over components alone.
""")


# ── Save all outputs ──────────────────────────────────────────
with open(os.path.join(CFG['save_path'], 'RESULTS.pkl'), 'wb') as f:
    pickle.dump(RESULTS, f, protocol=4)
with open(os.path.join(CFG['save_path'], 'HISTORIES.pkl'), 'wb') as f:
    pickle.dump(HISTORIES, f, protocol=4)

DM_DF.to_csv(
    os.path.join(CFG['save_path'], 'dm_results.csv'), index=False)
RESULTS_DF.to_csv(
    os.path.join(CFG['save_path'], 'results_df.csv'), index=False)

print("All outputs saved:")
print("  RESULTS.pkl    — full result dicts including arrays")
print("  HISTORIES.pkl  — Keras training histories")
print("  results_df.csv — scalar metrics for all 84 models")
print("  dm_results.csv — Diebold-Mariano test results")
print()
print("Modelling phase COMPLETE.")
```

```python
# ============================================================
# Phase   : CRISP-DM — Modelling, Run 1
#
# Changes from Run 0 (baseline):
#   learning rate : 1e-3  → 1e-4
#   LSTM units    : 128   → 64
#   CNN filters   : 64    → 32
#   patience      : 15    → 20
#
# RATIONALE:
#   Run 0 showed models stopping at epoch 1-5.
#   Root cause: lr=1e-3 is too large for financial sequences —
#   the loss spikes immediately and early stopping fires before
#   any real learning occurs.
#
#   The CNN-LSTM had 106,625 parameters trained on only 1,339
#   sequences — a 79:1 parameter-to-sample ratio. Best practice
#   is below 10:1 (Goodfellow et al., 2016 Ch.11). Reducing to
#   LSTM=64, filters=32 gives ~25K parameters (19:1 ratio).
#
#   These changes are bundled because they are all hyperparameter
#   choices, not structural changes. The architecture is identical
#   to Run 0. Only the training configuration changes.
#
# PREREQUISITE:
#   PREPARED, SCALERS, PREP_INFO in memory or reload from disk.
# ============================================================

import os, pickle, random, warnings, time
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats as scipy_stats

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, callbacks, regularizers
from tensorflow.keras.optimizers import Adam
from sklearn.metrics import (
    accuracy_score, roc_auc_score, f1_score,
    average_precision_score, confusion_matrix, roc_curve)
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
import xgboost as xgb

# ── Reproducibility ──────────────────────────────────────────
SEED = 42
os.environ['PYTHONHASHSEED']       = str(SEED)
os.environ['TF_DETERMINISTIC_OPS'] = '1'
random.seed(SEED); np.random.seed(SEED); tf.random.set_seed(SEED)

# ── Plot theme ───────────────────────────────────────────────
DARK   = '#0d1117'; BORDER = '#30363d'; TEXT = '#c9d1d9'
MUTED  = '#8b949e'
ACCENT = '#58a6ff'; GREEN = '#3fb950'; RED = '#f85149'
YELLOW = '#e3b341'; PURPLE = '#bc8cff'; ORANGE = '#f0883e'
COLORS = [ACCENT, GREEN, RED, YELLOW, PURPLE]

plt.rcParams.update({
    'figure.facecolor': DARK, 'axes.facecolor':  DARK,
    'axes.edgecolor':   BORDER,'axes.labelcolor': TEXT,
    'xtick.color': MUTED, 'ytick.color': MUTED, 'text.color': TEXT,
    'grid.color': '#21262d', 'grid.linestyle': '--', 'grid.alpha': 0.5,
    'font.family': 'monospace', 'legend.facecolor': '#161b22',
    'legend.edgecolor': BORDER, 'figure.dpi': 120,
})

SAVE_PATH = '/content/'

def load_prepared(path=SAVE_PATH):
    with open(os.path.join(path, 'PREPARED.pkl'),  'rb') as f:
        prepared  = pickle.load(f)
    with open(os.path.join(path, 'SCALERS.pkl'),   'rb') as f:
        scalers   = pickle.load(f)
    with open(os.path.join(path, 'PREP_INFO.pkl'), 'rb') as f:
        prep_info = pickle.load(f)
    print(f"Loaded from {path}")
    return prepared, scalers, prep_info

try:
    _ = PREPARED
    print("PREPARED in memory.")
except NameError:
    PREPARED, SCALERS, PREP_INFO = load_prepared()

# ── Run 1 configuration ───────────────────────────────────────
CFG = dict(
    tickers    = ['AAPL', 'AMZN', 'GOOG', 'MSFT', 'TSLA'],
    tickers_c3 = ['AAPL', 'AMZN', 'GOOG', 'TSLA'],
    conditions = ['C1', 'C2', 'C3'],
    look_back  = 20,
    learning_rate = 1e-4,
    lstm_units    = 64,
    cnn_filters   = 32,
    patience      = 20,
    cnn_kernel    = 3,
    dropout       = 0.3,
    l2_reg        = 1e-4,
    batch_size    = 32,
    epochs        = 150,
    rf_trees      = 200,
    rf_depth      = 8,
    xgb_trees     = 200,
    xgb_depth     = 4,
    xgb_lr        = 0.05,
    save_path     = SAVE_PATH,
    run_name      = 'run1',
)

print("Run 1 configuration:")
print(f"  lr={CFG['learning_rate']}  lstm={CFG['lstm_units']}  "
      f"filters={CFG['cnn_filters']}  patience={CFG['patience']}")
print(f"  epochs ceiling={CFG['epochs']}  "
      f"(early stopping governs actual stop)")


# ── Model builders ────────────────────────────────────────────
# Architecture identical to Run 0.
# Parameter counts reduced due to smaller lstm_units and cnn_filters.

def build_cnn_lstm(input_shape, name='CNN-LSTM'):
    reg = regularizers.l2(CFG['l2_reg'])
    inp = layers.Input(shape=input_shape)
    x   = layers.Conv1D(CFG['cnn_filters'], CFG['cnn_kernel'],
                        padding='causal', activation='relu',
                        kernel_regularizer=reg)(inp)
    x   = layers.MaxPooling1D(pool_size=2)(x)
    x   = layers.Dropout(CFG['dropout'])(x)
    x   = layers.LSTM(CFG['lstm_units'],
                      kernel_regularizer=reg,
                      recurrent_regularizer=reg)(x)
    x   = layers.BatchNormalization()(x)
    x   = layers.Dropout(CFG['dropout'])(x)
    x   = layers.Dense(32, activation='relu',
                       kernel_regularizer=reg)(x)
    out = layers.Dense(1, activation='sigmoid')(x)
    m   = keras.Model(inp, out, name=name)
    m.compile(optimizer=Adam(CFG['learning_rate']),
              loss='binary_crossentropy', metrics=['accuracy'])
    return m


def build_cnn_only(input_shape, name='CNN-only'):
    reg = regularizers.l2(CFG['l2_reg'])
    inp = layers.Input(shape=input_shape)
    x   = layers.Conv1D(CFG['cnn_filters'], CFG['cnn_kernel'],
                        padding='causal', activation='relu',
                        kernel_regularizer=reg)(inp)
    x   = layers.Conv1D(CFG['cnn_filters'], CFG['cnn_kernel'],
                        padding='causal', activation='relu',
                        kernel_regularizer=reg)(x)
    x   = layers.GlobalAveragePooling1D()(x)
    x   = layers.BatchNormalization()(x)
    x   = layers.Dropout(CFG['dropout'])(x)
    x   = layers.Dense(32, activation='relu',
                       kernel_regularizer=reg)(x)
    out = layers.Dense(1, activation='sigmoid')(x)
    m   = keras.Model(inp, out, name=name)
    m.compile(optimizer=Adam(CFG['learning_rate']),
              loss='binary_crossentropy', metrics=['accuracy'])
    return m


def build_lstm_only(input_shape, name='LSTM-only'):
    reg = regularizers.l2(CFG['l2_reg'])
    inp = layers.Input(shape=input_shape)
    x   = layers.LSTM(CFG['lstm_units'],
                      kernel_regularizer=reg,
                      recurrent_regularizer=reg)(inp)
    x   = layers.BatchNormalization()(x)
    x   = layers.Dropout(CFG['dropout'])(x)
    x   = layers.Dense(32, activation='relu',
                       kernel_regularizer=reg)(x)
    out = layers.Dense(1, activation='sigmoid')(x)
    m   = keras.Model(inp, out, name=name)
    m.compile(optimizer=Adam(CFG['learning_rate']),
              loss='binary_crossentropy', metrics=['accuracy'])
    return m


print("\nParameter counts — Run 0 vs Run 1:")
shape_test = PREPARED['AAPL']['C2']['X_train'].shape[1:]
for builder, nm in [(build_cnn_lstm,  'CNN-LSTM'),
                    (build_cnn_only,  'CNN-only'),
                    (build_lstm_only, 'LSTM-only')]:
    m = builder(shape_test, nm)
    print(f"  {nm:<12} Run1={m.count_params():>8,} params")
del m
print("  (Run 0: CNN-LSTM=106,625  CNN-only=5,505  LSTM-only=78,913)")


# ── Utilities ─────────────────────────────────────────────────

def evaluate_model(y_true, y_prob, ticker, condition, model_name):
    y_pred = (y_prob >= 0.5).astype(int)
    acc    = accuracy_score(y_true, y_pred)
    try:
        auc = roc_auc_score(y_true, y_prob)
    except Exception:
        auc = 0.5
    f1  = f1_score(y_true, y_pred, average='weighted',
                   zero_division=0)
    ap  = average_precision_score(y_true, y_prob)
    cm  = confusion_matrix(y_true, y_pred)
    return {
        'ticker': ticker, 'condition': condition,
        'model': model_name,
        'accuracy':      round(acc, 4),
        'roc_auc':       round(auc, 4),
        'f1_weighted':   round(f1,  4),
        'avg_precision': round(ap,  4),
        'y_true': y_true, 'y_prob': y_prob,
        'y_pred': y_pred, 'conf_matrix': cm,
    }


def train_dl_model(model, arrays, ticker, model_name):
    X_tr, y_tr = arrays['X_train'], arrays['y_train']
    X_va, y_va = arrays['X_val'],   arrays['y_val']
    cw         = arrays['class_weights']
    cb = [
        callbacks.EarlyStopping(
            monitor='val_loss', patience=CFG['patience'],
            restore_best_weights=True, verbose=0),
        callbacks.ReduceLROnPlateau(
            monitor='val_loss', factor=0.5, patience=10,
            min_lr=1e-7, verbose=0),
    ]
    t0   = time.time()
    hist = model.fit(
        X_tr, y_tr,
        validation_data=(X_va, y_va),
        epochs=CFG['epochs'],
        batch_size=CFG['batch_size'],
        class_weight=cw,
        callbacks=cb, verbose=0)
    elapsed = time.time() - t0
    best_ep = int(np.argmin(hist.history['val_loss'])) + 1
    val_acc = hist.history['val_accuracy'][best_ep - 1]
    print(f"    [{ticker} {model_name}] "
          f"best_epoch={best_ep:3d}  "
          f"val_loss={min(hist.history['val_loss']):.4f}  "
          f"val_acc={val_acc:.4f}  "
          f"time={elapsed:.0f}s")
    return model, hist


def train_classical(arrays, ticker, condition):
    X_tr_f = arrays['X_train'].reshape(len(arrays['X_train']), -1)
    X_te_f = arrays['X_test'].reshape(len(arrays['X_test']),   -1)
    y_tr   = arrays['y_train']
    y_te   = arrays['y_test']
    n_neg  = (y_tr == 0).sum(); n_pos = (y_tr == 1).sum()
    scale_pos = n_neg / n_pos if n_pos > 0 else 1.0
    results, models = [], {}

    lr = LogisticRegression(
             max_iter=2000, C=0.1, class_weight='balanced',
             random_state=SEED, n_jobs=-1)
    lr.fit(X_tr_f, y_tr)
    prob = lr.predict_proba(X_te_f)[:, 1]
    results.append(evaluate_model(
        y_te, prob, ticker, condition, 'Logistic Regression'))
    models['Logistic Regression'] = lr

    rf = RandomForestClassifier(
             n_estimators=CFG['rf_trees'], max_depth=CFG['rf_depth'],
             class_weight='balanced', random_state=SEED, n_jobs=-1)
    rf.fit(X_tr_f, y_tr)
    prob = rf.predict_proba(X_te_f)[:, 1]
    results.append(evaluate_model(
        y_te, prob, ticker, condition, 'Random Forest'))
    models['Random Forest'] = rf

    xgb_clf = xgb.XGBClassifier(
                  n_estimators=CFG['xgb_trees'],
                  max_depth=CFG['xgb_depth'],
                  learning_rate=CFG['xgb_lr'],
                  scale_pos_weight=scale_pos,
                  eval_metric='logloss',
                  use_label_encoder=False,
                  random_state=SEED, verbosity=0)
    xgb_clf.fit(X_tr_f, y_tr)
    prob = xgb_clf.predict_proba(X_te_f)[:, 1]
    results.append(evaluate_model(
        y_te, prob, ticker, condition, 'XGBoost'))
    models['XGBoost'] = xgb_clf
    return results, models


print("Utilities defined.")


# ── Main training loop ────────────────────────────────────────
RESULTS_R1   = []
HISTORIES_R1 = {}

print("=" * 65)
print("RUN 1 TRAINING LOOP")
print("=" * 65)

for ticker in CFG['tickers']:
    print(f"\n{'─'*55}\n  TICKER: {ticker}\n{'─'*55}")
    for condition in CFG['conditions']:
        if condition == 'C3' and ticker not in CFG['tickers_c3']:
            print(f"\n  [{ticker} C3] SKIPPED")
            continue

        arrays    = PREPARED[ticker][condition]
        y_te      = arrays['y_test']
        n_feat    = arrays['X_train'].shape[2]
        inp_shape = (CFG['look_back'], n_feat)

        print(f"\n  Condition {condition} ({n_feat} features)")
        print("  Classical ML:")
        classical_results, _ = train_classical(
            arrays, ticker, condition)
        RESULTS_R1.extend(classical_results)
        for r in classical_results:
            print(f"    [{r['model']:<22}] "
                  f"acc={r['accuracy']:.4f}  "
                  f"auc={r['roc_auc']:.4f}")

        print("  Deep Learning:")
        for builder, model_name in [
            (build_cnn_only,  'CNN-only'),
            (build_lstm_only, 'LSTM-only'),
            (build_cnn_lstm,  'CNN-LSTM'),
        ]:
            tf.random.set_seed(SEED)
            np.random.seed(SEED)
            model = builder(inp_shape, model_name)
            model, hist = train_dl_model(
                model, arrays, ticker, model_name)
            y_prob = model.predict(
                arrays['X_test'], verbose=0).ravel()
            result = evaluate_model(
                y_te, y_prob, ticker, condition, model_name)
            RESULTS_R1.append(result)
            key = f"{ticker}_{condition}_{model_name}"
            HISTORIES_R1[key] = hist
            print(f"    [{model_name:<22}] "
                  f"acc={result['accuracy']:.4f}  "
                  f"auc={result['roc_auc']:.4f}")

    interim = pd.DataFrame([
        {k: v for k, v in r.items()
         if k not in ('y_true','y_prob','y_pred','conf_matrix')}
        for r in RESULTS_R1])
    interim.to_csv(
        os.path.join(SAVE_PATH, f'r1_interim_{ticker}.csv'),
        index=False)
    print(f"\n  Saved interim after {ticker}.")

print("\n" + "=" * 65)
print("Run 1 complete.")


# ── Results DataFrame ─────────────────────────────────────────
RESULTS_DF_R1 = pd.DataFrame([
    {k: v for k, v in r.items()
     if k not in ('y_true','y_prob','y_pred','conf_matrix')}
    for r in RESULTS_R1])

RESULTS_DF_R1.to_csv(
    os.path.join(SAVE_PATH, 'results_r1.csv'), index=False)
print("Saved: results_r1.csv")


# ── Run 0 vs Run 1 comparison ─────────────────────────────────
try:
    R0     = pd.read_csv(os.path.join(SAVE_PATH, 'results_df.csv'))
    has_r0 = True
    print("Run 0 results loaded for comparison.")
except FileNotFoundError:
    has_r0 = False
    print("Run 0 results not found — showing Run 1 only.")

print("\n" + "=" * 65)
print("RUN 0 vs RUN 1 — CNN-LSTM ACCURACY COMPARISON")
print("=" * 65)

for condition in CFG['conditions']:
    r1_sub = RESULTS_DF_R1[
        (RESULTS_DF_R1['condition'] == condition) &
        (RESULTS_DF_R1['model']     == 'CNN-LSTM')]
    print(f"\n  Condition {condition} — CNN-LSTM accuracy:")
    print(f"  {'Ticker':<8} {'Run0':>8} {'Run1':>8} {'Delta':>8}")
    print(f"  {'-'*34}")

    tickers = (CFG['tickers_c3']
               if condition == 'C3' else CFG['tickers'])
    for ticker in tickers:
        r1_acc = float(r1_sub[
            r1_sub['ticker']==ticker]['accuracy'].values[0]) \
            if len(r1_sub[r1_sub['ticker']==ticker]) > 0 else None

        if has_r0:
            r0_row = R0[(R0['ticker']    == ticker) &
                        (R0['condition'] == condition) &
                        (R0['model']     == 'CNN-LSTM')]
            r0_acc    = float(r0_row['accuracy'].values[0]) \
                        if len(r0_row) > 0 else None
            delta     = (r1_acc - r0_acc) \
                        if (r1_acc and r0_acc) else None
            delta_str = f"{delta:+.4f}" if delta else "N/A"
            r0_str    = f"{r0_acc:.4f}"  if r0_acc else "N/A"
        else:
            r0_str, delta_str = "N/A", "N/A"

        r1_str = f"{r1_acc:.4f}" if r1_acc else "N/A"
        print(f"  {ticker:<8} {r0_str:>8} {r1_str:>8} "
              f"{delta_str:>8}")

print(f"\n  Mean CNN-LSTM accuracy:")
for condition in CFG['conditions']:
    r1_mean = RESULTS_DF_R1[
        (RESULTS_DF_R1['condition'] == condition) &
        (RESULTS_DF_R1['model']     == 'CNN-LSTM')]['accuracy'].mean()
    if has_r0:
        r0_mean = R0[
            (R0['condition'] == condition) &
            (R0['model']     == 'CNN-LSTM')]['accuracy'].mean()
        print(f"  {condition}: Run0={r0_mean:.4f}  "
              f"Run1={r1_mean:.4f}  "
              f"Delta={r1_mean - r0_mean:+.4f}")
    else:
        print(f"  {condition}: Run1={r1_mean:.4f}")


# ── Training history comparison figure ───────────────────────
print("\nGenerating training history comparison figure...")

try:
    with open(os.path.join(SAVE_PATH, 'HISTORIES.pkl'), 'rb') as f:
        HISTORIES_R0 = pickle.load(f)
    has_r0_hist = True
except FileNotFoundError:
    has_r0_hist = False
    print("Run 0 histories not found — showing Run 1 only.")

tickers_plot = CFG['tickers']
model_name   = 'CNN-LSTM'
conditions_p = ['C1', 'C2']

fig, axes = plt.subplots(
    len(tickers_plot),
    len(conditions_p)*2 if has_r0_hist else len(conditions_p),
    figsize=(20 if has_r0_hist else 12, 3*len(tickers_plot)),
    facecolor=DARK)
fig.suptitle('Training History: Run 0 vs Run 1 — CNN-LSTM val_loss',
             color=TEXT, fontsize=12, fontweight='bold')

for row, ticker in enumerate(tickers_plot):
    col = 0
    for condition in conditions_p:
        key = f"{ticker}_{condition}_{model_name}"

        if has_r0_hist:
            ax0 = axes[row, col]
            ax0.set_facecolor(DARK)
            hist0 = HISTORIES_R0.get(key)
            if hist0:
                ep    = range(1, len(hist0.history['val_loss']) + 1)
                ax0.plot(ep, hist0.history['val_loss'],
                         color=RED, linewidth=1.2, label='Run0')
                best0 = np.argmin(hist0.history['val_loss']) + 1
                ax0.axvline(best0, color=RED,
                            linestyle=':', linewidth=0.8)
            ax0.set_title(f'{ticker} {condition} R0',
                          color=TEXT, fontsize=8)
            ax0.grid(True, alpha=0.3)
            ax0.tick_params(labelsize=6)
            col += 1

        ax1 = axes[row, col]
        ax1.set_facecolor(DARK)
        hist1 = HISTORIES_R1.get(key)
        if hist1:
            ep    = range(1, len(hist1.history['val_loss']) + 1)
            ax1.plot(ep, hist1.history['val_loss'],
                     color=ACCENT, linewidth=1.2, label='Run1')
            best1 = np.argmin(hist1.history['val_loss']) + 1
            ax1.axvline(best1, color=ACCENT,
                        linestyle=':', linewidth=0.8)
            ax1.set_title(f'{ticker} {condition} R1 (ep={best1})',
                          color=TEXT, fontsize=8)
        ax1.grid(True, alpha=0.3)
        ax1.tick_params(labelsize=6)
        col += 1

plt.tight_layout()
plt.savefig('R1_vs_R0_training_history.png', dpi=150,
            bbox_inches='tight', facecolor=DARK)
plt.show()
print("Saved: R1_vs_R0_training_history.png")


# ── Best epoch summary ────────────────────────────────────────
print("\nBest epoch summary — Run 1:")
print(f"{'Model':<12} {'Ticker':<6} {'C1':>5} {'C2':>5} {'C3':>5}")
print("-" * 35)

for ticker in CFG['tickers']:
    row_str = f"{'CNN-LSTM':<12} {ticker:<6}"
    for condition in ['C1', 'C2', 'C3']:
        if condition == 'C3' and ticker not in CFG['tickers_c3']:
            row_str += f"{'---':>5}"
            continue
        key  = f"{ticker}_{condition}_CNN-LSTM"
        hist = HISTORIES_R1.get(key)
        if hist:
            best_ep = np.argmin(hist.history['val_loss']) + 1
            row_str += f"{best_ep:>5}"
        else:
            row_str += f"{'N/A':>5}"
    print(row_str)

print("\nIf best_epoch > 20 for most entries: Run 1 fix worked.")
print("If still 1-5: learning rate may need further reduction to 5e-5.")


# ── Full accuracy summary table ───────────────────────────────
print("\n" + "=" * 65)
print("RUN 1 — FULL ACCURACY TABLE")
print("=" * 65)

for condition in CFG['conditions']:
    sub = RESULTS_DF_R1[RESULTS_DF_R1['condition'] == condition]
    if sub.empty:
        continue
    pivot = sub.pivot_table(
        index='model', columns='ticker',
        values='accuracy', aggfunc='first')
    pivot['Mean'] = pivot.mean(axis=1)
    pivot = pivot.sort_values('Mean', ascending=False)
    print(f"\n  Condition {condition}:")
    print(pivot.round(4).to_string())

print("\n" + "=" * 65)
print("Summary: mean accuracy across all tickers and conditions")
print("=" * 65)
summary = RESULTS_DF_R1.groupby(['condition','model'])\
    ['accuracy'].mean().unstack('condition').round(4)
summary['Overall'] = summary.mean(axis=1)
print(summary.sort_values('Overall', ascending=False).to_string())


# ── Save ──────────────────────────────────────────────────────
with open(os.path.join(SAVE_PATH, 'RESULTS_R1.pkl'),  'wb') as f:
    pickle.dump(RESULTS_R1,   f, protocol=4)
with open(os.path.join(SAVE_PATH, 'HISTORIES_R1.pkl'), 'wb') as f:
    pickle.dump(HISTORIES_R1, f, protocol=4)

print(f"\nSaved: RESULTS_R1.pkl  HISTORIES_R1.pkl  results_r1.csv")
```

```python
# ============================================================
# Phase   : CRISP-DM — Modelling, Run 2
#
# Changes from Run 1:
#   1. Walk-forward validation — 4 expanding-window folds.
#      Each fold trains on progressively more data and validates
#      on the following year. This ensures val windows cover
#      multiple market regimes rather than one anomalous window.
#      Reference: Prado (2018) Advances in Financial ML, Ch.7
#
#   2. Gradient clipping (clipnorm=1.0).
#      Prevents exploding gradients on TSLA high-volatility data.
#      Reference: Pascanu et al. (2013) — On the difficulty of
#      training recurrent neural networks.
#
# Protocol:
#   Step A — Walk-forward phase: train on each fold, record
#     best_epoch per fold. Average across folds = optimal_epochs.
#   Step B — Final training: retrain on ALL pre-test data for
#     exactly optimal_epochs (no early stopping).
#     Reference: Bergmeir & Benitez (2012, Information Sciences)
#   Step C — Evaluate final model on test set.
#
# PREREQUISITE:
#   PREPARED, SCALERS, PREP_INFO in memory or reload from disk.
# ============================================================

import os, pickle, random, warnings, time
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats as scipy_stats

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, callbacks, regularizers
from tensorflow.keras.optimizers import Adam
from sklearn.metrics import (
    accuracy_score, roc_auc_score, f1_score,
    average_precision_score, confusion_matrix, roc_curve)
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
import xgboost as xgb

# ── Reproducibility ──────────────────────────────────────────
SEED = 42
os.environ['PYTHONHASHSEED']       = str(SEED)
os.environ['TF_DETERMINISTIC_OPS'] = '1'
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# ── Plot theme ───────────────────────────────────────────────
DARK   = '#0d1117'; BORDER = '#30363d'; TEXT = '#c9d1d9'
MUTED  = '#8b949e'; GRID   = '#21262d'
ACCENT = '#58a6ff'; GREEN  = '#3fb950'; RED    = '#f85149'
YELLOW = '#e3b341'; PURPLE = '#bc8cff'; ORANGE = '#f0883e'
COLORS = [ACCENT, GREEN, RED, YELLOW, PURPLE]

plt.rcParams.update({
    'figure.facecolor': DARK, 'axes.facecolor':  DARK,
    'axes.edgecolor':  BORDER, 'axes.labelcolor': TEXT,
    'xtick.color': MUTED, 'ytick.color': MUTED,
    'text.color': TEXT,   'grid.color':  GRID,
    'grid.linestyle': '--', 'grid.alpha': 0.5,
    'font.family': 'monospace',
    'legend.facecolor': '#161b22',
    'legend.edgecolor': BORDER,
    'figure.dpi': 120,
})

SAVE_PATH = '/content/'

def load_prepared(path=SAVE_PATH):
    with open(os.path.join(path, 'PREPARED.pkl'),  'rb') as f:
        p = pickle.load(f)
    with open(os.path.join(path, 'SCALERS.pkl'),   'rb') as f:
        s = pickle.load(f)
    with open(os.path.join(path, 'PREP_INFO.pkl'), 'rb') as f:
        i = pickle.load(f)
    print(f"Loaded from {path}")
    return p, s, i

try:
    _ = PREPARED
    print("PREPARED in memory.")
except NameError:
    PREPARED, SCALERS, PREP_INFO = load_prepared()

# ── Configuration — carries Run 1 forward ────────────────────
CFG = dict(
    tickers    = ['AAPL', 'AMZN', 'GOOG', 'MSFT', 'TSLA'],
    tickers_c3 = ['AAPL', 'AMZN', 'GOOG', 'TSLA'],
    conditions = ['C1', 'C2', 'C3'],
    look_back  = 20,
    learning_rate = 1e-4,
    lstm_units    = 64,
    cnn_filters   = 32,
    patience      = 20,
    cnn_kernel    = 3,
    dropout       = 0.3,
    l2_reg        = 1e-4,
    batch_size    = 32,
    epochs        = 150,
    clip_norm     = 1.0,
    n_folds       = 4,
    min_train_yrs = 2,
    rf_trees  = 200, rf_depth  = 8,
    xgb_trees = 200, xgb_depth = 4, xgb_lr = 0.05,
    save_path = SAVE_PATH,
    run_name  = 'run2',
)

print("Run 2 configuration:")
print(f"  lr={CFG['learning_rate']}  lstm={CFG['lstm_units']}  "
      f"filters={CFG['cnn_filters']}  patience={CFG['patience']}")
print(f"  clip_norm={CFG['clip_norm']}  n_folds={CFG['n_folds']}")


# ── Walk-forward fold constructor ─────────────────────────────
# Creates 4 expanding-window folds from the pre-test training data.
# Each fold adds one more year to the training set and validates
# on the following year. This ensures early stopping decisions are
# made on val windows covering 2017, 2018, 2019, and 2020 —
# four distinct market regimes — rather than a single COVID window.

def make_wf_folds(X_all, y_all, n_folds=4, min_train=504):
    """
    Build n_folds expanding walk-forward folds.

    Parameters
    ----------
    X_all     : scaled feature array (n_total, look_back, n_feat)
    y_all     : label array (n_total,)
    n_folds   : number of folds
    min_train : minimum training samples in fold 1

    Returns
    -------
    List of (X_train, y_train, X_val, y_val) tuples.
    """
    n    = len(X_all)
    step = (n - min_train) // n_folds
    folds = []
    for f in range(n_folds):
        train_end = min_train + f * step
        val_start = train_end
        val_end   = min(train_end + step, n)
        if val_end <= val_start:
            continue
        folds.append((
            X_all[:train_end], y_all[:train_end],
            X_all[val_start:val_end], y_all[val_start:val_end],
        ))
    return folds


# ── Model builders — identical to Run 1, clipnorm added ───────

def build_cnn_lstm(input_shape, name='CNN-LSTM'):
    reg = regularizers.l2(CFG['l2_reg'])
    inp = layers.Input(shape=input_shape)
    x   = layers.Conv1D(CFG['cnn_filters'], CFG['cnn_kernel'],
                        padding='causal', activation='relu',
                        kernel_regularizer=reg)(inp)
    x   = layers.MaxPooling1D(pool_size=2)(x)
    x   = layers.Dropout(CFG['dropout'])(x)
    x   = layers.LSTM(CFG['lstm_units'],
                      kernel_regularizer=reg,
                      recurrent_regularizer=reg)(x)
    x   = layers.BatchNormalization()(x)
    x   = layers.Dropout(CFG['dropout'])(x)
    x   = layers.Dense(32, activation='relu',
                       kernel_regularizer=reg)(x)
    out = layers.Dense(1, activation='sigmoid')(x)
    m   = keras.Model(inp, out, name=name)
    m.compile(
        optimizer=Adam(CFG['learning_rate'],
                       clipnorm=CFG['clip_norm']),
        loss='binary_crossentropy', metrics=['accuracy'])
    return m


def build_cnn_only(input_shape, name='CNN-only'):
    reg = regularizers.l2(CFG['l2_reg'])
    inp = layers.Input(shape=input_shape)
    x   = layers.Conv1D(CFG['cnn_filters'], CFG['cnn_kernel'],
                        padding='causal', activation='relu',
                        kernel_regularizer=reg)(inp)
    x   = layers.Conv1D(CFG['cnn_filters'], CFG['cnn_kernel'],
                        padding='causal', activation='relu',
                        kernel_regularizer=reg)(x)
    x   = layers.GlobalAveragePooling1D()(x)
    x   = layers.BatchNormalization()(x)
    x   = layers.Dropout(CFG['dropout'])(x)
    x   = layers.Dense(32, activation='relu',
                       kernel_regularizer=reg)(x)
    out = layers.Dense(1, activation='sigmoid')(x)
    m   = keras.Model(inp, out, name=name)
    m.compile(
        optimizer=Adam(CFG['learning_rate'],
                       clipnorm=CFG['clip_norm']),
        loss='binary_crossentropy', metrics=['accuracy'])
    return m


def build_lstm_only(input_shape, name='LSTM-only'):
    reg = regularizers.l2(CFG['l2_reg'])
    inp = layers.Input(shape=input_shape)
    x   = layers.LSTM(CFG['lstm_units'],
                      kernel_regularizer=reg,
                      recurrent_regularizer=reg)(inp)
    x   = layers.BatchNormalization()(x)
    x   = layers.Dropout(CFG['dropout'])(x)
    x   = layers.Dense(32, activation='relu',
                       kernel_regularizer=reg)(x)
    out = layers.Dense(1, activation='sigmoid')(x)
    m   = keras.Model(inp, out, name=name)
    m.compile(
        optimizer=Adam(CFG['learning_rate'],
                       clipnorm=CFG['clip_norm']),
        loss='binary_crossentropy', metrics=['accuracy'])
    return m

BUILDERS = {
    'CNN-only':  build_cnn_only,
    'LSTM-only': build_lstm_only,
    'CNN-LSTM':  build_cnn_lstm,
}

print("Model builders defined with clipnorm=1.0.")


# ── Core walk-forward training function ──────────────────────

def train_wf_model(builder, input_shape, arrays,
                   ticker, model_name, condition):
    """
    Walk-forward training protocol.

    Step A: train on each fold, record best_epoch.
    Step B: retrain on full pre-test data for mean(best_epochs)+2.
            The +2 buffer accounts for the larger final training
            set having slightly more learning capacity.

    Returns
    -------
    final_model : Keras model trained on full pre-test data
    fold_info   : list of per-fold diagnostics
    optimal_ep  : epochs used for final training
    fold_losses : val_loss per epoch per fold (for plotting)
    """
    X_all = np.concatenate([
        arrays['X_train'], arrays['X_val']], axis=0)
    y_all = np.concatenate([
        arrays['y_train'], arrays['y_val']], axis=0)

    n_pos = y_all.sum()
    n_neg = len(y_all) - n_pos
    cw    = {0: len(y_all)/(2*n_neg) if n_neg > 0 else 1.0,
             1: len(y_all)/(2*n_pos) if n_pos > 0 else 1.0}

    folds = make_wf_folds(
        X_all, y_all,
        n_folds=CFG['n_folds'],
        min_train=max(200, len(X_all)//(CFG['n_folds']+1)))

    fold_info   = []
    fold_losses = []
    best_epochs = []

    for fold_idx, (X_tr, y_tr, X_va, y_va) in enumerate(folds):
        tf.random.set_seed(SEED + fold_idx)
        np.random.seed(SEED + fold_idx)
        model = builder(input_shape, model_name)
        cb = [
            callbacks.EarlyStopping(
                monitor='val_loss', patience=CFG['patience'],
                restore_best_weights=True, verbose=0),
            callbacks.ReduceLROnPlateau(
                monitor='val_loss', factor=0.5, patience=10,
                min_lr=1e-7, verbose=0),
        ]
        hist = model.fit(
            X_tr, y_tr,
            validation_data=(X_va, y_va),
            epochs=CFG['epochs'],
            batch_size=CFG['batch_size'],
            class_weight=cw,
            callbacks=cb, verbose=0)
        best_ep   = int(np.argmin(hist.history['val_loss'])) + 1
        best_loss = min(hist.history['val_loss'])
        best_acc  = hist.history['val_accuracy'][best_ep - 1]
        best_epochs.append(best_ep)
        fold_losses.append(hist.history['val_loss'])
        fold_info.append({
            'fold':          fold_idx + 1,
            'train_size':    len(X_tr),
            'val_size':      len(X_va),
            'best_epoch':    best_ep,
            'best_val_loss': round(best_loss, 4),
            'best_val_acc':  round(best_acc,  4),
        })
        del model

    optimal_ep = int(np.ceil(np.mean(best_epochs))) + 2
    optimal_ep = min(optimal_ep, CFG['epochs'])

    print(f"    [{ticker} {condition} {model_name}] "
          f"fold best_epochs={best_epochs}  "
          f"optimal={optimal_ep}")

    tf.random.set_seed(SEED)
    np.random.seed(SEED)
    final_model = builder(input_shape, model_name)
    final_model.fit(
        X_all, y_all,
        epochs=optimal_ep,
        batch_size=CFG['batch_size'],
        class_weight=cw, verbose=0)

    return final_model, fold_info, optimal_ep, fold_losses


# ── Utilities ─────────────────────────────────────────────────

def evaluate_model(y_true, y_prob, ticker, condition, model_name):
    y_pred = (y_prob >= 0.5).astype(int)
    acc    = accuracy_score(y_true, y_pred)
    try:
        auc = roc_auc_score(y_true, y_prob)
    except Exception:
        auc = 0.5
    f1  = f1_score(y_true, y_pred,
                   average='weighted', zero_division=0)
    ap  = average_precision_score(y_true, y_prob)
    cm  = confusion_matrix(y_true, y_pred)
    return {
        'ticker': ticker, 'condition': condition,
        'model': model_name,
        'accuracy':      round(acc, 4),
        'roc_auc':       round(auc, 4),
        'f1_weighted':   round(f1,  4),
        'avg_precision': round(ap,  4),
        'y_true': y_true, 'y_prob': y_prob,
        'y_pred': y_pred, 'conf_matrix': cm,
    }


def train_classical(arrays, ticker, condition):
    X_tr_f  = arrays['X_train'].reshape(len(arrays['X_train']), -1)
    X_te_f  = arrays['X_test'].reshape(len(arrays['X_test']),   -1)
    X_all_f = np.concatenate([
        X_tr_f,
        arrays['X_val'].reshape(len(arrays['X_val']), -1)])
    y_all     = np.concatenate([arrays['y_train'], arrays['y_val']])
    y_te      = arrays['y_test']
    n_neg     = (y_all == 0).sum(); n_pos = (y_all == 1).sum()
    scale_pos = n_neg / n_pos if n_pos > 0 else 1.0
    results, models = [], {}

    lr_clf = LogisticRegression(
        max_iter=2000, C=0.1, class_weight='balanced',
        random_state=SEED, n_jobs=-1)
    lr_clf.fit(X_all_f, y_all)
    prob = lr_clf.predict_proba(X_te_f)[:, 1]
    results.append(evaluate_model(
        y_te, prob, ticker, condition, 'Logistic Regression'))
    models['Logistic Regression'] = lr_clf

    rf = RandomForestClassifier(
        n_estimators=CFG['rf_trees'], max_depth=CFG['rf_depth'],
        class_weight='balanced', random_state=SEED, n_jobs=-1)
    rf.fit(X_all_f, y_all)
    prob = rf.predict_proba(X_te_f)[:, 1]
    results.append(evaluate_model(
        y_te, prob, ticker, condition, 'Random Forest'))
    models['Random Forest'] = rf

    xgb_clf = xgb.XGBClassifier(
        n_estimators=CFG['xgb_trees'], max_depth=CFG['xgb_depth'],
        learning_rate=CFG['xgb_lr'], scale_pos_weight=scale_pos,
        eval_metric='logloss', use_label_encoder=False,
        random_state=SEED, verbosity=0)
    xgb_clf.fit(X_all_f, y_all)
    prob = xgb_clf.predict_proba(X_te_f)[:, 1]
    results.append(evaluate_model(
        y_te, prob, ticker, condition, 'XGBoost'))
    models['XGBoost'] = xgb_clf

    return results, models


print("Utilities defined.")


# ── Main training loop ────────────────────────────────────────
RESULTS_R2   = []
FOLD_INFO_R2 = {}
WF_LOSSES_R2 = {}
OPT_EPOCHS   = {}

print("=" * 65)
print("RUN 2 — WALK-FORWARD TRAINING LOOP")
print("=" * 65)
print(f"Protocol: {CFG['n_folds']}-fold walk-forward + final retrain")
print()

t_total = time.time()

for ticker in CFG['tickers']:
    print(f"\n{'─'*55}\n  TICKER: {ticker}\n{'─'*55}")

    for condition in CFG['conditions']:
        if condition == 'C3' and ticker not in CFG['tickers_c3']:
            print(f"\n  [{ticker} C3] SKIPPED")
            continue

        arrays    = PREPARED[ticker][condition]
        y_te      = arrays['y_test']
        n_feat    = arrays['X_train'].shape[2]
        inp_shape = (CFG['look_back'], n_feat)

        print(f"\n  Condition {condition} ({n_feat} features, "
              f"{len(y_te)} test sequences)")

        print("  Classical ML (trained on full pre-test data):")
        classical_results, _ = train_classical(
            arrays, ticker, condition)
        RESULTS_R2.extend(classical_results)
        for r in classical_results:
            print(f"    [{r['model']:<22}] "
                  f"acc={r['accuracy']:.4f}  "
                  f"auc={r['roc_auc']:.4f}")

        print("  Deep Learning (walk-forward):")
        for model_name, builder in BUILDERS.items():
            tf.random.set_seed(SEED)
            np.random.seed(SEED)
            t0 = time.time()
            final_model, fold_info, opt_ep, fold_losses = \
                train_wf_model(builder, inp_shape, arrays,
                               ticker, model_name, condition)
            elapsed = time.time() - t0
            y_prob  = final_model.predict(
                arrays['X_test'], verbose=0).ravel()
            result  = evaluate_model(
                y_te, y_prob, ticker, condition, model_name)
            RESULTS_R2.append(result)
            key = f"{ticker}_{condition}_{model_name}"
            FOLD_INFO_R2[key] = fold_info
            WF_LOSSES_R2[key] = fold_losses
            OPT_EPOCHS[key]   = opt_ep
            print(f"    [{model_name:<12}] "
                  f"opt_ep={opt_ep:3d}  "
                  f"acc={result['accuracy']:.4f}  "
                  f"auc={result['roc_auc']:.4f}  "
                  f"time={elapsed:.0f}s")

    interim = pd.DataFrame([
        {k: v for k, v in r.items()
         if k not in ('y_true','y_prob','y_pred','conf_matrix')}
        for r in RESULTS_R2])
    interim.to_csv(
        os.path.join(SAVE_PATH, f'r2_interim_{ticker}.csv'),
        index=False)
    print(f"\n  Saved interim after {ticker}.")

total_time = time.time() - t_total
print(f"\n{'='*65}")
print(f"Run 2 complete. Total time: {total_time/3600:.1f} hours")
print(f"{'='*65}")


# ── Results DataFrame and save ────────────────────────────────
RESULTS_DF_R2 = pd.DataFrame([
    {k: v for k, v in r.items()
     if k not in ('y_true','y_prob','y_pred','conf_matrix')}
    for r in RESULTS_R2])

RESULTS_DF_R2.to_csv(
    os.path.join(SAVE_PATH, 'results_r2.csv'), index=False)

with open(os.path.join(SAVE_PATH, 'RESULTS_R2.pkl'),   'wb') as f:
    pickle.dump(RESULTS_R2,   f, protocol=4)
with open(os.path.join(SAVE_PATH, 'FOLD_INFO_R2.pkl'), 'wb') as f:
    pickle.dump(FOLD_INFO_R2, f, protocol=4)
with open(os.path.join(SAVE_PATH, 'WF_LOSSES_R2.pkl'), 'wb') as f:
    pickle.dump(WF_LOSSES_R2, f, protocol=4)

print("Saved: results_r2.csv, RESULTS_R2.pkl, "
      "FOLD_INFO_R2.pkl, WF_LOSSES_R2.pkl")


# ── Three-run comparison ──────────────────────────────────────
print("\n" + "=" * 65)
print("THREE-RUN COMPARISON — CNN-LSTM ACCURACY")
print("=" * 65)

try:
    R0 = pd.read_csv(os.path.join(SAVE_PATH, 'results_df.csv'))
    R1 = pd.read_csv(os.path.join(SAVE_PATH, 'results_r1.csv'))
    has_prev = True
except FileNotFoundError:
    has_prev = False
    print("Previous run CSVs not found — showing Run 2 only.")

def get_acc(df, ticker, condition, model='CNN-LSTM'):
    row = df[(df['ticker']    == ticker) &
             (df['condition'] == condition) &
             (df['model']     == model)]
    return float(row['accuracy'].values[0]) if len(row) > 0 else None

def fmt(val):
    return f"{val:.4f}" if val is not None else "N/A"

for condition in CFG['conditions']:
    tickers = (CFG['tickers_c3']
               if condition == 'C3' else CFG['tickers'])
    r2_sub  = RESULTS_DF_R2[
        (RESULTS_DF_R2['condition'] == condition) &
        (RESULTS_DF_R2['model']     == 'CNN-LSTM')]

    print(f"\n  Condition {condition}:")
    if has_prev:
        print(f"  {'Ticker':<8} {'Run0':>8} {'Run1':>8} "
              f"{'Run2':>8} {'R0->R2':>8}")
        print(f"  {'-'*42}")
    else:
        print(f"  {'Ticker':<8} {'Run2':>8}")

    for ticker in tickers:
        r2_row = r2_sub[r2_sub['ticker'] == ticker]
        r2_acc = float(r2_row['accuracy'].values[0]) \
                 if len(r2_row) > 0 else None
        if has_prev:
            r0a   = get_acc(R0, ticker, condition)
            r1a   = get_acc(R1, ticker, condition)
            delta = (r2_acc - r0a) \
                    if (r2_acc is not None and r0a is not None) \
                    else None
            delta_str = f"{delta:+.4f}" if delta is not None \
                        else "N/A"
            print(f"  {ticker:<8} {fmt(r0a):>8} {fmt(r1a):>8} "
                  f"{fmt(r2_acc):>8} {delta_str:>8}")
        else:
            print(f"  {ticker:<8} {fmt(r2_acc):>8}")

print(f"\n  Mean CNN-LSTM accuracy:")
if has_prev:
    print(f"  {'Cond':<6} {'Run0':>8} {'Run1':>8} "
          f"{'Run2':>8} {'R0->R2':>8}")
    print(f"  {'-'*38}")

for cond in CFG['conditions']:
    r2m = RESULTS_DF_R2[
        (RESULTS_DF_R2['condition'] == cond) &
        (RESULTS_DF_R2['model']     == 'CNN-LSTM')]['accuracy'].mean()
    if has_prev:
        r0m = R0[(R0['condition'] == cond) &
                 (R0['model']     == 'CNN-LSTM')]['accuracy'].mean()
        r1m = R1[(R1['condition'] == cond) &
                 (R1['model']     == 'CNN-LSTM')]['accuracy'].mean()
        print(f"  {cond:<6} {r0m:.4f}  {r1m:.4f}  "
              f"{r2m:.4f}  {r2m - r0m:+.4f}")
    else:
        print(f"  {cond:<6} Run2={r2m:.4f}")


# ── Optimal epochs table ──────────────────────────────────────
print("\n" + "=" * 65)
print("OPTIMAL EPOCHS — WALK-FORWARD RESULTS")
print("=" * 65)
print(f"{'Ticker':<8} {'C1':>6} {'C2':>6} {'C3':>6}")
print("-" * 28)

for ticker in CFG['tickers']:
    row = f"{ticker:<8}"
    for cond in ['C1', 'C2', 'C3']:
        if cond == 'C3' and ticker not in CFG['tickers_c3']:
            row += f"{'---':>6}"
            continue
        key = f"{ticker}_{cond}_CNN-LSTM"
        ep  = OPT_EPOCHS.get(key, None)
        row += f"{ep:>6}" if ep else f"{'N/A':>6}"
    print(row)

print("\nFold details (CNN-LSTM, AAPL, C2):")
for fi in FOLD_INFO_R2.get('AAPL_C2_CNN-LSTM', []):
    print(f"  Fold {fi['fold']}: "
          f"train={fi['train_size']}  val={fi['val_size']}  "
          f"best_ep={fi['best_epoch']}  "
          f"val_loss={fi['best_val_loss']:.4f}  "
          f"val_acc={fi['best_val_acc']:.4f}")


# ── Walk-forward validation curves figure ────────────────────
print("\nGenerating walk-forward validation curves figure...")

fig, axes = plt.subplots(
    len(CFG['tickers']), CFG['n_folds'],
    figsize=(4*CFG['n_folds'], 3*len(CFG['tickers'])),
    facecolor=DARK, sharey='row')
fig.suptitle(
    'Walk-Forward Validation Curves — CNN-LSTM, Condition C2\n'
    'Each panel = one fold. Vertical line = best epoch.',
    color=TEXT, fontsize=12, fontweight='bold')

fold_colors = [ACCENT, GREEN, YELLOW, ORANGE]

for row, ticker in enumerate(CFG['tickers']):
    key    = f"{ticker}_C2_CNN-LSTM"
    losses = WF_LOSSES_R2.get(key, [])
    finfos = FOLD_INFO_R2.get(key, [])
    opt_ep = OPT_EPOCHS.get(key, None)

    for fold_idx in range(CFG['n_folds']):
        ax = axes[row, fold_idx]
        ax.set_facecolor(DARK)
        if fold_idx < len(losses):
            ep = range(1, len(losses[fold_idx]) + 1)
            ax.plot(ep, losses[fold_idx],
                    color=fold_colors[fold_idx], linewidth=1.2)
            if fold_idx < len(finfos):
                best = finfos[fold_idx]['best_epoch']
                ax.axvline(best,
                           color=fold_colors[fold_idx],
                           linestyle=':', linewidth=1.0)
        if fold_idx == 0:
            ax.set_ylabel(ticker, color=COLORS[row], fontsize=9)
        if row == 0:
            ax.set_title(f'Fold {fold_idx+1}',
                         color=TEXT, fontsize=9)
        if opt_ep and fold_idx == CFG['n_folds'] - 1:
            ax.text(0.98, 0.95, f'opt_ep={opt_ep}',
                    transform=ax.transAxes,
                    ha='right', va='top',
                    fontsize=7, color=MUTED)
        ax.grid(True, alpha=0.3)
        ax.tick_params(labelsize=6)

plt.tight_layout()
plt.savefig('R2_wf_curves.png', dpi=150,
            bbox_inches='tight', facecolor=DARK)
plt.show()
print("Saved: R2_wf_curves.png")


# ── Full results summary ──────────────────────────────────────
print("\n" + "=" * 65)
print("RUN 2 — FULL RESULTS SUMMARY")
print("=" * 65)

for condition in CFG['conditions']:
    sub = RESULTS_DF_R2[RESULTS_DF_R2['condition'] == condition]
    if sub.empty:
        continue
    pivot = sub.pivot_table(
        index='model', columns='ticker',
        values='accuracy', aggfunc='first')
    pivot['Mean'] = pivot.mean(axis=1)
    pivot = pivot.sort_values('Mean', ascending=False)
    print(f"\n  Condition {condition} — Accuracy:")
    print(pivot.round(4).to_string())

print("\n" + "=" * 65)
print("Mean accuracy summary — all conditions")
print("=" * 65)
smry = RESULTS_DF_R2.groupby(
    ['condition','model'])['accuracy'].mean()\
    .unstack('condition').round(4)
smry['Overall'] = smry.mean(axis=1)
print(smry.sort_values('Overall', ascending=False).to_string())

print("\nBenchmark comparison (CNN-LSTM mean per condition):")
print(f"  {'Model/Study':<35} {'C1':>8} {'C2':>8} {'C3':>8}")
print(f"  {'-'*63}")
print(f"  {'Fischer & Krauss (2018) LSTM':<35} "
      f"{'---':>8} {'53.20%':>8} {'---':>8}")
print(f"  {'Lu et al. (2020) CNN-LSTM':<35} "
      f"{'---':>8} {'55.90%':>8} {'---':>8}")
for cond in ['C1', 'C2', 'C3']:
    val = RESULTS_DF_R2[
        (RESULTS_DF_R2['model']     == 'CNN-LSTM') &
        (RESULTS_DF_R2['condition'] == cond)]['accuracy'].mean()
    row_label = f"  This study CNN-LSTM Run2 ({cond})"
    print(f"  {row_label:<35} {val*100:>7.2f}%")


# ── Confusion matrices ────────────────────────────────────────
from matplotlib.colors import LinearSegmentedColormap
cmap_dark = LinearSegmentedColormap.from_list(
    'dark_blue', [DARK, ACCENT])

for condition in CFG['conditions']:
    tickers_this = (CFG['tickers_c3']
                    if condition == 'C3' else CFG['tickers'])
    n   = len(tickers_this)
    fig, axes = plt.subplots(
        1, n, figsize=(3.5*n, 4), facecolor=DARK)
    if n == 1:
        axes = [axes]
    fig.suptitle(
        f'Confusion Matrices — CNN-LSTM Run 2 | Condition {condition}',
        color=TEXT, fontsize=12, fontweight='bold')

    for idx, ticker in enumerate(tickers_this):
        r = next(
            (x for x in RESULTS_R2
             if x['ticker']    == ticker
             and x['condition'] == condition
             and x['model']     == 'CNN-LSTM'), None)
        if r is None:
            continue
        cm  = r['conf_matrix']
        ax  = axes[idx]
        ax.set_facecolor(DARK)
        ax.imshow(cm, cmap=cmap_dark)
        ax.set_title(f"{ticker}\nacc={r['accuracy']:.4f}",
                     color=TEXT, fontsize=10)
        ax.set_xticks([0, 1]); ax.set_yticks([0, 1])
        ax.set_xticklabels(['DOWN', 'UP'], fontsize=9)
        ax.set_yticklabels(['DOWN', 'UP'], fontsize=9)
        if idx == 0:
            ax.set_ylabel('Actual', fontsize=9)
        ax.set_xlabel('Predicted', fontsize=9)
        for i in range(2):
            for j in range(2):
                ax.text(j, i, str(cm[i, j]),
                        ha='center', va='center',
                        fontsize=13, fontweight='bold',
                        color=TEXT)

    plt.tight_layout()
    fname = f'R2_confusion_{condition}.png'
    plt.savefig(fname, dpi=150,
                bbox_inches='tight', facecolor=DARK)
    plt.show()
    print(f"Saved: {fname}")

print("\nRun 2 complete.")
```

```python
# ============================================================

# Phase   : CRISP-DM — Modelling, Run 3
#
# Change from Run 2:
#   optimal_epochs = mean(fold best_epochs[0:3]) + 2
#   Fold 4 is trained but EXCLUDED from the epoch vote.
#
# SCIENTIFIC JUSTIFICATION:
#   Fold 4 validation window covers approximately 2020 —
#   the COVID crash and recovery. Walk-forward curves showed
#   Fold 4 val_loss rising monotonically from epoch 1 for most
#   tickers, dragging the mean best_epoch down to 4-11 and
#   preventing the model from reaching its genuine optimum.
#
#   The COVID period is a documented structural break in
#   financial time series (Mishkin, 2020). Using it as a
#   validation signal for models tested on the 2022 rate-hike
#   regime is methodologically unsound — neither regime is
#   representative of the other.
#
#   Fold 4 is retained in the fold training runs but excluded
#   from the epoch selection vote. This is analogous to the
#   purged and embargoed cross-validation approach of
#   Prado (2018, Ch.7) where structurally anomalous periods
#   are excluded from the validation signal.
#
# PREREQUISITE:
#   PREPARED, SCALERS, PREP_INFO in memory or reload from disk.
# ============================================================

import os, pickle, random, warnings, time
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats as scipy_stats

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, callbacks, regularizers
from tensorflow.keras.optimizers import Adam
from sklearn.metrics import (
    accuracy_score, roc_auc_score, f1_score,
    average_precision_score, confusion_matrix, roc_curve)
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
import xgboost as xgb

# ── Reproducibility ──────────────────────────────────────────
SEED = 42
os.environ['PYTHONHASHSEED']       = str(SEED)
os.environ['TF_DETERMINISTIC_OPS'] = '1'
random.seed(SEED); np.random.seed(SEED); tf.random.set_seed(SEED)

# ── Plot theme ───────────────────────────────────────────────
DARK   = '#0d1117'; BORDER = '#30363d'; TEXT = '#c9d1d9'
MUTED  = '#8b949e'; GRID   = '#21262d'
ACCENT = '#58a6ff'; GREEN  = '#3fb950'; RED    = '#f85149'
YELLOW = '#e3b341'; PURPLE = '#bc8cff'; ORANGE = '#f0883e'
COLORS = [ACCENT, GREEN, RED, YELLOW, PURPLE]

plt.rcParams.update({
    'figure.facecolor': DARK, 'axes.facecolor':  DARK,
    'axes.edgecolor':  BORDER, 'axes.labelcolor': TEXT,
    'xtick.color': MUTED, 'ytick.color': MUTED,
    'text.color': TEXT,   'grid.color':  GRID,
    'grid.linestyle': '--', 'grid.alpha': 0.5,
    'font.family': 'monospace',
    'legend.facecolor': '#161b22',
    'legend.edgecolor': BORDER, 'figure.dpi': 120,
})

SAVE_PATH = '/content/'

def load_prepared(path=SAVE_PATH):
    with open(os.path.join(path, 'PREPARED.pkl'),  'rb') as f:
        p = pickle.load(f)
    with open(os.path.join(path, 'SCALERS.pkl'),   'rb') as f:
        s = pickle.load(f)
    with open(os.path.join(path, 'PREP_INFO.pkl'), 'rb') as f:
        i = pickle.load(f)
    print(f"Loaded from {path}")
    return p, s, i

try:
    _ = PREPARED
    print("PREPARED in memory.")
except NameError:
    PREPARED, SCALERS, PREP_INFO = load_prepared()

# ── Configuration — identical to Run 2, one addition ─────────
CFG = dict(
    tickers    = ['AAPL', 'AMZN', 'GOOG', 'MSFT', 'TSLA'],
    tickers_c3 = ['AAPL', 'AMZN', 'GOOG', 'TSLA'],
    conditions = ['C1', 'C2', 'C3'],
    look_back  = 20,
    learning_rate    = 1e-4,
    lstm_units       = 64,
    cnn_filters      = 32,
    patience         = 20,
    cnn_kernel       = 3,
    dropout          = 0.3,
    l2_reg           = 1e-4,
    batch_size       = 32,
    epochs           = 150,
    clip_norm        = 1.0,
    n_folds          = 4,
    epoch_vote_folds = [0, 1, 2],   # Fold 4 (index 3) excluded
    rf_trees=200, rf_depth=8,
    xgb_trees=200, xgb_depth=4, xgb_lr=0.05,
    save_path=SAVE_PATH,
    run_name='run3',
)

print("Run 3 configuration:")
print(f"  All Run 2 settings carried forward.")
print(f"  epoch_vote_folds={CFG['epoch_vote_folds']}")
print(f"  Fold 4 (COVID window) excluded from optimal epoch vote.")


# ── Walk-forward fold constructor — identical to Run 2 ────────

def make_wf_folds(X_all, y_all, n_folds=4, min_train=504):
    n     = len(X_all)
    step  = (n - min_train) // n_folds
    folds = []
    for f in range(n_folds):
        train_end = min_train + f * step
        val_start = train_end
        val_end   = min(train_end + step, n)
        if val_end <= val_start:
            continue
        folds.append((
            X_all[:train_end], y_all[:train_end],
            X_all[val_start:val_end], y_all[val_start:val_end],
        ))
    return folds


# ── Model builders — identical to Run 2 ──────────────────────

def build_cnn_lstm(input_shape, name='CNN-LSTM'):
    reg = regularizers.l2(CFG['l2_reg'])
    inp = layers.Input(shape=input_shape)
    x   = layers.Conv1D(CFG['cnn_filters'], CFG['cnn_kernel'],
                        padding='causal', activation='relu',
                        kernel_regularizer=reg)(inp)
    x   = layers.MaxPooling1D(pool_size=2)(x)
    x   = layers.Dropout(CFG['dropout'])(x)
    x   = layers.LSTM(CFG['lstm_units'],
                      kernel_regularizer=reg,
                      recurrent_regularizer=reg)(x)
    x   = layers.BatchNormalization()(x)
    x   = layers.Dropout(CFG['dropout'])(x)
    x   = layers.Dense(32, activation='relu',
                       kernel_regularizer=reg)(x)
    out = layers.Dense(1, activation='sigmoid')(x)
    m   = keras.Model(inp, out, name=name)
    m.compile(optimizer=Adam(CFG['learning_rate'],
                             clipnorm=CFG['clip_norm']),
              loss='binary_crossentropy', metrics=['accuracy'])
    return m


def build_cnn_only(input_shape, name='CNN-only'):
    reg = regularizers.l2(CFG['l2_reg'])
    inp = layers.Input(shape=input_shape)
    x   = layers.Conv1D(CFG['cnn_filters'], CFG['cnn_kernel'],
                        padding='causal', activation='relu',
                        kernel_regularizer=reg)(inp)
    x   = layers.Conv1D(CFG['cnn_filters'], CFG['cnn_kernel'],
                        padding='causal', activation='relu',
                        kernel_regularizer=reg)(x)
    x   = layers.GlobalAveragePooling1D()(x)
    x   = layers.BatchNormalization()(x)
    x   = layers.Dropout(CFG['dropout'])(x)
    x   = layers.Dense(32, activation='relu',
                       kernel_regularizer=reg)(x)
    out = layers.Dense(1, activation='sigmoid')(x)
    m   = keras.Model(inp, out, name=name)
    m.compile(optimizer=Adam(CFG['learning_rate'],
                             clipnorm=CFG['clip_norm']),
              loss='binary_crossentropy', metrics=['accuracy'])
    return m


def build_lstm_only(input_shape, name='LSTM-only'):
    reg = regularizers.l2(CFG['l2_reg'])
    inp = layers.Input(shape=input_shape)
    x   = layers.LSTM(CFG['lstm_units'],
                      kernel_regularizer=reg,
                      recurrent_regularizer=reg)(inp)
    x   = layers.BatchNormalization()(x)
    x   = layers.Dropout(CFG['dropout'])(x)
    x   = layers.Dense(32, activation='relu',
                       kernel_regularizer=reg)(x)
    out = layers.Dense(1, activation='sigmoid')(x)
    m   = keras.Model(inp, out, name=name)
    m.compile(optimizer=Adam(CFG['learning_rate'],
                             clipnorm=CFG['clip_norm']),
              loss='binary_crossentropy', metrics=['accuracy'])
    return m

BUILDERS = {
    'CNN-only':  build_cnn_only,
    'LSTM-only': build_lstm_only,
    'CNN-LSTM':  build_cnn_lstm,
}
print("Model builders defined.")


# ── Core walk-forward training ────────────────────────────────
# One change from Run 2: optimal_epoch vote uses only
# CFG['epoch_vote_folds'] = [0, 1, 2], excluding Fold 4.

def train_wf_model(builder, input_shape, arrays,
                   ticker, model_name, condition):
    X_all = np.concatenate([
        arrays['X_train'], arrays['X_val']], axis=0)
    y_all = np.concatenate([
        arrays['y_train'], arrays['y_val']], axis=0)
    n_pos = y_all.sum()
    n_neg = len(y_all) - n_pos
    cw    = {0: len(y_all)/(2*n_neg) if n_neg > 0 else 1.0,
             1: len(y_all)/(2*n_pos) if n_pos > 0 else 1.0}

    folds = make_wf_folds(
        X_all, y_all,
        n_folds=CFG['n_folds'],
        min_train=max(200, len(X_all)//(CFG['n_folds']+1)))

    fold_info   = []
    fold_losses = []
    best_epochs = []

    for fold_idx, (X_tr, y_tr, X_va, y_va) in enumerate(folds):
        tf.random.set_seed(SEED + fold_idx)
        np.random.seed(SEED + fold_idx)
        model = builder(input_shape, model_name)
        cb = [
            callbacks.EarlyStopping(
                monitor='val_loss', patience=CFG['patience'],
                restore_best_weights=True, verbose=0),
            callbacks.ReduceLROnPlateau(
                monitor='val_loss', factor=0.5, patience=10,
                min_lr=1e-7, verbose=0),
        ]
        hist = model.fit(
            X_tr, y_tr,
            validation_data=(X_va, y_va),
            epochs=CFG['epochs'],
            batch_size=CFG['batch_size'],
            class_weight=cw,
            callbacks=cb, verbose=0)
        best_ep   = int(np.argmin(hist.history['val_loss'])) + 1
        best_loss = min(hist.history['val_loss'])
        best_acc  = hist.history['val_accuracy'][best_ep - 1]
        best_epochs.append(best_ep)
        fold_losses.append(hist.history['val_loss'])
        fold_info.append({
            'fold':          fold_idx + 1,
            'train_size':    len(X_tr),
            'val_size':      len(X_va),
            'best_epoch':    best_ep,
            'best_val_loss': round(best_loss, 4),
            'best_val_acc':  round(best_acc,  4),
            'used_in_vote':  fold_idx in CFG['epoch_vote_folds'],
        })
        del model

    vote_epochs = [best_epochs[i]
                   for i in CFG['epoch_vote_folds']
                   if i < len(best_epochs)]
    optimal_ep  = int(np.ceil(np.mean(vote_epochs))) + 2
    optimal_ep  = min(optimal_ep, CFG['epochs'])

    print(f"    [{ticker} {condition} {model_name}] "
          f"all_folds={str(best_epochs)}  "
          f"vote_folds={str(vote_epochs)}  "
          f"optimal={optimal_ep}")

    tf.random.set_seed(SEED); np.random.seed(SEED)
    final_model = builder(input_shape, model_name)
    final_model.fit(
        X_all, y_all,
        epochs=optimal_ep,
        batch_size=CFG['batch_size'],
        class_weight=cw, verbose=0)

    return final_model, fold_info, optimal_ep, fold_losses


# ── Utilities ─────────────────────────────────────────────────

def evaluate_model(y_true, y_prob, ticker, condition, model_name):
    y_pred = (y_prob >= 0.5).astype(int)
    acc    = accuracy_score(y_true, y_pred)
    try:
        auc = roc_auc_score(y_true, y_prob)
    except Exception:
        auc = 0.5
    f1  = f1_score(y_true, y_pred,
                   average='weighted', zero_division=0)
    ap  = average_precision_score(y_true, y_prob)
    cm  = confusion_matrix(y_true, y_pred)
    return {
        'ticker': ticker, 'condition': condition,
        'model': model_name,
        'accuracy':      round(acc, 4),
        'roc_auc':       round(auc, 4),
        'f1_weighted':   round(f1,  4),
        'avg_precision': round(ap,  4),
        'y_true': y_true, 'y_prob': y_prob,
        'y_pred': y_pred, 'conf_matrix': cm,
    }


def train_classical(arrays, ticker, condition):
    X_tr_f  = arrays['X_train'].reshape(len(arrays['X_train']), -1)
    X_val_f = arrays['X_val'].reshape(len(arrays['X_val']),   -1)
    X_te_f  = arrays['X_test'].reshape(len(arrays['X_test']),  -1)
    X_all_f = np.concatenate([X_tr_f, X_val_f])
    y_all   = np.concatenate([arrays['y_train'], arrays['y_val']])
    y_te    = arrays['y_test']
    n_neg   = (y_all == 0).sum(); n_pos = (y_all == 1).sum()
    scale_pos = n_neg / n_pos if n_pos > 0 else 1.0
    results, models = [], {}

    lr_clf = LogisticRegression(
        max_iter=2000, C=0.1, class_weight='balanced',
        random_state=SEED, n_jobs=-1)
    lr_clf.fit(X_all_f, y_all)
    prob = lr_clf.predict_proba(X_te_f)[:, 1]
    results.append(evaluate_model(
        y_te, prob, ticker, condition, 'Logistic Regression'))
    models['Logistic Regression'] = lr_clf

    rf = RandomForestClassifier(
        n_estimators=CFG['rf_trees'], max_depth=CFG['rf_depth'],
        class_weight='balanced', random_state=SEED, n_jobs=-1)
    rf.fit(X_all_f, y_all)
    prob = rf.predict_proba(X_te_f)[:, 1]
    results.append(evaluate_model(
        y_te, prob, ticker, condition, 'Random Forest'))
    models['Random Forest'] = rf

    xgb_clf = xgb.XGBClassifier(
        n_estimators=CFG['xgb_trees'], max_depth=CFG['xgb_depth'],
        learning_rate=CFG['xgb_lr'], scale_pos_weight=scale_pos,
        eval_metric='logloss', use_label_encoder=False,
        random_state=SEED, verbosity=0)
    xgb_clf.fit(X_all_f, y_all)
    prob = xgb_clf.predict_proba(X_te_f)[:, 1]
    results.append(evaluate_model(
        y_te, prob, ticker, condition, 'XGBoost'))
    models['XGBoost'] = xgb_clf
    return results, models

print("Utilities defined.")


# ── Main training loop ────────────────────────────────────────
RESULTS_R3    = []
FOLD_INFO_R3  = {}
WF_LOSSES_R3  = {}
OPT_EPOCHS_R3 = {}

print("=" * 65)
print("RUN 3 — WALK-FORWARD, COVID FOLD EXCLUDED FROM VOTE")
print("=" * 65)
print(f"Voting folds: {CFG['epoch_vote_folds']} (0-indexed)")
print(f"Excluded: Fold 4 (index 3) — COVID crash 2020")
print()

t_total = time.time()

for ticker in CFG['tickers']:
    print(f"\n{'─'*55}\n  TICKER: {ticker}\n{'─'*55}")

    for condition in CFG['conditions']:
        if condition == 'C3' and ticker not in CFG['tickers_c3']:
            print(f"\n  [{ticker} C3] SKIPPED")
            continue

        arrays    = PREPARED[ticker][condition]
        y_te      = arrays['y_test']
        n_feat    = arrays['X_train'].shape[2]
        inp_shape = (CFG['look_back'], n_feat)

        print(f"\n  Condition {condition} ({n_feat} features)")
        print("  Classical ML:")
        classical_results, _ = train_classical(
            arrays, ticker, condition)
        RESULTS_R3.extend(classical_results)
        for r in classical_results:
            print(f"    [{r['model']:<22}] "
                  f"acc={r['accuracy']:.4f}  "
                  f"auc={r['roc_auc']:.4f}")

        print("  Deep Learning:")
        for model_name, builder in BUILDERS.items():
            tf.random.set_seed(SEED); np.random.seed(SEED)
            t0 = time.time()
            final_model, fold_info, opt_ep, fold_losses = \
                train_wf_model(builder, inp_shape, arrays,
                               ticker, model_name, condition)
            elapsed = time.time() - t0
            y_prob  = final_model.predict(
                arrays['X_test'], verbose=0).ravel()
            result  = evaluate_model(
                y_te, y_prob, ticker, condition, model_name)
            RESULTS_R3.append(result)
            key = f"{ticker}_{condition}_{model_name}"
            FOLD_INFO_R3[key]  = fold_info
            WF_LOSSES_R3[key]  = fold_losses
            OPT_EPOCHS_R3[key] = opt_ep
            print(f"    [{model_name:<12}] "
                  f"opt_ep={opt_ep:3d}  "
                  f"acc={result['accuracy']:.4f}  "
                  f"auc={result['roc_auc']:.4f}  "
                  f"time={elapsed:.0f}s")

    interim = pd.DataFrame([
        {k: v for k, v in r.items()
         if k not in ('y_true','y_prob','y_pred','conf_matrix')}
        for r in RESULTS_R3])
    interim.to_csv(
        os.path.join(SAVE_PATH, f'r3_interim_{ticker}.csv'),
        index=False)
    print(f"\n  Saved interim after {ticker}.")

total_time = time.time() - t_total
print(f"\n{'='*65}")
print(f"Run 3 complete. Total time: {total_time/3600:.1f} hours")
print(f"{'='*65}")


# ── Save ──────────────────────────────────────────────────────
RESULTS_DF_R3 = pd.DataFrame([
    {k: v for k, v in r.items()
     if k not in ('y_true','y_prob','y_pred','conf_matrix')}
    for r in RESULTS_R3])

RESULTS_DF_R3.to_csv(
    os.path.join(SAVE_PATH, 'results_r3.csv'), index=False)
with open(os.path.join(SAVE_PATH, 'RESULTS_R3.pkl'),   'wb') as f:
    pickle.dump(RESULTS_R3,   f, protocol=4)
with open(os.path.join(SAVE_PATH, 'FOLD_INFO_R3.pkl'), 'wb') as f:
    pickle.dump(FOLD_INFO_R3, f, protocol=4)
with open(os.path.join(SAVE_PATH, 'WF_LOSSES_R3.pkl'), 'wb') as f:
    pickle.dump(WF_LOSSES_R3, f, protocol=4)
print("Saved: results_r3.csv and pkl files.")


# ── Four-run comparison ───────────────────────────────────────
print("\n" + "=" * 65)
print("FOUR-RUN COMPARISON — CNN-LSTM ACCURACY")
print("=" * 65)

def get_acc(df, ticker, condition, model='CNN-LSTM'):
    row = df[(df['ticker']    == ticker) &
             (df['condition'] == condition) &
             (df['model']     == model)]
    return float(row['accuracy'].values[0]) if len(row) > 0 else None

def fmt(val):
    return f"{val:.4f}" if val is not None else " N/A "

try:
    R0 = pd.read_csv(os.path.join(SAVE_PATH, 'results_df.csv'))
    R1 = pd.read_csv(os.path.join(SAVE_PATH, 'results_r1.csv'))
    R2 = pd.read_csv(os.path.join(SAVE_PATH, 'results_r2.csv'))
    has_prev = True
except FileNotFoundError:
    has_prev = False

for condition in CFG['conditions']:
    tickers = (CFG['tickers_c3']
               if condition == 'C3' else CFG['tickers'])
    r3_sub  = RESULTS_DF_R3[
        (RESULTS_DF_R3['condition'] == condition) &
        (RESULTS_DF_R3['model']     == 'CNN-LSTM')]
    print(f"\n  Condition {condition}:")
    if has_prev:
        print(f"  {'Ticker':<8} {'Run0':>8} {'Run1':>8} "
              f"{'Run2':>8} {'Run3':>8} {'R0->R3':>8}")
        print(f"  {'-'*52}")
    for ticker in tickers:
        r3_row = r3_sub[r3_sub['ticker'] == ticker]
        r3a    = float(r3_row['accuracy'].values[0]) \
                 if len(r3_row) > 0 else None
        if has_prev:
            r0a   = get_acc(R0, ticker, condition)
            r1a   = get_acc(R1, ticker, condition)
            r2a   = get_acc(R2, ticker, condition)
            delta = (r3a - r0a) \
                    if (r3a is not None and r0a is not None) \
                    else None
            d_str = f"{delta:+.4f}" if delta is not None \
                    else " N/A "
            print(f"  {ticker:<8} {fmt(r0a):>8} {fmt(r1a):>8} "
                  f"{fmt(r2a):>8} {fmt(r3a):>8} {d_str:>8}")
        else:
            print(f"  {ticker:<8} {fmt(r3a):>8}")

print(f"\n  Mean CNN-LSTM accuracy:")
if has_prev:
    print(f"  {'Cond':<6} {'Run0':>8} {'Run1':>8} "
          f"{'Run2':>8} {'Run3':>8} {'R0->R3':>8}")
    print(f"  {'-'*50}")
for cond in CFG['conditions']:
    r3m = RESULTS_DF_R3[
        (RESULTS_DF_R3['condition'] == cond) &
        (RESULTS_DF_R3['model']     == 'CNN-LSTM')]['accuracy'].mean()
    if has_prev:
        r0m = R0[(R0['condition'] == cond) &
                 (R0['model']     == 'CNN-LSTM')]['accuracy'].mean()
        r1m = R1[(R1['condition'] == cond) &
                 (R1['model']     == 'CNN-LSTM')]['accuracy'].mean()
        r2m = R2[(R2['condition'] == cond) &
                 (R2['model']     == 'CNN-LSTM')]['accuracy'].mean()
        print(f"  {cond:<6} {r0m:.4f}  {r1m:.4f}  "
              f"{r2m:.4f}  {r3m:.4f}  {r3m - r0m:+.4f}")
    else:
        print(f"  {cond:<6} Run3={r3m:.4f}")


# ── Optimal epochs comparison ─────────────────────────────────
print("\n" + "=" * 65)
print("OPTIMAL EPOCHS: RUN 2 vs RUN 3 — CNN-LSTM")
print("=" * 65)
print("Expected: Run 3 epochs should be higher (Fold 4 excluded)")
print()

print(f"{'Ticker':<8} {'Cond':<5} "
      f"{'R2_opt':>8} {'R3_opt':>8} "
      f"{'R3_vote_eps':>20} {'R3_all_eps':>25}")
print("-" * 72)

for ticker in CFG['tickers']:
    for cond in ['C1', 'C2', 'C3']:
        if cond == 'C3' and ticker not in CFG['tickers_c3']:
            continue
        key     = f"{ticker}_{cond}_CNN-LSTM"
        r3_opt  = OPT_EPOCHS_R3.get(key, None)
        r3_info = FOLD_INFO_R3.get(key, [])
        r3_all  = [fi['best_epoch'] for fi in r3_info]
        r3_vote = [r3_all[i] for i in CFG['epoch_vote_folds']
                   if i < len(r3_all)]
        try:
            r2_opt = int(list({
                k: v for k, v in
                __import__('pickle').load(
                    open(os.path.join(
                        SAVE_PATH, 'FOLD_INFO_R2.pkl'), 'rb')
                ).items()
            }[key][-1]['best_epoch']))
        except Exception:
            r2_opt = None
        r3_opt_str = str(r3_opt) if r3_opt else 'N/A'
        r2_opt_str = 'N/A'
        print(f"{ticker:<8} {cond:<5} "
              f"{r2_opt_str:>8} {r3_opt_str:>8} "
              f"{str(r3_vote):>20} {str(r3_all):>25}")


# ── Walk-forward curves figure ────────────────────────────────
print("\nGenerating Run 3 walk-forward curves...")

fig, axes = plt.subplots(
    len(CFG['tickers']), CFG['n_folds'],
    figsize=(4*CFG['n_folds'], 3*len(CFG['tickers'])),
    facecolor=DARK, sharey='row')
fig.suptitle(
    'Walk-Forward Curves — CNN-LSTM C2 | Run 3\n'
    'Folds 1-3 used for epoch vote (blue/green/yellow). '
    'Fold 4 excluded (orange — COVID window).',
    color=TEXT, fontsize=11, fontweight='bold')

fold_colors = [ACCENT, GREEN, YELLOW, ORANGE]
fold_labels = ['Fold 1 (vote)', 'Fold 2 (vote)',
               'Fold 3 (vote)', 'Fold 4 (excluded)']

for row, ticker in enumerate(CFG['tickers']):
    key    = f"{ticker}_C2_CNN-LSTM"
    losses = WF_LOSSES_R3.get(key, [])
    finfos = FOLD_INFO_R3.get(key, [])
    opt_ep = OPT_EPOCHS_R3.get(key, None)

    for fold_idx in range(CFG['n_folds']):
        ax  = axes[row, fold_idx]
        ax.set_facecolor(DARK)
        col = fold_colors[fold_idx]
        if fold_idx < len(losses):
            ep = range(1, len(losses[fold_idx]) + 1)
            ls = '--' if fold_idx not in CFG['epoch_vote_folds'] \
                 else '-'
            ax.plot(ep, losses[fold_idx],
                    color=col, linewidth=1.4, linestyle=ls)
            if fold_idx < len(finfos):
                best = finfos[fold_idx]['best_epoch']
                ax.axvline(best, color=col,
                           linestyle=':', linewidth=0.9)
        if fold_idx not in CFG['epoch_vote_folds']:
            ax.text(0.5, 0.92, 'EXCLUDED',
                    transform=ax.transAxes,
                    ha='center', fontsize=7,
                    color=ORANGE, alpha=0.8)
        if fold_idx == 0:
            ax.set_ylabel(ticker, color=COLORS[row], fontsize=9)
        if row == 0:
            ax.set_title(fold_labels[fold_idx],
                         color=fold_colors[fold_idx], fontsize=8)
        if opt_ep and fold_idx == CFG['n_folds'] - 1:
            ax.text(0.98, 0.95, f'opt={opt_ep}',
                    transform=ax.transAxes,
                    ha='right', va='top',
                    fontsize=7, color=MUTED)
        ax.grid(True, alpha=0.3)
        ax.tick_params(labelsize=6)

plt.tight_layout()
plt.savefig('R3_wf_curves.png', dpi=150,
            bbox_inches='tight', facecolor=DARK)
plt.show()
print("Saved: R3_wf_curves.png")


# ── Full accuracy summary ─────────────────────────────────────
print("\n" + "=" * 65)
print("RUN 3 — FULL RESULTS SUMMARY")
print("=" * 65)

for condition in CFG['conditions']:
    sub = RESULTS_DF_R3[RESULTS_DF_R3['condition'] == condition]
    if sub.empty:
        continue
    pivot = sub.pivot_table(
        index='model', columns='ticker',
        values='accuracy', aggfunc='first')
    pivot['Mean'] = pivot.mean(axis=1)
    pivot = pivot.sort_values('Mean', ascending=False)
    print(f"\n  Condition {condition} — Accuracy:")
    print(pivot.round(4).to_string())

print("\n" + "=" * 65)
print("Mean accuracy — all conditions, all models")
print("=" * 65)
smry = RESULTS_DF_R3.groupby(
    ['condition','model'])['accuracy'].mean()\
    .unstack('condition').round(4)
smry['Overall'] = smry.mean(axis=1)
print(smry.sort_values('Overall', ascending=False).to_string())

print("\nBenchmark comparison (CNN-LSTM mean per condition):")
for cond in ['C1', 'C2', 'C3']:
    val = RESULTS_DF_R3[
        (RESULTS_DF_R3['model']     == 'CNN-LSTM') &
        (RESULTS_DF_R3['condition'] == cond)]['accuracy'].mean()
    print(f"  CNN-LSTM Run 3 ({cond}): {val*100:.2f}%  "
          f"(Fischer & Krauss benchmark: 53.20%)")


# ── Confusion matrices ────────────────────────────────────────
from matplotlib.colors import LinearSegmentedColormap
cmap_dark = LinearSegmentedColormap.from_list(
    'dark_blue', [DARK, ACCENT])

for condition in CFG['conditions']:
    tickers_this = (CFG['tickers_c3']
                    if condition == 'C3' else CFG['tickers'])
    n   = len(tickers_this)
    fig, axes = plt.subplots(
        1, n, figsize=(3.5*n, 4), facecolor=DARK)
    if n == 1:
        axes = [axes]
    fig.suptitle(
        f'Confusion Matrices — CNN-LSTM Run 3 | {condition}',
        color=TEXT, fontsize=12, fontweight='bold')

    for idx, ticker in enumerate(tickers_this):
        r = next(
            (x for x in RESULTS_R3
             if x['ticker']    == ticker
             and x['condition'] == condition
             and x['model']     == 'CNN-LSTM'), None)
        if r is None:
            continue
        cm  = r['conf_matrix']
        ax  = axes[idx]
        ax.set_facecolor(DARK)
        ax.imshow(cm, cmap=cmap_dark)
        ax.set_title(f"{ticker}\nacc={r['accuracy']:.4f}",
                     color=TEXT, fontsize=10)
        ax.set_xticks([0, 1]); ax.set_yticks([0, 1])
        ax.set_xticklabels(['DOWN', 'UP'], fontsize=9)
        ax.set_yticklabels(['DOWN', 'UP'], fontsize=9)
        if idx == 0:
            ax.set_ylabel('Actual', fontsize=9)
        ax.set_xlabel('Predicted', fontsize=9)
        for i in range(2):
            for j in range(2):
                ax.text(j, i, str(cm[i, j]),
                        ha='center', va='center',
                        fontsize=13, fontweight='bold',
                        color=TEXT)

    plt.tight_layout()
    fname = f'R3_confusion_{condition}.png'
    plt.savefig(fname, dpi=150,
                bbox_inches='tight', facecolor=DARK)
    plt.show()
    print(f"Saved: {fname}")

print("\nRun 3 complete.")
```

```python
# ============================================================
# Phase   : CRISP-DM — Modelling, Run 4
#
# Change from Run 3:
#   A temporal self-attention layer is inserted between the CNN
#   block and the LSTM layer in the CNN-LSTM architecture only.
#   CNN-only and LSTM-only remain unchanged as ablation baselines.
#
# ATTENTION MECHANISM:
#   After the CNN+MaxPool+Dropout block the sequence has shape
#   (batch, reduced_timesteps, cnn_filters). The attention layer
#   computes a scalar weight for each timestep, then returns the
#   weighted sum. This gives the LSTM a soft selection over
#   which parts of the 20-day window matter most.
#
#   Implementation: single-head additive (Bahdanau) attention.
#   Reference: Bahdanau et al. (2015) — Neural Machine
#   Translation by Jointly Learning to Align and Translate.
#   Adapted for financial time series:
#   Qin et al. (2017) — A Dual-Stage Attention-Based
#   Recurrent Neural Network for Time Series Prediction.
#
#   Residual connection added for gradient stability.
#   Reference: He et al. (2016) — Deep Residual Learning.
#
#   Attention adds ~2K-4K parameters relative to the 28K
#   total in the CNN-LSTM Run 3 model.
#
# Note: the proposed model is named 'CNN-LSTM-Attn' so that
#   it is distinguishable from Run 3's 'CNN-LSTM' in results.
#
# PREREQUISITE:
#   PREPARED, SCALERS, PREP_INFO in memory or reload from disk.
# ============================================================

import os, pickle, random, warnings, time
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats as scipy_stats

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, callbacks, regularizers
from tensorflow.keras.optimizers import Adam
from sklearn.metrics import (
    accuracy_score, roc_auc_score, f1_score,
    average_precision_score, confusion_matrix, roc_curve)
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
import xgboost as xgb

# ── Reproducibility ──────────────────────────────────────────
SEED = 42
os.environ['PYTHONHASHSEED']       = str(SEED)
os.environ['TF_DETERMINISTIC_OPS'] = '1'
random.seed(SEED); np.random.seed(SEED); tf.random.set_seed(SEED)

# ── Plot theme ───────────────────────────────────────────────
DARK   = '#0d1117'; BORDER = '#30363d'; TEXT = '#c9d1d9'
MUTED  = '#8b949e'; GRID   = '#21262d'
ACCENT = '#58a6ff'; GREEN  = '#3fb950'; RED    = '#f85149'
YELLOW = '#e3b341'; PURPLE = '#bc8cff'; ORANGE = '#f0883e'
COLORS = [ACCENT, GREEN, RED, YELLOW, PURPLE]

plt.rcParams.update({
    'figure.facecolor': DARK, 'axes.facecolor':  DARK,
    'axes.edgecolor':  BORDER, 'axes.labelcolor': TEXT,
    'xtick.color': MUTED, 'ytick.color': MUTED,
    'text.color': TEXT,   'grid.color':  GRID,
    'grid.linestyle': '--', 'grid.alpha': 0.5,
    'font.family': 'monospace',
    'legend.facecolor': '#161b22',
    'legend.edgecolor': BORDER, 'figure.dpi': 120,
})

SAVE_PATH = '/content/'

def load_prepared(path=SAVE_PATH):
    with open(os.path.join(path, 'PREPARED.pkl'),  'rb') as f:
        p = pickle.load(f)
    with open(os.path.join(path, 'SCALERS.pkl'),   'rb') as f:
        s = pickle.load(f)
    with open(os.path.join(path, 'PREP_INFO.pkl'), 'rb') as f:
        i = pickle.load(f)
    print(f"Loaded from {path}")
    return p, s, i

try:
    _ = PREPARED
    print("PREPARED in memory.")
except NameError:
    PREPARED, SCALERS, PREP_INFO = load_prepared()

# ── Configuration — identical to Run 3 ───────────────────────
CFG = dict(
    tickers    = ['AAPL', 'AMZN', 'GOOG', 'MSFT', 'TSLA'],
    tickers_c3 = ['AAPL', 'AMZN', 'GOOG', 'TSLA'],
    conditions = ['C1', 'C2', 'C3'],
    look_back        = 20,
    learning_rate    = 1e-4,
    lstm_units       = 64,
    cnn_filters      = 32,
    patience         = 20,
    cnn_kernel       = 3,
    dropout          = 0.3,
    l2_reg           = 1e-4,
    batch_size       = 32,
    epochs           = 150,
    clip_norm        = 1.0,
    n_folds          = 4,
    epoch_vote_folds = [0, 1, 2],
    rf_trees=200, rf_depth=8,
    xgb_trees=200, xgb_depth=4, xgb_lr=0.05,
    save_path = SAVE_PATH,
    run_name  = 'run4',
)

print("Run 4 configuration:")
print("  All Run 3 settings carried forward.")
print("  ONE CHANGE: attention layer added to CNN-LSTM only.")


# ── Custom attention layer ────────────────────────────────────

class TemporalAttention(layers.Layer):
    """
    Single-head temporal self-attention.
    Computes attention weights over the time dimension and
    returns the weighted sequence for the downstream LSTM.

    Mathematically:
        score(t) = tanh(W * h_t + b)       [alignment model]
        alpha(t) = softmax(v * score(t))    [attention weights]
        context  = sum(alpha(t) * h_t)      [context vector]

    Reference: Bahdanau et al. (2015); Qin et al. (2017)
    """
    def __init__(self, units=32, **kwargs):
        super().__init__(**kwargs)
        self.units = units
        self.W = layers.Dense(units, use_bias=True,
                              activation='tanh')
        self.v = layers.Dense(1, use_bias=False)

    def call(self, hidden_states):
        score   = self.W(hidden_states)
        score   = self.v(score)
        weights = tf.nn.softmax(score, axis=1)
        context = weights * hidden_states
        return context, weights

    def get_config(self):
        config = super().get_config()
        config.update({'units': self.units})
        return config


# ── Model builders ────────────────────────────────────────────

def build_cnn_lstm_attention(input_shape, name='CNN-LSTM-Attn'):
    """
    CNN-LSTM with temporal attention between CNN and LSTM.

    Architecture:
      Input → Conv1D → MaxPool → Dropout
            → TemporalAttention + residual + LayerNorm
            → LSTM → BatchNorm → Dropout → Dense(32) → Sigmoid
    """
    reg = regularizers.l2(CFG['l2_reg'])
    inp = layers.Input(shape=input_shape)
    x   = layers.Conv1D(
              CFG['cnn_filters'], CFG['cnn_kernel'],
              padding='causal', activation='relu',
              kernel_regularizer=reg)(inp)
    x   = layers.MaxPooling1D(pool_size=2)(x)
    x   = layers.Dropout(CFG['dropout'])(x)
    attn_out, attn_weights = TemporalAttention(
        units=32, name='temporal_attention')(x)
    x   = layers.Add()([x, attn_out])
    x   = layers.LayerNormalization()(x)
    x   = layers.LSTM(
              CFG['lstm_units'],
              kernel_regularizer=reg,
              recurrent_regularizer=reg)(x)
    x   = layers.BatchNormalization()(x)
    x   = layers.Dropout(CFG['dropout'])(x)
    x   = layers.Dense(32, activation='relu',
                       kernel_regularizer=reg)(x)
    out = layers.Dense(1, activation='sigmoid')(x)
    model = keras.Model(inp, out, name=name)
    model.compile(
        optimizer=Adam(CFG['learning_rate'],
                       clipnorm=CFG['clip_norm']),
        loss='binary_crossentropy', metrics=['accuracy'])
    return model


def build_cnn_only(input_shape, name='CNN-only'):
    reg = regularizers.l2(CFG['l2_reg'])
    inp = layers.Input(shape=input_shape)
    x   = layers.Conv1D(CFG['cnn_filters'], CFG['cnn_kernel'],
                        padding='causal', activation='relu',
                        kernel_regularizer=reg)(inp)
    x   = layers.Conv1D(CFG['cnn_filters'], CFG['cnn_kernel'],
                        padding='causal', activation='relu',
                        kernel_regularizer=reg)(x)
    x   = layers.GlobalAveragePooling1D()(x)
    x   = layers.BatchNormalization()(x)
    x   = layers.Dropout(CFG['dropout'])(x)
    x   = layers.Dense(32, activation='relu',
                       kernel_regularizer=reg)(x)
    out = layers.Dense(1, activation='sigmoid')(x)
    m   = keras.Model(inp, out, name=name)
    m.compile(
        optimizer=Adam(CFG['learning_rate'],
                       clipnorm=CFG['clip_norm']),
        loss='binary_crossentropy', metrics=['accuracy'])
    return m


def build_lstm_only(input_shape, name='LSTM-only'):
    reg = regularizers.l2(CFG['l2_reg'])
    inp = layers.Input(shape=input_shape)
    x   = layers.LSTM(CFG['lstm_units'],
                      kernel_regularizer=reg,
                      recurrent_regularizer=reg)(inp)
    x   = layers.BatchNormalization()(x)
    x   = layers.Dropout(CFG['dropout'])(x)
    x   = layers.Dense(32, activation='relu',
                       kernel_regularizer=reg)(x)
    out = layers.Dense(1, activation='sigmoid')(x)
    m   = keras.Model(inp, out, name=name)
    m.compile(
        optimizer=Adam(CFG['learning_rate'],
                       clipnorm=CFG['clip_norm']),
        loss='binary_crossentropy', metrics=['accuracy'])
    return m

BUILDERS = {
    'CNN-only':      build_cnn_only,
    'LSTM-only':     build_lstm_only,
    'CNN-LSTM-Attn': build_cnn_lstm_attention,
}

# Parameter count comparison (Run 3 plain vs Run 4 attention)
print("\nParameter counts — Run 3 vs Run 4 (CNN-LSTM):")
shape_test = PREPARED['AAPL']['C2']['X_train'].shape[1:]

def build_cnn_lstm_plain(input_shape):
    reg = regularizers.l2(CFG['l2_reg'])
    inp = layers.Input(shape=input_shape)
    x   = layers.Conv1D(CFG['cnn_filters'], CFG['cnn_kernel'],
                        padding='causal', activation='relu',
                        kernel_regularizer=reg)(inp)
    x   = layers.MaxPooling1D(pool_size=2)(x)
    x   = layers.Dropout(CFG['dropout'])(x)
    x   = layers.LSTM(CFG['lstm_units'],
                      kernel_regularizer=reg,
                      recurrent_regularizer=reg)(x)
    x   = layers.BatchNormalization()(x)
    x   = layers.Dropout(CFG['dropout'])(x)
    x   = layers.Dense(32, activation='relu',
                       kernel_regularizer=reg)(x)
    out = layers.Dense(1, activation='sigmoid')(x)
    return keras.Model(inp, out)

m_plain = build_cnn_lstm_plain(shape_test)
m_attn  = build_cnn_lstm_attention(shape_test)
print(f"  CNN-LSTM (Run 3):      {m_plain.count_params():>8,} params")
print(f"  CNN-LSTM-Attn (Run 4): {m_attn.count_params():>8,} params")
print(f"  Parameter increase:    "
      f"{m_attn.count_params() - m_plain.count_params():>+8,} params")
del m_plain, m_attn


# ── Walk-forward fold constructor — identical to Run 3 ────────

def make_wf_folds(X_all, y_all, n_folds=4, min_train=504):
    n     = len(X_all)
    step  = (n - min_train) // n_folds
    folds = []
    for f in range(n_folds):
        train_end = min_train + f * step
        val_start = train_end
        val_end   = min(train_end + step, n)
        if val_end <= val_start:
            continue
        folds.append((
            X_all[:train_end], y_all[:train_end],
            X_all[val_start:val_end], y_all[val_start:val_end],
        ))
    return folds


def train_wf_model(builder, input_shape, arrays,
                   ticker, model_name, condition):
    """Walk-forward training protocol — identical to Run 3."""
    X_all = np.concatenate([
        arrays['X_train'], arrays['X_val']], axis=0)
    y_all = np.concatenate([
        arrays['y_train'], arrays['y_val']], axis=0)
    n_pos = y_all.sum()
    n_neg = len(y_all) - n_pos
    cw    = {0: len(y_all)/(2*n_neg) if n_neg > 0 else 1.0,
             1: len(y_all)/(2*n_pos) if n_pos > 0 else 1.0}

    folds = make_wf_folds(
        X_all, y_all,
        n_folds=CFG['n_folds'],
        min_train=max(200, len(X_all)//(CFG['n_folds']+1)))

    fold_info   = []
    fold_losses = []
    best_epochs = []

    for fold_idx, (X_tr, y_tr, X_va, y_va) in enumerate(folds):
        tf.random.set_seed(SEED + fold_idx)
        np.random.seed(SEED + fold_idx)
        model = builder(input_shape, model_name)
        cb = [
            callbacks.EarlyStopping(
                monitor='val_loss', patience=CFG['patience'],
                restore_best_weights=True, verbose=0),
            callbacks.ReduceLROnPlateau(
                monitor='val_loss', factor=0.5, patience=10,
                min_lr=1e-7, verbose=0),
        ]
        hist = model.fit(
            X_tr, y_tr,
            validation_data=(X_va, y_va),
            epochs=CFG['epochs'],
            batch_size=CFG['batch_size'],
            class_weight=cw,
            callbacks=cb, verbose=0)
        best_ep   = int(np.argmin(hist.history['val_loss'])) + 1
        best_loss = min(hist.history['val_loss'])
        best_acc  = hist.history['val_accuracy'][best_ep - 1]
        best_epochs.append(best_ep)
        fold_losses.append(hist.history['val_loss'])
        fold_info.append({
            'fold':          fold_idx + 1,
            'train_size':    len(X_tr),
            'val_size':      len(X_va),
            'best_epoch':    best_ep,
            'best_val_loss': round(best_loss, 4),
            'best_val_acc':  round(best_acc,  4),
        })
        del model

    vote_epochs = [best_epochs[i]
                   for i in CFG['epoch_vote_folds']
                   if i < len(best_epochs)]
    optimal_ep  = int(np.ceil(np.mean(vote_epochs))) + 2
    optimal_ep  = min(optimal_ep, CFG['epochs'])

    print(f"    [{ticker} {condition} {model_name}] "
          f"vote={vote_epochs}  optimal={optimal_ep}")

    tf.random.set_seed(SEED); np.random.seed(SEED)
    final_model = builder(input_shape, model_name)
    final_model.fit(
        X_all, y_all,
        epochs=optimal_ep,
        batch_size=CFG['batch_size'],
        class_weight=cw, verbose=0)

    return final_model, fold_info, optimal_ep, fold_losses


# ── Utilities ─────────────────────────────────────────────────

def evaluate_model(y_true, y_prob, ticker, condition, model_name):
    y_pred = (y_prob >= 0.5).astype(int)
    acc    = accuracy_score(y_true, y_pred)
    try:
        auc = roc_auc_score(y_true, y_prob)
    except Exception:
        auc = 0.5
    f1  = f1_score(y_true, y_pred,
                   average='weighted', zero_division=0)
    ap  = average_precision_score(y_true, y_prob)
    cm  = confusion_matrix(y_true, y_pred)
    return {
        'ticker': ticker, 'condition': condition,
        'model': model_name,
        'accuracy':      round(acc, 4),
        'roc_auc':       round(auc, 4),
        'f1_weighted':   round(f1,  4),
        'avg_precision': round(ap,  4),
        'y_true': y_true, 'y_prob': y_prob,
        'y_pred': y_pred, 'conf_matrix': cm,
    }


def train_classical(arrays, ticker, condition):
    X_tr_f  = arrays['X_train'].reshape(len(arrays['X_train']), -1)
    X_val_f = arrays['X_val'].reshape(len(arrays['X_val']),   -1)
    X_te_f  = arrays['X_test'].reshape(len(arrays['X_test']),  -1)
    X_all_f = np.concatenate([X_tr_f, X_val_f])
    y_all   = np.concatenate([arrays['y_train'], arrays['y_val']])
    y_te    = arrays['y_test']
    n_neg   = (y_all == 0).sum(); n_pos = (y_all == 1).sum()
    scale_pos = n_neg / n_pos if n_pos > 0 else 1.0
    results, models = [], {}

    lr_clf = LogisticRegression(
        max_iter=2000, C=0.1, class_weight='balanced',
        random_state=SEED, n_jobs=-1)
    lr_clf.fit(X_all_f, y_all)
    prob = lr_clf.predict_proba(X_te_f)[:, 1]
    results.append(evaluate_model(
        y_te, prob, ticker, condition, 'Logistic Regression'))

    rf = RandomForestClassifier(
        n_estimators=CFG['rf_trees'], max_depth=CFG['rf_depth'],
        class_weight='balanced', random_state=SEED, n_jobs=-1)
    rf.fit(X_all_f, y_all)
    prob = rf.predict_proba(X_te_f)[:, 1]
    results.append(evaluate_model(
        y_te, prob, ticker, condition, 'Random Forest'))

    xgb_clf = xgb.XGBClassifier(
        n_estimators=CFG['xgb_trees'], max_depth=CFG['xgb_depth'],
        learning_rate=CFG['xgb_lr'], scale_pos_weight=scale_pos,
        eval_metric='logloss', use_label_encoder=False,
        random_state=SEED, verbosity=0)
    xgb_clf.fit(X_all_f, y_all)
    prob = xgb_clf.predict_proba(X_te_f)[:, 1]
    results.append(evaluate_model(
        y_te, prob, ticker, condition, 'XGBoost'))
    return results, models

print("Utilities defined.")


# ── Main training loop ────────────────────────────────────────
RESULTS_R4    = []
FOLD_INFO_R4  = {}
WF_LOSSES_R4  = {}
OPT_EPOCHS_R4 = {}

print("=" * 65)
print("RUN 4 — CNN-LSTM WITH ATTENTION")
print("=" * 65)
print("Proposed model: CNN-LSTM-Attn")
print("Ablations unchanged: CNN-only, LSTM-only")
print()

t_total = time.time()

for ticker in CFG['tickers']:
    print(f"\n{'─'*55}\n  TICKER: {ticker}\n{'─'*55}")

    for condition in CFG['conditions']:
        if condition == 'C3' and ticker not in CFG['tickers_c3']:
            print(f"\n  [{ticker} C3] SKIPPED")
            continue

        arrays    = PREPARED[ticker][condition]
        y_te      = arrays['y_test']
        n_feat    = arrays['X_train'].shape[2]
        inp_shape = (CFG['look_back'], n_feat)

        print(f"\n  Condition {condition} ({n_feat} features)")
        print("  Classical ML:")
        classical_results, _ = train_classical(
            arrays, ticker, condition)
        RESULTS_R4.extend(classical_results)
        for r in classical_results:
            print(f"    [{r['model']:<22}] "
                  f"acc={r['accuracy']:.4f}  "
                  f"auc={r['roc_auc']:.4f}")

        print("  Deep Learning:")
        for model_name, builder in BUILDERS.items():
            tf.random.set_seed(SEED); np.random.seed(SEED)
            t0 = time.time()
            final_model, fold_info, opt_ep, fold_losses = \
                train_wf_model(builder, inp_shape, arrays,
                               ticker, model_name, condition)
            elapsed = time.time() - t0
            y_prob  = final_model.predict(
                arrays['X_test'], verbose=0).ravel()
            result  = evaluate_model(
                y_te, y_prob, ticker, condition, model_name)
            RESULTS_R4.append(result)
            key = f"{ticker}_{condition}_{model_name}"
            FOLD_INFO_R4[key]  = fold_info
            WF_LOSSES_R4[key]  = fold_losses
            OPT_EPOCHS_R4[key] = opt_ep
            print(f"    [{model_name:<16}] "
                  f"opt_ep={opt_ep:3d}  "
                  f"acc={result['accuracy']:.4f}  "
                  f"auc={result['roc_auc']:.4f}  "
                  f"time={elapsed:.0f}s")

    interim = pd.DataFrame([
        {k: v for k, v in r.items()
         if k not in ('y_true','y_prob','y_pred','conf_matrix')}
        for r in RESULTS_R4])
    interim.to_csv(
        os.path.join(SAVE_PATH, f'r4_interim_{ticker}.csv'),
        index=False)
    print(f"\n  Saved interim after {ticker}.")

total_time = time.time() - t_total
print(f"\n{'='*65}")
print(f"Run 4 complete. Total time: {total_time/3600:.1f} hours")


# ── Save ──────────────────────────────────────────────────────
RESULTS_DF_R4 = pd.DataFrame([
    {k: v for k, v in r.items()
     if k not in ('y_true','y_prob','y_pred','conf_matrix')}
    for r in RESULTS_R4])

RESULTS_DF_R4.to_csv(
    os.path.join(SAVE_PATH, 'results_r4.csv'), index=False)
with open(os.path.join(SAVE_PATH, 'RESULTS_R4.pkl'),   'wb') as f:
    pickle.dump(RESULTS_R4,   f, protocol=4)
with open(os.path.join(SAVE_PATH, 'FOLD_INFO_R4.pkl'), 'wb') as f:
    pickle.dump(FOLD_INFO_R4, f, protocol=4)
print("Saved: results_r4.csv, RESULTS_R4.pkl, FOLD_INFO_R4.pkl")


# ── Key comparison: Run 3 CNN-LSTM vs Run 4 CNN-LSTM-Attn ─────
print("\n" + "=" * 65)
print("KEY COMPARISON: CNN-LSTM (Run 3) vs CNN-LSTM-Attn (Run 4)")
print("=" * 65)

def fmt(val):
    return f"{val:.4f}" if val is not None else " N/A "

def get_acc(df, ticker, condition, model):
    row = df[(df['ticker']    == ticker) &
             (df['condition'] == condition) &
             (df['model']     == model)]
    return float(row['accuracy'].values[0]) if len(row) > 0 else None

def get_auc(df, ticker, condition, model):
    row = df[(df['ticker']    == ticker) &
             (df['condition'] == condition) &
             (df['model']     == model)]
    return float(row['roc_auc'].values[0]) if len(row) > 0 else None

try:
    R3     = pd.read_csv(os.path.join(SAVE_PATH, 'results_r3.csv'))
    has_r3 = True
except FileNotFoundError:
    has_r3 = False
    print("Run 3 CSV not found — showing Run 4 only.")

for condition in CFG['conditions']:
    tickers = (CFG['tickers_c3']
               if condition == 'C3' else CFG['tickers'])
    print(f"\n  Condition {condition} — Accuracy:")
    if has_r3:
        print(f"  {'Ticker':<8} {'R3_CNNLSTM':>12} "
              f"{'R4_Attn':>10} {'Delta':>8} "
              f"{'R3_AUC':>8} {'R4_AUC':>8}")
        print(f"  {'-'*56}")
    for ticker in tickers:
        r4a  = get_acc(RESULTS_DF_R4, ticker, condition,
                       'CNN-LSTM-Attn')
        r4au = get_auc(RESULTS_DF_R4, ticker, condition,
                       'CNN-LSTM-Attn')
        if has_r3:
            r3a   = get_acc(R3, ticker, condition, 'CNN-LSTM')
            r3au  = get_auc(R3, ticker, condition, 'CNN-LSTM')
            delta = (r4a - r3a) \
                    if (r4a is not None and r3a is not None) \
                    else None
            d_str = f"{delta:+.4f}" if delta is not None \
                    else " N/A "
            print(f"  {ticker:<8} {fmt(r3a):>12} "
                  f"{fmt(r4a):>10} {d_str:>8} "
                  f"{fmt(r3au):>8} {fmt(r4au):>8}")
        else:
            print(f"  {ticker:<8} {fmt(r4a):>10} {fmt(r4au):>8}")

print(f"\n  Mean accuracy:")
if has_r3:
    print(f"  {'Cond':<6} {'R3_CNNLSTM':>12} "
          f"{'R4_Attn':>10} {'Delta':>8}")
    print(f"  {'-'*38}")
for cond in CFG['conditions']:
    r4m = RESULTS_DF_R4[
        (RESULTS_DF_R4['condition'] == cond) &
        (RESULTS_DF_R4['model']     == 'CNN-LSTM-Attn')
    ]['accuracy'].mean()
    if has_r3:
        r3m = R3[(R3['condition'] == cond) &
                 (R3['model']     == 'CNN-LSTM')]['accuracy'].mean()
        print(f"  {cond:<6} {r3m:.4f}      "
              f"{r4m:.4f}  {r4m - r3m:+.4f}")
    else:
        print(f"  {cond:<6} Run4={r4m:.4f}")


# ── Full results summary ──────────────────────────────────────
print("\n" + "=" * 65)
print("RUN 4 — FULL RESULTS SUMMARY")
print("=" * 65)

for condition in CFG['conditions']:
    sub = RESULTS_DF_R4[RESULTS_DF_R4['condition'] == condition]
    if sub.empty:
        continue
    pivot = sub.pivot_table(
        index='model', columns='ticker',
        values='accuracy', aggfunc='first')
    pivot['Mean'] = pivot.mean(axis=1)
    pivot = pivot.sort_values('Mean', ascending=False)
    print(f"\n  Condition {condition} — Accuracy:")
    print(pivot.round(4).to_string())

print("\n" + "=" * 65)
print("Mean accuracy — all conditions")
print("=" * 65)
smry = RESULTS_DF_R4.groupby(
    ['condition','model'])['accuracy'].mean()\
    .unstack('condition').round(4)
smry['Overall'] = smry.mean(axis=1)
print(smry.sort_values('Overall', ascending=False).to_string())

print("\nFinal benchmark comparison:")
print(f"  {'Study':<40} {'C1':>8} {'C2':>8} {'C3':>8}")
print(f"  {'-'*66}")
print(f"  {'Fischer & Krauss (2018) LSTM':<40} "
      f"{'---':>8} {'53.20%':>8} {'---':>8}")
print(f"  {'Lu et al. (2020) CNN-LSTM':<40} "
      f"{'---':>8} {'55.90%':>8} {'---':>8}")
if has_r3:
    for lbl, model, df in [
        ('This study CNN-LSTM Run3',     'CNN-LSTM',      R3),
        ('This study CNN-LSTM-Attn Run4','CNN-LSTM-Attn', RESULTS_DF_R4),
    ]:
        for cond in ['C1', 'C2', 'C3']:
            sub = df[(df['condition'] == cond) &
                     (df['model']     == model)]
            if sub.empty:
                continue
        vals = {c: df[(df['condition'] == c) &
                      (df['model']     == model)]['accuracy'].mean()
                for c in ['C1', 'C2', 'C3']}
        c1s = f"{vals['C1']*100:.2f}%" if 'C1' in vals else '---'
        c2s = f"{vals['C2']*100:.2f}%" if 'C2' in vals else '---'
        c3s = f"{vals['C3']*100:.2f}%" if 'C3' in vals else '---'
        print(f"  {lbl:<40} {c1s:>8} {c2s:>8} {c3s:>8}")


# ── Confusion matrices ────────────────────────────────────────
from matplotlib.colors import LinearSegmentedColormap
cmap_dark = LinearSegmentedColormap.from_list(
    'dark_blue', [DARK, ACCENT])

for condition in CFG['conditions']:
    tickers_this = (CFG['tickers_c3']
                    if condition == 'C3' else CFG['tickers'])
    n   = len(tickers_this)
    fig, axes = plt.subplots(
        1, n, figsize=(3.5*n, 4), facecolor=DARK)
    if n == 1:
        axes = [axes]
    fig.suptitle(
        f'Confusion Matrices — CNN-LSTM-Attn Run 4 | {condition}',
        color=TEXT, fontsize=12, fontweight='bold')

    for idx, ticker in enumerate(tickers_this):
        r = next(
            (x for x in RESULTS_R4
             if x['ticker']    == ticker
             and x['condition'] == condition
             and x['model']     == 'CNN-LSTM-Attn'), None)
        if r is None:
            continue
        cm  = r['conf_matrix']
        ax  = axes[idx]
        ax.set_facecolor(DARK)
        ax.imshow(cm, cmap=cmap_dark)
        ax.set_title(f"{ticker}\nacc={r['accuracy']:.4f}",
                     color=TEXT, fontsize=10)
        ax.set_xticks([0, 1]); ax.set_yticks([0, 1])
        ax.set_xticklabels(['DOWN', 'UP'], fontsize=9)
        ax.set_yticklabels(['DOWN', 'UP'], fontsize=9)
        if idx == 0:
            ax.set_ylabel('Actual', fontsize=9)
        ax.set_xlabel('Predicted', fontsize=9)
        for i in range(2):
            for j in range(2):
                ax.text(j, i, str(cm[i, j]),
                        ha='center', va='center',
                        fontsize=13, fontweight='bold',
                        color=TEXT)

    plt.tight_layout()
    fname = f'R4_confusion_{condition}.png'
    plt.savefig(fname, dpi=150,
                bbox_inches='tight', facecolor=DARK)
    plt.show()
    print(f"Saved: {fname}")


# ── ROC curves: Run 3 CNN-LSTM vs Run 4 CNN-LSTM-Attn ────────
print("\nGenerating ROC comparison figure...")

for condition in CFG['conditions']:
    tickers_this = (CFG['tickers_c3']
                    if condition == 'C3' else CFG['tickers'])
    n   = len(tickers_this)
    fig, axes = plt.subplots(
        1, n, figsize=(4*n, 4), facecolor=DARK)
    if n == 1:
        axes = [axes]
    fig.suptitle(
        f'ROC Curves: CNN-LSTM R3 vs CNN-LSTM-Attn R4 | {condition}',
        color=TEXT, fontsize=11, fontweight='bold')

    for idx, ticker in enumerate(tickers_this):
        ax = axes[idx]
        ax.set_facecolor(DARK)
        ax.plot([0, 1], [0, 1], color=BORDER,
                linestyle='--', linewidth=0.8)

        if has_r3:
            r3_result = next(
                (r for r in pickle.load(
                    open(os.path.join(SAVE_PATH,
                                      'RESULTS_R3.pkl'), 'rb'))
                 if r['ticker']    == ticker
                 and r['condition'] == condition
                 and r['model']     == 'CNN-LSTM'), None)
            if r3_result:
                fpr, tpr, _ = roc_curve(
                    r3_result['y_true'], r3_result['y_prob'])
                ax.plot(fpr, tpr, color=YELLOW, linewidth=1.4,
                        label=f"R3 CNN-LSTM "
                              f"({r3_result['roc_auc']:.3f})")

        r4_result = next(
            (r for r in RESULTS_R4
             if r['ticker']    == ticker
             and r['condition'] == condition
             and r['model']     == 'CNN-LSTM-Attn'), None)
        if r4_result:
            fpr, tpr, _ = roc_curve(
                r4_result['y_true'], r4_result['y_prob'])
            ax.plot(fpr, tpr, color=ACCENT, linewidth=1.4,
                    label=f"R4 Attn "
                          f"({r4_result['roc_auc']:.3f})")

        ax.set_title(ticker, color=TEXT, fontsize=10)
        ax.set_xlabel('FPR', fontsize=8)
        if idx == 0:
            ax.set_ylabel('TPR', fontsize=8)
        ax.legend(fontsize=7, loc='lower right')
        ax.grid(True, alpha=0.3)
        ax.tick_params(labelsize=7)

    plt.tight_layout()
    fname = f'R4_roc_{condition}.png'
    plt.savefig(fname, dpi=150,
                bbox_inches='tight', facecolor=DARK)
    plt.show()
    print(f"Saved: {fname}")

print("\nRun 4 complete.")
```

```python
# ============================================================
# Phase   : CRISP-DM — Evaluation
#
# Outputs:
#   E1 — Diebold-Mariano tests (final models vs all baselines)
#   E2 — Feature importance via Integrated Gradients
#   E3 — Feature importance bar charts
#   E4 — Temporal importance heatmap (which of 20 days matter)
#   E5 — Attention weight visualisation (Run 4 C3)
#   E6 — Random Forest feature importance (C2 comparison)
#   E7 — Final thesis summary tables
#   E8 — Four-run progression chart
#
# PREREQUISITE:
#   PREPARED in memory from data_preparation.py.
#   results_r3.csv, results_r4.csv in /content/.
#   RESULTS_R3.pkl, RESULTS_R4.pkl in /content/.
#   In a fresh session: re-run data_preparation.py first.
#
# XAI METHOD NOTE:
#   The shap library patches TF's gradient registry at import
#   time, corrupting normal model.fit() with LookupError on
#   any op it has not pre-registered (shap_DivNoNan, etc.).
#   This patch is process-level and cannot be undone once
#   applied. This file uses a self-contained Integrated
#   Gradients implementation built on tf.GradientTape, which
#   requires no registry modification and is mathematically
#   equivalent to SHAP GradientExplainer.
#   Reference: Sundararajan et al. (2017) — Axiomatic
#   Attribution for Deep Networks (Integrated Gradients).
#   Equivalence to SHAP: Lundberg & Lee (2017, NeurIPS).
# ============================================================

import os, pickle, random, warnings, time
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from matplotlib.colors import LinearSegmentedColormap
from scipy import stats as scipy_stats

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, regularizers
from tensorflow.keras.optimizers import Adam

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, roc_auc_score, f1_score,
    average_precision_score, confusion_matrix, roc_curve)

# ── Reproducibility ──────────────────────────────────────────
SEED = 42
os.environ['PYTHONHASHSEED']       = str(SEED)
os.environ['TF_DETERMINISTIC_OPS'] = '1'
random.seed(SEED); np.random.seed(SEED); tf.random.set_seed(SEED)

# ── Plot theme ───────────────────────────────────────────────
DARK   = '#0d1117'; BORDER = '#30363d'; TEXT = '#c9d1d9'
MUTED  = '#8b949e'; GRID   = '#21262d'
ACCENT = '#58a6ff'; GREEN  = '#3fb950'; RED    = '#f85149'
YELLOW = '#e3b341'; PURPLE = '#bc8cff'; ORANGE = '#f0883e'
TEAL   = '#39d353'
COLORS = [ACCENT, GREEN, RED, YELLOW, PURPLE]

plt.rcParams.update({
    'figure.facecolor': DARK, 'axes.facecolor':  DARK,
    'axes.edgecolor':  BORDER, 'axes.labelcolor': TEXT,
    'xtick.color': MUTED, 'ytick.color': MUTED,
    'text.color': TEXT,   'grid.color':  GRID,
    'grid.linestyle': '--', 'grid.alpha': 0.5,
    'font.family': 'monospace',
    'legend.facecolor': '#161b22',
    'legend.edgecolor': BORDER, 'figure.dpi': 120,
})

SAVE_PATH = '/content/'

def load_prepared(path=SAVE_PATH):
    with open(os.path.join(path, 'PREPARED.pkl'),  'rb') as f:
        p = pickle.load(f)
    with open(os.path.join(path, 'SCALERS.pkl'),   'rb') as f:
        s = pickle.load(f)
    with open(os.path.join(path, 'PREP_INFO.pkl'), 'rb') as f:
        i = pickle.load(f)
    print(f"Loaded from {path}")
    return p, s, i

try:
    _ = PREPARED
    print("PREPARED in memory.")
except NameError:
    PREPARED, SCALERS, PREP_INFO = load_prepared()

R3 = pd.read_csv(os.path.join(SAVE_PATH, 'results_r3.csv'))
R4 = pd.read_csv(os.path.join(SAVE_PATH, 'results_r4.csv'))

with open(os.path.join(SAVE_PATH, 'RESULTS_R3.pkl'), 'rb') as f:
    RESULTS_R3 = pickle.load(f)
with open(os.path.join(SAVE_PATH, 'RESULTS_R4.pkl'), 'rb') as f:
    RESULTS_R4 = pickle.load(f)

print("All result files loaded.")
print(f"R3: {len(R3)} rows   R4: {len(R4)} rows")

# ── Configuration ─────────────────────────────────────────────
CFG = dict(
    tickers    = ['AAPL', 'AMZN', 'GOOG', 'MSFT', 'TSLA'],
    tickers_c3 = ['AAPL', 'AMZN', 'GOOG', 'TSLA'],
    look_back  = 20,
    learning_rate = 1e-4, lstm_units = 64, cnn_filters = 32,
    cnn_kernel = 3, dropout = 0.3, l2_reg = 1e-4,
    clip_norm = 1.0, batch_size = 32,
    save_path = SAVE_PATH,
)

FEAT_C1 = ['Open', 'High', 'Low', 'Close', 'Volume']
FEAT_C2 = ['Open', 'High', 'Low', 'Close', 'Volume',
           'EMA_20', 'EMA_50', 'RSI', 'RSI_MA',
           'Histogram', 'BB_Basis', 'BB_Upper', 'BB_Lower',
           'MACD', 'Signal', 'VIX']
FEAT_C3 = FEAT_C2 + ['Sentiment']
FEATURE_SETS = {'C1': FEAT_C1, 'C2': FEAT_C2, 'C3': FEAT_C3}

MAJORITY_CLASS = {
    'AAPL': 0.5020, 'AMZN': 0.5139,
    'GOOG': 0.5100, 'MSFT': 0.5179, 'TSLA': 0.5299}

print("Configuration loaded.")


# ── Model builders ────────────────────────────────────────────

class TemporalAttention(layers.Layer):
    """
    Single-head temporal self-attention.
    Reference: Bahdanau et al. (2015); Qin et al. (2017)
    """
    def __init__(self, units=32, **kwargs):
        super().__init__(**kwargs)
        self.units = units
        self.W = layers.Dense(units, use_bias=True,
                              activation='tanh')
        self.v = layers.Dense(1, use_bias=False)

    def call(self, hidden_states):
        score   = self.W(hidden_states)
        score   = self.v(score)
        weights = tf.nn.softmax(score, axis=1)
        context = weights * hidden_states
        return context, weights

    def get_config(self):
        config = super().get_config()
        config.update({'units': self.units})
        return config


def build_cnn_lstm(input_shape):
    reg = regularizers.l2(CFG['l2_reg'])
    inp = layers.Input(shape=input_shape)
    x   = layers.Conv1D(CFG['cnn_filters'], CFG['cnn_kernel'],
                        padding='causal', activation='relu',
                        kernel_regularizer=reg)(inp)
    x   = layers.MaxPooling1D(pool_size=2)(x)
    x   = layers.Dropout(CFG['dropout'])(x)
    x   = layers.LSTM(CFG['lstm_units'],
                      kernel_regularizer=reg,
                      recurrent_regularizer=reg)(x)
    x   = layers.BatchNormalization()(x)
    x   = layers.Dropout(CFG['dropout'])(x)
    x   = layers.Dense(32, activation='relu',
                       kernel_regularizer=reg)(x)
    out = layers.Dense(1, activation='sigmoid')(x)
    m   = keras.Model(inp, out)
    m.compile(optimizer=Adam(CFG['learning_rate'],
                             clipnorm=CFG['clip_norm']),
              loss='binary_crossentropy', metrics=['accuracy'])
    return m


def build_cnn_lstm_attn(input_shape):
    reg = regularizers.l2(CFG['l2_reg'])
    inp = layers.Input(input_shape)
    x   = layers.Conv1D(CFG['cnn_filters'], CFG['cnn_kernel'],
                        padding='causal', activation='relu',
                        kernel_regularizer=reg)(inp)
    x   = layers.MaxPooling1D(pool_size=2)(x)
    x   = layers.Dropout(CFG['dropout'])(x)
    attn_out, attn_weights = TemporalAttention(
        units=32, name='temporal_attention')(x)
    x   = layers.Add()([x, attn_out])
    x   = layers.LayerNormalization()(x)
    x   = layers.LSTM(CFG['lstm_units'],
                      kernel_regularizer=reg,
                      recurrent_regularizer=reg)(x)
    x   = layers.BatchNormalization()(x)
    x   = layers.Dropout(CFG['dropout'])(x)
    x   = layers.Dense(32, activation='relu',
                       kernel_regularizer=reg)(x)
    out = layers.Dense(1, activation='sigmoid')(x)
    m   = keras.Model(inp, out)
    m.compile(optimizer=Adam(CFG['learning_rate'],
                             clipnorm=CFG['clip_norm']),
              loss='binary_crossentropy', metrics=['accuracy'])
    return m


def retrain_model(ticker, condition, model_type, n_epochs):
    """
    Retrain on full pre-test data for n_epochs.
    model_type: 'cnn_lstm' or 'cnn_lstm_attn'
    """
    arrays    = PREPARED[ticker][condition]
    X_all     = np.concatenate([arrays['X_train'],
                                 arrays['X_val']], axis=0)
    y_all     = np.concatenate([arrays['y_train'],
                                 arrays['y_val']], axis=0)
    n_pos     = y_all.sum(); n_neg = len(y_all) - n_pos
    cw        = {0: len(y_all)/(2*n_neg) if n_neg > 0 else 1.0,
                 1: len(y_all)/(2*n_pos) if n_pos > 0 else 1.0}
    inp_shape = (CFG['look_back'], arrays['X_train'].shape[2])
    tf.random.set_seed(SEED); np.random.seed(SEED)
    model = (build_cnn_lstm(inp_shape)
             if model_type == 'cnn_lstm'
             else build_cnn_lstm_attn(inp_shape))
    model.fit(X_all, y_all, epochs=n_epochs,
              batch_size=CFG['batch_size'],
              class_weight=cw, verbose=0)
    return model, X_all, y_all


print("Model builders defined.")


# ── Integrated Gradients implementation ───────────────────────
# Pure TF implementation — no shap library dependency.
# Mathematically equivalent to SHAP GradientExplainer.
# Reference: Sundararajan et al. (2017); Lundberg & Lee (2017).

def integrated_gradients(model, X, baseline=None,
                          n_steps=50, batch_size=32):
    """
    Compute Integrated Gradients attributions.

    For each input sample x, attributions are:
        IG(x) = (x - baseline) * mean(grad(f, alpha))
    where alpha interpolates from baseline to x over n_steps.

    Parameters
    ----------
    model      : fitted Keras model
    X          : input array (n_samples, look_back, n_feat)
    baseline   : reference input (default: zero tensor)
    n_steps    : number of interpolation steps (default 50)
    batch_size : interpolation batch size

    Returns
    -------
    attributions : array (n_samples, look_back, n_feat)
    mean_abs     : mean |attribution| per feature
    """
    X        = X.astype(np.float32)
    if baseline is None:
        baseline = np.zeros_like(X[:1], dtype=np.float32)
    else:
        baseline = baseline.astype(np.float32)

    # Interpolation steps from baseline to input
    alphas      = np.linspace(0, 1, n_steps + 1,
                               dtype=np.float32)
    all_grads   = []

    for i in range(0, len(X), batch_size):
        X_batch = X[i:i + batch_size]
        # Broadcast: (n_steps+1, batch, look_back, n_feat)
        interp  = (baseline
                   + alphas[:, None, None, None]
                   * (X_batch[None] - baseline))
        # Reshape to (n_steps+1 * batch, look_back, n_feat)
        n_s, n_b = interp.shape[0], interp.shape[1]
        interp_flat = interp.reshape(n_s * n_b,
                                     *interp.shape[2:])
        interp_tf   = tf.constant(interp_flat)

        with tf.GradientTape() as tape:
            tape.watch(interp_tf)
            preds = model(interp_tf, training=False)
        grads = tape.gradient(preds, interp_tf).numpy()
        # Reshape back to (n_steps+1, batch, look_back, n_feat)
        grads = grads.reshape(n_s, n_b, *grads.shape[1:])
        # Trapezoidal integration over steps axis
        avg_grads = np.trapz(grads, axis=0) / n_steps
        # Element-wise multiply by (x - baseline)
        delta = X_batch - baseline
        ig    = avg_grads * delta
        all_grads.append(ig)

    attributions = np.concatenate(all_grads, axis=0)
    mean_abs     = np.abs(attributions).mean(axis=(0, 1))
    return attributions, mean_abs


print("Integrated Gradients implementation defined.")
print("No shap library dependency — works with all TF ops.")


# ── E1 — Diebold-Mariano Tests ────────────────────────────────
# Compares best model per condition against all baselines.
# Uses 0-1 loss (misclassification) and Harvey et al. (1997)
# small-sample correction. H0: equal predictive accuracy.
# Reference: Diebold & Mariano (1995, JBES)

def dm_test(y_true, prob1, prob2, h=1):
    """
    Diebold-Mariano test for equal predictive accuracy.
    prob1 = baseline probabilities, prob2 = proposed model.
    Negative DM stat: baseline has lower loss (baseline better).
    Reference: Harvey, Leybourne & Newbold (1997, IJF)
    """
    pred1  = (prob1 >= 0.5).astype(int)
    pred2  = (prob2 >= 0.5).astype(int)
    e1     = (pred1 != y_true).astype(float)
    e2     = (pred2 != y_true).astype(float)
    d      = e1 - e2
    T      = len(d)
    d_bar  = d.mean()
    gamma0 = np.var(d, ddof=1)
    if gamma0 < 1e-12:
        return 0.0, 1.0
    hac    = np.sqrt((T + 1 - 2*h + h*(h-1)/T) / T)
    dm_raw = d_bar / np.sqrt(gamma0 / T)
    dm_hc  = dm_raw * hac
    p_val  = 2 * scipy_stats.t.sf(abs(dm_hc), df=T-1)
    return round(dm_hc, 4), round(p_val, 4)


print("=" * 65)
print("E1 — DIEBOLD-MARIANO TESTS (Final Models)")
print("=" * 65)
print("Best model per condition: C1/C2=CNN-LSTM Run3, "
      "C3=CNN-LSTM-Attn Run4")
print("H0: equal predictive accuracy")
print("Negative DM: baseline has MORE errors (proposed better)")
print()

BEST_MODELS = {
    'C1': ('CNN-LSTM',      RESULTS_R3),
    'C2': ('CNN-LSTM',      RESULTS_R3),
    'C3': ('CNN-LSTM-Attn', RESULTS_R4),
}
BASELINES = ['Logistic Regression', 'Random Forest',
             'XGBoost', 'CNN-only', 'LSTM-only']
DM_RESULTS = []

for condition, (best_model_name, best_results) in \
        BEST_MODELS.items():
    tickers = (CFG['tickers_c3']
               if condition == 'C3' else CFG['tickers'])
    print(f"\n  Condition {condition} — Best: {best_model_name}")
    print(f"  {'Ticker':<6} {'Baseline':<24} {'DM':>8} "
          f"{'p-val':>8} {'Sig':>6} "
          f"{'Best_acc':>10} {'Base_acc':>10}")
    print(f"  {'-'*76}")

    for ticker in tickers:
        best_r = next(
            (r for r in best_results
             if r['ticker']    == ticker
             and r['condition'] == condition
             and r['model']     == best_model_name), None)
        if best_r is None:
            continue
        for baseline in BASELINES:
            base_r = next(
                (r for r in best_results
                 if r['ticker']    == ticker
                 and r['condition'] == condition
                 and r['model']     == baseline), None)
            if base_r is None:
                continue
            n  = min(len(best_r['y_true']),
                     len(base_r['y_true']))
            dm, pv = dm_test(
                best_r['y_true'][:n],
                base_r['y_prob'][:n],
                best_r['y_prob'][:n])
            sig = ('***' if pv < 0.01 else
                   '**'  if pv < 0.05 else
                   '*'   if pv < 0.10 else 'ns')
            print(f"  {ticker:<6} {baseline:<24} {dm:>8.4f} "
                  f"{pv:>8.4f} {sig:>6} "
                  f"{best_r['accuracy']:>10.4f} "
                  f"{base_r['accuracy']:>10.4f}")
            DM_RESULTS.append({
                'condition':      condition,
                'ticker':         ticker,
                'baseline':       baseline,
                'dm_stat':        dm,
                'p_value':        pv,
                'significant_05': pv < 0.05,
                'significant_10': pv < 0.10,
                'best_acc':       best_r['accuracy'],
                'base_acc':       base_r['accuracy'],
            })

DM_DF = pd.DataFrame(DM_RESULTS)
DM_DF.to_csv(os.path.join(SAVE_PATH, 'eval_dm_tests.csv'),
             index=False)
print(f"\n  Total DM tests: {len(DM_DF)}")
print(f"  p < 0.05 (significant): {DM_DF['significant_05'].sum()}")
print(f"  p < 0.10 (trend):       {DM_DF['significant_10'].sum()}")
sig_rows = DM_DF[DM_DF['significant_10']]
if len(sig_rows) > 0:
    print("\n  Significant/trend results:")
    print(sig_rows[['condition', 'ticker', 'baseline',
                    'dm_stat', 'p_value',
                    'best_acc', 'base_acc']].to_string(
                        index=False))


# ── E2 — Feature Importance (Integrated Gradients) ───────────
# Integrated Gradients attributes each feature's contribution
# by integrating gradients along a path from a baseline
# (zero input) to the actual input. This is equivalent to
# SHAP GradientExplainer and satisfies the completeness and
# sensitivity axioms of attribution methods.
# Reference: Sundararajan et al. (2017, ICML)

print("=" * 65)
print("E2 — FEATURE IMPORTANCE (Integrated Gradients)")
print("=" * 65)
print("Note: pure TF implementation — no shap dependency.")
print("Expected time: ~10-15 minutes total.")
print()

IG_RESULTS = {}

# E2a: CNN-LSTM C2 — AAPL (Run 3, optimal epoch = 14)
print("Computing attributions for CNN-LSTM C2 (Run 3) — AAPL...")
model_c2, X_all_c2, _ = retrain_model(
    'AAPL', 'C2', 'cnn_lstm', n_epochs=14)
arrays_aapl_c2  = PREPARED['AAPL']['C2']
attr_c2, ma_c2  = integrated_gradients(
    model_c2, arrays_aapl_c2['X_test'])
IG_RESULTS['AAPL_C2_CNNLSTM'] = {
    'attributions': attr_c2, 'mean_abs': ma_c2,
    'features':     FEAT_C2,
    'X_test':       arrays_aapl_c2['X_test'],
    'y_test':       arrays_aapl_c2['y_test'],
}
print(f"  Done. Attributions shape: {attr_c2.shape}")
with open(os.path.join(SAVE_PATH, 'IG_C2_AAPL.pkl'), 'wb') as f:
    pickle.dump(IG_RESULTS['AAPL_C2_CNNLSTM'], f, protocol=4)
print("  Saved: IG_C2_AAPL.pkl")

# E2b: CNN-LSTM-Attn C3 — all 4 tickers (Run 4)
print("\nComputing attributions for CNN-LSTM-Attn C3 (Run 4)...")
R4_C3_EPOCHS = {'AAPL': 10, 'AMZN': 6, 'GOOG': 6, 'TSLA': 6}
IG_C3 = {}
for ticker in CFG['tickers_c3']:
    ep = R4_C3_EPOCHS.get(ticker, 8)
    print(f"  [{ticker}] Retraining with {ep} epochs...")
    model_c3, X_all_c3, _ = retrain_model(
        ticker, 'C3', 'cnn_lstm_attn', n_epochs=ep)
    arrays_c3        = PREPARED[ticker]['C3']
    attr_c3, ma_c3   = integrated_gradients(
        model_c3, arrays_c3['X_test'])
    IG_C3[ticker] = {
        'attributions': attr_c3, 'mean_abs': ma_c3,
        'features':     FEAT_C3,
        'X_test':       arrays_c3['X_test'],
        'y_test':       arrays_c3['y_test'],
    }
    print(f"  [{ticker}] Done. Attributions shape: {attr_c3.shape}")

IG_RESULTS['C3_all'] = IG_C3
with open(os.path.join(SAVE_PATH, 'IG_C3_all.pkl'), 'wb') as f:
    pickle.dump(IG_C3, f, protocol=4)
print("  Saved: IG_C3_all.pkl")


# ── E3 — Feature Importance Figures ──────────────────────────
print("=" * 65)
print("E3 — FEATURE IMPORTANCE FIGURES")
print("=" * 65)

# E3a: C2 model — AAPL
fig, ax = plt.subplots(figsize=(10, 7), facecolor=DARK)
ax.set_facecolor(DARK)
ig_data     = IG_RESULTS['AAPL_C2_CNNLSTM']
mean_abs    = ig_data['mean_abs']
features    = ig_data['features']
order       = np.argsort(mean_abs)[::-1]
feat_sorted = [features[i] for i in order]
vals_sorted = mean_abs[order]
colors_bar  = [ACCENT if f != 'Volume' else GREEN
               for f in feat_sorted]
ax.barh(range(len(feat_sorted)), vals_sorted,
        color=colors_bar, alpha=0.85)
ax.set_yticks(range(len(feat_sorted)))
ax.set_yticklabels(feat_sorted, fontsize=10)
ax.set_xlabel('Mean |Integrated Gradient|', fontsize=11)
ax.set_title(
    'Feature Importance — CNN-LSTM C2 | AAPL\n'
    'Mean |Integrated Gradient| across test predictions '
    '(aggregated over look_back=20 days)',
    color=TEXT, fontsize=11, fontweight='bold')
ax.axvline(0, color=BORDER, linewidth=0.8)
ax.grid(True, axis='x', alpha=0.3)
ax.tick_params(axis='y', labelsize=9)
plt.tight_layout()
plt.savefig('E3a_ig_importance_C2_AAPL.png', dpi=150,
            bbox_inches='tight', facecolor=DARK)
plt.show()
print("Saved: E3a_ig_importance_C2_AAPL.png")

# E3b: C3 model — all 4 tickers
fig, axes = plt.subplots(
    1, len(CFG['tickers_c3']),
    figsize=(5*len(CFG['tickers_c3']), 8),
    facecolor=DARK)
fig.suptitle(
    'Feature Importance — CNN-LSTM-Attn C3\n'
    'Mean |Integrated Gradient| per feature per ticker '
    '(Sentiment highlighted)',
    color=TEXT, fontsize=12, fontweight='bold')

for idx, ticker in enumerate(CFG['tickers_c3']):
    ax   = axes[idx]
    ax.set_facecolor(DARK)
    ig   = IG_C3[ticker]
    ma   = ig['mean_abs']
    ft   = ig['features']
    ord_ = np.argsort(ma)[::-1]
    fs   = [ft[i] for i in ord_]
    vs   = ma[ord_]
    cols = [RED if f == 'Sentiment'
            else (GREEN if f == 'Volume' else ACCENT)
            for f in fs]
    ax.barh(range(len(fs)), vs, color=cols, alpha=0.85)
    ax.set_yticks(range(len(fs)))
    ax.set_yticklabels(fs if idx == 0 else [], fontsize=8)
    ax.set_title(ticker, color=TEXT, fontsize=11)
    ax.set_xlabel('Mean |IG|', fontsize=8)
    ax.grid(True, axis='x', alpha=0.3)
    sent_rank = fs.index('Sentiment') + 1
    ax.text(0.98, 0.02, f'Sentiment rank: #{sent_rank}',
            transform=ax.transAxes,
            ha='right', va='bottom', fontsize=8, color=RED)

plt.tight_layout()
plt.savefig('E3b_ig_importance_C3.png', dpi=150,
            bbox_inches='tight', facecolor=DARK)
plt.show()
print("Saved: E3b_ig_importance_C3.png")

print("\nSentiment attribution summary (C3 model):")
print(f"{'Ticker':<8} {'Sent_IG':>12} {'Rank':>6} "
      f"{'Interpretation':>25}")
print("-" * 55)
for ticker in CFG['tickers_c3']:
    ma        = IG_C3[ticker]['mean_abs']
    ft        = IG_C3[ticker]['features']
    si        = ft.index('Sentiment')
    rank      = int(np.argsort(ma)[::-1].tolist().index(si)) + 1
    ig_sent   = ma[si]
    attr_raw  = IG_C3[ticker]['attributions'][:, :, si]
    mean_sign = np.mean(attr_raw)
    direction = "Contrarian (neg)" if mean_sign < 0 \
                else "Confirming (pos)"
    print(f"{ticker:<8} {ig_sent:>12.6f} {rank:>6} "
          f"{direction:>25}")


# ── E4 — Temporal Attribution Heatmap ────────────────────────
# Rows = features, Columns = timesteps (t-19 to t-0).
# Colour = mean |attribution|. Reveals whether the model
# relies on recent momentum or mid-window regime signals.
print("=" * 65)
print("E4 — TEMPORAL ATTRIBUTION HEATMAP")
print("=" * 65)

ig_data      = IG_RESULTS['AAPL_C2_CNNLSTM']
attr_3d      = ig_data['attributions']
features     = ig_data['features']
mean_abs_2d  = np.abs(attr_3d).mean(axis=0)  # (20, n_feat)

fig, ax = plt.subplots(figsize=(14, 7), facecolor=DARK)
ax.set_facecolor(DARK)
im = ax.imshow(mean_abs_2d.T, aspect='auto',
               cmap='hot', interpolation='nearest')
ax.set_xticks(range(0, 20, 2))
ax.set_xticklabels([f't-{19-i}' for i in range(0, 20, 2)],
                   fontsize=8, rotation=45)
ax.set_yticks(range(len(features)))
ax.set_yticklabels(features, fontsize=9)
ax.set_xlabel(
    'Timestep in look_back window (t-19=oldest, t-0=most recent)',
    fontsize=10)
ax.set_title(
    'Temporal Attribution Heatmap — CNN-LSTM C2 | AAPL\n'
    'Mean |Integrated Gradient| per (timestep, feature). '
    'Brighter = more influential.',
    color=TEXT, fontsize=11, fontweight='bold')
plt.colorbar(im, ax=ax, fraction=0.02, pad=0.02)
plt.tight_layout()
plt.savefig('E4_temporal_heatmap.png', dpi=150,
            bbox_inches='tight', facecolor=DARK)
plt.show()
print("Saved: E4_temporal_heatmap.png")

ts_importance = mean_abs_2d.mean(axis=1)
most_imp_t    = np.argmax(ts_importance)
print(f"\nMost influential timestep: "
      f"t-{19 - most_imp_t} "
      f"(day {most_imp_t+1} of 20-day window)")
print("Timestep importance (mean |IG| over features):")
for t in range(20):
    bar = '#' * int(ts_importance[t] / ts_importance.max() * 30)
    print(f"  t-{19-t:2d}  {ts_importance[t]:.6f}  {bar}")


# ── E5 — Attention Weight Visualisation ──────────────────────
# Attention weights complement Integrated Gradients — two
# independent methods for attributing temporal importance.
# Agreement between them strengthens the interpretability
# argument.
print("=" * 65)
print("E5 — ATTENTION WEIGHT VISUALISATION")
print("=" * 65)

print("Retraining AAPL C3 model for attention extraction...")
ep_aapl_c3           = R4_C3_EPOCHS['AAPL']
model_attn_aapl, _, _ = retrain_model(
    'AAPL', 'C3', 'cnn_lstm_attn', n_epochs=ep_aapl_c3)

attn_layer     = model_attn_aapl.get_layer('temporal_attention')
attn_extractor = keras.Model(
    inputs  = model_attn_aapl.input,
    outputs = attn_layer.output[1])

X_test_c3        = PREPARED['AAPL']['C3']['X_test']
attn_weights_raw = attn_extractor.predict(X_test_c3, verbose=0)
attn_weights     = attn_weights_raw.squeeze(-1)
mean_attn        = attn_weights.mean(axis=0)

print(f"Attention weights shape: {attn_weights.shape}")
print(f"Timesteps after MaxPool: {len(mean_attn)}")

fig, axes = plt.subplots(1, 2, figsize=(14, 5), facecolor=DARK)
fig.suptitle(
    'Attention Weights — CNN-LSTM-Attn C3 | AAPL\n'
    'Left: mean weight per timestep  '
    'Right: attention vs IG comparison',
    color=TEXT, fontsize=11, fontweight='bold')

ax1 = axes[0]
ax1.set_facecolor(DARK)
t_ax = range(len(mean_attn))
ax1.bar(t_ax, mean_attn, color=ACCENT, alpha=0.85)
ax1.axhline(1/len(mean_attn), color=RED, linestyle='--',
            linewidth=1.0, label='Uniform (no preference)')
ax1.set_xlabel('Timestep (after MaxPool)', fontsize=10)
ax1.set_ylabel('Mean attention weight', fontsize=10)
ax1.set_title('Average attention across test predictions',
              color=TEXT, fontsize=10)
ax1.legend(fontsize=8)
ax1.grid(True, axis='y', alpha=0.3)

ax2 = axes[1]
ax2.set_facecolor(DARK)
ts_ig  = mean_abs_2d.mean(axis=1)
n_attn = len(mean_attn)
ig_ds  = np.array([
    ts_ig[int(i*20/n_attn):int((i+1)*20/n_attn)].mean()
    for i in range(n_attn)])
attn_norm = mean_attn / mean_attn.sum()
ig_norm   = ig_ds / ig_ds.sum() if ig_ds.sum() > 0 else ig_ds
ax2.plot(t_ax, attn_norm, color=ACCENT, linewidth=2,
         marker='o', markersize=5,
         label='Attention (normalised)')
ax2.plot(t_ax, ig_norm, color=YELLOW, linewidth=2,
         marker='s', markersize=5, linestyle='--',
         label='IG (normalised)')
ax2.set_xlabel('Timestep (pooled)', fontsize=10)
ax2.set_ylabel('Normalised importance', fontsize=10)
ax2.set_title('Attention vs Integrated Gradients — do they agree?',
              color=TEXT, fontsize=10)
ax2.legend(fontsize=8)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('E5_attention_weights.png', dpi=150,
            bbox_inches='tight', facecolor=DARK)
plt.show()
print("Saved: E5_attention_weights.png")

corr = np.corrcoef(attn_norm, ig_norm)[0, 1]
print(f"\nCorrelation between attention and IG "
      f"temporal profiles: r={corr:.4f}")
if abs(corr) > 0.5:
    print("  Attention and IG agree on which timesteps "
          "matter — mutually reinforcing.")
else:
    print("  Attention and IG focus on different timesteps "
          "— they capture different aspects of the sequence.")


# ── E6 — Random Forest Feature Importance (C2 comparison) ────
# RF built-in impurity importance compared against CNN-LSTM IG.
# If both methods rank the same features highly the finding
# is model-agnostic (Strobl et al., 2008).
print("=" * 65)
print("E6 — RANDOM FOREST FEATURE IMPORTANCE (AAPL C2)")
print("=" * 65)

arrays_c2 = PREPARED['AAPL']['C2']
X_tr_f    = arrays_c2['X_train'].reshape(
    len(arrays_c2['X_train']), -1)
X_val_f   = arrays_c2['X_val'].reshape(
    len(arrays_c2['X_val']), -1)
X_te_f    = arrays_c2['X_test'].reshape(
    len(arrays_c2['X_test']), -1)
X_all_f   = np.concatenate([X_tr_f, X_val_f])
y_all_rf  = np.concatenate([
    arrays_c2['y_train'], arrays_c2['y_val']])

rf = RandomForestClassifier(
    n_estimators=200, max_depth=8,
    class_weight='balanced', random_state=SEED, n_jobs=-1)
rf.fit(X_all_f, y_all_rf)
print(f"RF test accuracy: "
      f"{accuracy_score(arrays_c2['y_test'], rf.predict(X_te_f)):.4f}")

# Aggregate RF impurity importance per original feature
n_feat   = len(FEAT_C2)
lb       = CFG['look_back']
rf_imp   = rf.feature_importances_
rf_feat  = np.zeros(n_feat)
for fi in range(n_feat):
    idx_cols    = [t * n_feat + fi for t in range(lb)]
    rf_feat[fi] = rf_imp[idx_cols].mean()

fig, axes = plt.subplots(1, 2, figsize=(14, 7), facecolor=DARK)
fig.suptitle(
    'Feature Importance: Random Forest vs CNN-LSTM | AAPL C2\n'
    'Agreement between models = robust feature importance',
    color=TEXT, fontsize=11, fontweight='bold')

for ax_idx, (vals, model_nm, col) in enumerate([
    (IG_RESULTS['AAPL_C2_CNNLSTM']['mean_abs'],
     'CNN-LSTM (Integrated Gradients)', ACCENT),
    (rf_feat, 'Random Forest (Impurity)', GREEN),
]):
    ax    = axes[ax_idx]
    ax.set_facecolor(DARK)
    order = np.argsort(vals)[::-1]
    fs    = [FEAT_C2[i] for i in order]
    vs    = vals[order]
    ax.barh(range(len(fs)), vs, color=col, alpha=0.85)
    ax.set_yticks(range(len(fs)))
    ax.set_yticklabels(fs if ax_idx == 0 else [], fontsize=9)
    ax.set_title(model_nm, color=TEXT, fontsize=10)
    ax.set_xlabel('Mean importance', fontsize=9)
    ax.grid(True, axis='x', alpha=0.3)

plt.tight_layout()
plt.savefig('E6_rf_vs_dl_importance.png', dpi=150,
            bbox_inches='tight', facecolor=DARK)
plt.show()
print("Saved: E6_rf_vs_dl_importance.png")

dl_ranks = np.argsort(
    np.argsort(IG_RESULTS['AAPL_C2_CNNLSTM']['mean_abs']))
rf_ranks = np.argsort(np.argsort(rf_feat))
spearman = scipy_stats.spearmanr(dl_ranks, rf_ranks)
print(f"\nSpearman rank correlation (RF vs CNN-LSTM): "
      f"r={spearman.statistic:.4f}  "
      f"p={spearman.pvalue:.4f}")
if spearman.statistic > 0.5:
    print("  Both models agree on which features matter.")
elif spearman.statistic > 0.2:
    print("  Moderate agreement on feature importance.")
else:
    print("  Models prioritise different features — "
          "DL extracts patterns classical ML cannot.")


# ── E7 — Final Thesis Tables ──────────────────────────────────
print("=" * 65)
print("E7 — FINAL THESIS TABLES")
print("=" * 65)

print("\nTHESIS TABLE 1: Directional Accuracy — "
      "Best Run per Condition")
print("(C1/C2: Run 3 CNN-LSTM  |  C3: Run 4 CNN-LSTM-Attn)")
print()

def build_table(results_df, model_name, conditions, tickers):
    rows = []
    for cond in conditions:
        t_list = (CFG['tickers_c3']
                  if cond == 'C3' else tickers)
        for ticker in t_list:
            row = results_df[
                (results_df['ticker']    == ticker) &
                (results_df['condition'] == cond) &
                (results_df['model']     == model_name)]
            if len(row) == 0:
                continue
            rows.append({
                'Condition': cond, 'Ticker': ticker,
                'Accuracy':  row['accuracy'].values[0],
                'ROC-AUC':   row['roc_auc'].values[0],
                'F1':        row['f1_weighted'].values[0],
                'Model':     model_name,
            })
    return pd.DataFrame(rows)

t1_c1c2 = build_table(R3, 'CNN-LSTM', ['C1', 'C2'],
                       CFG['tickers'])
t1_c3   = build_table(R4, 'CNN-LSTM-Attn', ['C3'],
                       CFG['tickers_c3'])
t1      = pd.concat([t1_c1c2, t1_c3], ignore_index=True)
print(t1.to_string(index=False))

print("\n\nTHESIS TABLE 2: Mean Accuracy — All Models, All Conditions")
print("(CNN-LSTM = Run3  |  CNN-LSTM-Attn = Run4)")
print()
r3_means = R3.groupby(['condition', 'model'])\
    ['accuracy'].mean().unstack('condition').round(4)
r4_means = R4.groupby(['condition', 'model'])\
    ['accuracy'].mean().unstack('condition').round(4)
t2 = r3_means.copy()
if 'CNN-LSTM-Attn' in r4_means.index:
    t2.loc['CNN-LSTM-Attn'] = r4_means.loc['CNN-LSTM-Attn']
t2['Overall'] = t2.mean(axis=1)
t2 = t2.sort_values('Overall', ascending=False)
print(t2.to_string())

print("\n\nTHESIS TABLE 3: Benchmark Comparison")
print()
benchmarks = {
    'Fischer & Krauss (2018) LSTM':  {'C2': 53.20},
    'Lu et al. (2020) CNN-LSTM':     {'C2': 55.90},
    'Nelson et al. (2017) LSTM':     {'C2': 55.90},
    'Sezer & Ozbayoglu (2018) CNN':  {'C2': 55.00},
}
print(f"{'Study':<42} {'C1':>8} {'C2':>8} {'C3':>8}")
print("-" * 70)
for study, vals in benchmarks.items():
    c1s = f"{vals.get('C1', '---')}"
    c2s = f"{vals.get('C2', '---')}"
    c3s = f"{vals.get('C3', '---')}"
    print(f"  {study:<40} {c1s:>8} {c2s:>8} {c3s:>8}")

for model_nm, df, label in [
    ('CNN-LSTM',      R3, 'This study CNN-LSTM (Run3)'),
    ('CNN-LSTM-Attn', R4, 'This study CNN-LSTM-Attn (Run4)'),
]:
    vals_str = ""
    for cond in ['C1', 'C2', 'C3']:
        sub = df[(df['condition'] == cond) &
                 (df['model']     == model_nm)]
        vals_str += f"{'---':>8}" if sub.empty \
                    else f"{sub['accuracy'].mean()*100:>7.2f}%"
    print(f"  {label:<40} {vals_str}")

t1.to_csv(os.path.join(SAVE_PATH, 'eval_table1.csv'),
          index=False)
t2.to_csv(os.path.join(SAVE_PATH, 'eval_table2.csv'))
print("\nTables saved: eval_table1.csv, eval_table2.csv")


# ── E8 — Four-run progression figure ─────────────────────────
print("=" * 65)
print("E8 — FOUR-RUN PROGRESSION FIGURE")
print("=" * 65)

run_labels = ['Run 0\nBaseline', 'Run 1\nlr+size',
              'Run 2\nWalk-fwd', 'Run 3\nCovidX',
              'Run 4\nAttention']
c2_means   = [0.4805, 0.5152, 0.4987, 0.5125, 0.4918]
c3_means   = [0.5065, 0.5141, 0.5065, 0.5098, 0.5335]

fig, ax = plt.subplots(figsize=(12, 6), facecolor=DARK)
ax.set_facecolor(DARK)
x = np.arange(5)
ax.plot(x, c2_means, color=ACCENT, linewidth=2,
        marker='o', markersize=8, label='CNN-LSTM C2')
ax.plot(x, c3_means, color=GREEN, linewidth=2,
        marker='s', markersize=8, label='CNN-LSTM-Attn C3')

best_c2_idx = np.argmax(c2_means)
best_c3_idx = np.argmax(c3_means)
ax.plot(x[best_c2_idx], c2_means[best_c2_idx],
        'o', color=ACCENT, markersize=14, alpha=0.4)
ax.plot(x[best_c3_idx], c3_means[best_c3_idx],
        's', color=GREEN,  markersize=14, alpha=0.4)
ax.axhline(0.50,  color=RED,    linestyle='--',
           linewidth=1.0, alpha=0.7, label='50% random')
ax.axhline(0.532, color=PURPLE, linestyle=':',
           linewidth=1.0, alpha=0.7,
           label='Fischer & Krauss (2018) 53.2%')

for i, (c2, c3) in enumerate(zip(c2_means, c3_means)):
    ax.annotate(f'{c2*100:.1f}%', (i, c2),
                textcoords='offset points', xytext=(0, 10),
                ha='center', fontsize=8, color=ACCENT)
    ax.annotate(f'{c3*100:.1f}%', (i, c3),
                textcoords='offset points', xytext=(0, -16),
                ha='center', fontsize=8, color=GREEN)

ax.set_xticks(x)
ax.set_xticklabels(run_labels, fontsize=10)
ax.set_ylabel('Mean Directional Accuracy', fontsize=11)
ax.set_ylim(0.46, 0.57)
ax.set_title(
    'Four-Run Progression — CNN-LSTM Accuracy Improvement\n'
    'Each run applies one targeted fix to the training protocol',
    color=TEXT, fontsize=12, fontweight='bold')
ax.legend(fontsize=9, loc='upper left')
ax.grid(True, axis='y', alpha=0.3)

changes = ['', '+lr fix\n+size', '+walk-fwd\n+clip',
           '+excl\nCOVID', '+attention']
for i, ch in enumerate(changes):
    if ch:
        ax.text(i, 0.463, ch, ha='center',
                fontsize=7, color=MUTED)

plt.tight_layout()
plt.savefig('E8_progression.png', dpi=150,
            bbox_inches='tight', facecolor=DARK)
plt.show()
print("Saved: E8_progression.png")


print()
print("=" * 65)
print("EVALUATION PHASE COMPLETE")
print("=" * 65)
print("""
Files saved:
  eval_dm_tests.csv, eval_table1.csv, eval_table2.csv
  IG_C2_AAPL.pkl, IG_C3_all.pkl

Figures saved:
  E3a_ig_importance_C2_AAPL.png
  E3b_ig_importance_C3.png
  E4_temporal_heatmap.png
  E5_attention_weights.png
  E6_rf_vs_dl_importance.png
  E8_progression.png
""")
```
